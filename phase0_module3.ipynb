```python
import numpy as np
p1 = np.array([1, 1])
p2 = np.array([4, 5])

diff = p2 - p1

# Algebraic method
squared_diff = diff ** 2
summed_dimensions = np.sum(squared_diff)
distance_alg = np.sqrt(summed_dimensions)

# Dot product method
distance_dot = np.sqrt(np.dot(diff, diff))

print(distance_alg)
print(distance_dot)
```

```python
### ML INTUITION
# input matrix
A = np.array([[2, 3], [1, 4]])
print(A.shape)

# weight matrix
x = np.array([2, 1])
print(x.shape)

# the forward pass
bill = np.matmul(A, x)
print(bill)
```
