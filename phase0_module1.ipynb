# Module 1

```python
import numpy as np
positions = np.array([0,3,-2])
print(f"Starting positions: {positions}")

# broadcasting
positions += 2
print(f"New positions: {positions}")
```

```python
# Scaling
positions = np.array([0, 3, -2])
positions *= 3
print(positions)
```

# Module 2

```python
x = np.array([1,2,3])
y = np.array([4,5,6])
z_add = x+y
z_dot = np.dot(x, y)
print(z_add)
print(z_dot)
import matplotlib.pyplot as plt
plt.plot(x, marker='o')
plt.title("Vector plot")
plt.show()
```
