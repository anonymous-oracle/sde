### Phase 2 | Module 1: Sequences, Series & Convergence

To train a Machine Learning model, the computer takes "steps" to find the lowest error. To understand those steps, we need to understand Sequences and Series.

#### 1. What is a Sequence?
A **Sequence** is just an ordered list of numbers following a specific mathematical rule. 
* **Arithmetic Sequence:** You add or subtract a constant number each time (like walking up stairs). 
  * Example: 2, 5, 8, 11... (Common difference $d = 3$).
  * The formula for any term $a_n$ is: $$a_n = a_1 + (n-1)d$$
* **Geometric Sequence:** You multiply or divide by a constant number each time (like compound interest).
  * Example: 100, 50, 25, 12.5... (Common ratio $r = 0.5$).
  * The formula for any term $a_n$ is: $$a_n = a_1 \cdot r^{n-1}$$

#### 2. What is a Series? (The Sum)
If a Sequence is a *list* of numbers, a **Series** is what you get when you *add them all together*. 
What happens if you add an infinite geometric sequence together? Does it grow to infinity?
Not always! If your common ratio $r$ is a fraction between -1 and 1 (meaning the steps get smaller and smaller), the infinite sum actually **converges** to a perfectly finite, strict boundary. 


The formula for the sum of an infinite converging geometric series is beautifully simple:
$$S = \frac{a_1}{1 - r}$$

#### 3. The ML Tie-in: Learning Rate Decay
When we train a Neural Network, we use a "Learning Rate" to determine how big of a step the AI takes toward the correct answer. 
* If the steps are too big, the AI overshoots the answer.
* If the steps are too small, it takes a billion years to train.
To fix this, Staff Scientists use **Exponential Learning Rate Decay**. We start with a large step (e.g., 0.1), and after every epoch (training round), we multiply it by a ratio $r$ (e.g., 0.9). The AI's step sizes form a **Geometric Sequence** that converges safely to the perfect answer!

Let's build and plot both an Arithmetic Sequence and an Exponential Decay (Geometric) Sequence in pure Python to see how they differ visually.

**The Task:**
Open a new Jupyter cell and follow these instructions:

1. **Imports:** `import numpy as np` and `import matplotlib.pyplot as plt`.
2. **The Time Steps (n):** Create an array of epochs from 1 to 15: `epochs = np.arange(1, 16)`.

```python
import numpy as np
import matplotlib.pyplot as plt

epochs = np.arange(1, 16)
a1_arith = 10
d = -0.5
arithmetic_seq = a1_arith + (epochs - 1) * d
r = 0.8
geometric_seq = a1_arith * r**(epochs - 1)

plt.figure(figsize=(8, 6))
plt.plot(epochs, arithmetic_seq, marker="o", color="blue", label="Arithmetic (d = -0.5)")
plt.plot(epochs, geometric_seq, marker='x', color='red', label='Geometric (r = 0.8)')

plt.axhline(0, color='black', linestyle='--')
plt.title("Sequence Behaviors")
plt.xlabel("Epoch (n)")
plt.ylabel("Value")
plt.legend()
plt.grid(True)
plt.show()
```
