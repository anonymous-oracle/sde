### Phase 2 | Module 2: Complex Numbers & The Complex Plane

We are taught in early algebra that you cannot take the square root of a negative number. $\sqrt{-9}$ breaks a standard calculator because no real number multiplied by itself equals a negative. 

But what if we just *invented* a number to solve this? 

#### 1. The Imaginary Unit ($i$)
Mathematicians defined a new concept called the imaginary unit, **$i$**. 
It has one strict rule: **$i = \sqrt{-1}$**.
Therefore, if you square it, you get a negative number: **$i^2 = -1$**.

This allows us to solve impossible roots. $\sqrt{-9}$ simply becomes $\sqrt{9 \times -1}$, which simplifies to **$3i$**.

#### 2. The Standard Form ($z = a + bi$)
A **Complex Number** (usually written as $z$) is a hybrid. It has a **Real Part** ($a$) and an **Imaginary Part** ($b$). 
Example: $z = 4 + 3i$. 
You can add them just like standard algebra by combining like terms:
$$(4 + 3i) + (2 + 5i) = (4+2) + (3i+5i) = 6 + 8i$$

If you multiply them, you use the standard FOIL method, remembering that $i^2 = -1$:
$$(2 + 3i)(1 + 2i) = 2 + 4i + 3i + 6i^2$$
Because $6i^2 = 6(-1) = -6$, the answer simplifies to $(2 - 6) + 7i = \mathbf{-4 + 7i}$.

#### 3. The Complex Plane & Magnitude

Because a complex number has two distinct parts ($a$ and $b$), we can plot it as a 2D coordinate! 
* The **x-axis** becomes the Real line.
* The **y-axis** becomes the Imaginary line.
The number $z = 4 + 3i$ is just a point plotted at $(4, 3)$. 

Because it forms a right triangle on the grid, we can find its **Magnitude** (its distance from the center $(0,0)$) using the exact same Pythagorean theorem you mastered in Phase 1!
The magnitude is written as $|z|$:
$$|z| = \sqrt{a^2 + b^2}$$
For $4 + 3i$, the magnitude is $\sqrt{4^2 + 3^2} = \sqrt{16 + 9} = \sqrt{25} = \mathbf{5}$.

#### 4. The ML Tie-in: RoPE (Rotary Positional Embeddings)
Why do Deep Learning engineers care about imaginary numbers? 
If you plot a point on the complex plane and multiply it by $i$, the point perfectly **rotates 90 degrees** around the center. Complex numbers are the ultimate mathematical engine for 2D rotation. 

Modern Large Language Models (like Llama-3 and Mistral) do not use Sine/Cosine waves directly for word order anymore. They use **RoPE (Rotary Positional Embeddings)**. They treat every word's embedding vector as a coordinate on the Complex Plane and apply complex multiplication to physically *rotate* the word by a specific angle based on its position in the sentence. This allows the AI to understand relative distances between words flawlessly!

### Your Next Action: Main Track Guided Implementation

Let's use Python to plot Complex Numbers on a 2D grid. Python natively supports complex numbers using `j` instead of `i` (engineers use `j` because `i` is used for electrical current). 

**The Task:**
Open a new Jupyter cell and follow these instructions:

1. **Imports:** `import matplotlib.pyplot as plt`.
2. **Define the Numbers:** Create two complex numbers in Python:
   * `z1 = 4 + 3j`
   * `z2 = -2 + 5j`
3. **The Canvas:** `plt.figure(figsize=(6, 6))`.
4. **Plotting:** * You can extract the real part using `.real` and the imaginary part using `.imag`.
   * Plot `z1` as a red dot: `plt.plot(z1.real, z1.imag, 'ro', label='z1 = 4 + 3i')`.
   * Plot `z2` as a blue dot: `plt.plot(z2.real, z2.imag, 'bo', label='z2 = -2 + 5i')`.
5. **Draw the Vectors:** Draw lines from the origin $(0,0)$ to your points so they look like vectors.
   * `plt.plot([0, z1.real], [0, z1.imag], 'r--')`
   * `plt.plot([0, z2.real], [0, z2.imag], 'b--')`
6. **Formatting:**
   * Set axes limits: `plt.xlim(-6, 6)` and `plt.ylim(-6, 6)`.
   * Add zero-lines: `plt.axhline(0, color='black')` and `plt.axvline(0, color='black')`.
   * Add `plt.xlabel("Real Axis")`, `plt.ylabel("Imaginary Axis")`, `plt.grid(True)`, `plt.legend()`, and `plt.show()`.
