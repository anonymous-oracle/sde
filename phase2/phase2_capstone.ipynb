### **Mastery Consolidation Side Quest: Advanced Trigonometry & Identities**

#### **1. Rigorous Theory: The Reciprocal Functions**
We defined $\sin(\theta)$ and $\cos(\theta)$ on the unit circle. You correctly identified that $\tan(\theta) = \frac{\sin(\theta)}{\cos(\theta)}$. 
Calculus requires three more functions, which are simply the reciprocals of the first three:

* **Secant ($\sec(\theta)$):** $\frac{1}{\cos(\theta)}$
* **Cosecant ($\csc(\theta)$):** $\frac{1}{\sin(\theta)}$
* **Cotangent ($\cot(\theta)$):** $\frac{1}{\tan(\theta)} = \frac{\cos(\theta)}{\sin(\theta)}$

#### **2. Rigorous Theory: The Pythagorean Identities**
You already know the master identity:
$$\sin^2(\theta) + \cos^2(\theta) = 1$$
From this single truth, we can derive the others.
* **Derivation 1:** Divide the entire master equation by $\cos^2(\theta)$:
    $$\frac{\sin^2(\theta)}{\cos^2(\theta)} + \frac{\cos^2(\theta)}{\cos^2(\theta)} = \frac{1}{\cos^2(\theta)}$$
    This yields the second identity: **$\tan^2(\theta) + 1 = \sec^2(\theta)$**.
* **Derivation 2:** Divide the entire master equation by $\sin^2(\theta)$:
    This yields the third identity: **$1 + \cot^2(\theta) = \csc^2(\theta)$**.

These three identities are the hammers you will use to smash apart complex integrals in Phase 3. 

#### **3. Rigorous Theory: Solving Trigonometric Equations**
Because trig functions represent cycles, they don't have just one answer.
If I ask you to solve $\sin(\theta) = 0.5$, you might say $\theta = \frac{\pi}{6}$ ($30^\circ$). But because the Ferris wheel keeps turning, you hit a height of $0.5$ again on the other side of the circle at $\frac{5\pi}{6}$ ($150^\circ$). And if you go around the circle another full time ($+2\pi$), you hit it again at $\frac{13\pi}{6}$. 

Unless you are given a strict domain (e.g., $0 \le \theta < 2\pi$), a trigonometric equation has *infinite* solutions, written algebraically as $\theta = \text{base angle} + 2\pi k$ (where $k$ is any integer).
