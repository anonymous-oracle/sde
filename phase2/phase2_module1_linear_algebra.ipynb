### Phase 2 | Module 1: Linear Algebra Foundations (Vectors & Matrices)

Deep Learning is, at its absolute core, just giant spreadsheets of numbers being multiplied together. Linear Algebra is the mathematical language we use to manage those spreadsheets.

#### 1. What is a Vector? (The Arrow vs. The List)
* **To a Physicist:** A Vector is an arrow in space. It has a length (magnitude) and a direction.
* **To a Computer Scientist:** A Vector is just a Python List (or an Array) of numbers. That's it. 
For example, a vector `v = [3, 4]` means "start at 0, walk 3 steps right on the x-axis, and 4 steps up on the y-axis."

In ML, we use vectors to represent features. If we are predicting house prices, our input vector might be `x = [2000_sqft, 3_bedrooms, 2_bathrooms]`.

#### 2. The Dot Product (The Engine of Neural Networks)
If you only learn one thing in Linear Algebra, it must be the **Dot Product**.
The Dot Product takes two vectors of the exact same size, multiplies their matching numbers together, and adds everything up into a single number.

**The Math:**
If Vector $A = [a_1, a_2, a_3]$ and Vector $B = [b_1, b_2, b_3]$, then:
$$A \cdot B = (a_1 \cdot b_1) + (a_2 \cdot b_2) + (a_3 \cdot b_3)$$

**The ML Tie-in:** Every single "Neuron" in a neural network is just doing a Dot Product!
A neuron receives an input vector `x` (e.g., pixel values), and it holds a weight vector `w` (how much it cares about each pixel). It calculates the dot product `w · x`. If the resulting number is high, the neuron "fires". If it's low, it stays quiet.
