### Phase 2 | Module 1: Linear Algebra (Vectors & The Dot Product)

Neural networks do not understand text, images, or audio. They only understand lists of numbers. Linear Algebra is the mathematical language of lists of numbers. 

#### 1. What is a Vector? (The Supermarket Intuition)
Imagine you are writing a grocery list. 
* Apples: 3
* Bananas: 5
* Milk: 1

In Python, you might write this as a list: `[3, 5, 1]`. 
In Linear Algebra, this is called a **Vector**. A vector is just an array of numbers representing a single point or direction in space. Because this vector has 3 numbers, we say it lives in a "3-Dimensional Space". 

In modern NLP (like the `word2vec` models we will build), a word like "king" isn't just an integer ID anymore. It becomes a massive vector of 300 floating-point numbers (a 300-Dimensional space) where each number represents a hidden semantic meaning!

#### 2. The Dot Product (Measuring Similarity)
If I give you a vector for "king" and a vector for "queen", how does the computer know they mean similar things? It uses the **Dot Product**.

The Dot Product is the ultimate mathematical tool for comparing two vectors to see how much they "overlap" or point in the same direction.
**The Math:** You multiply the matching positions of the two vectors together, and then sum them all up.

Let Vector $A = [1, 2, 3]$ and Vector $B = [4, 5, 6]$.
$$A \cdot B = (1 \times 4) + (2 \times 5) + (3 \times 6)$$
$$A \cdot B = 4 + 10 + 18 = 32$$

#### 3. The Geometric Proof (Cosine Similarity)
There is a second, magical formula for the Dot Product. It equals the length (magnitude) of Vector $A$ multiplied by the length of Vector $B$, multiplied by the **Cosine** of the angle between them ($\theta$).
$$A \cdot B = \|A\| \|B\| \cos(\theta)$$

Why does this matter? Because if we isolate the Cosine, we get **Cosine Similarity**:
$$\cos(\theta) = \frac{A \cdot B}{\|A\| \|B\|}$$

If two word vectors point in the exact same direction (they mean the same thing), the angle between them is 0. And as you learned in your Unit Circle side quest, the Cosine of 0 is **$1.0$**. If they are completely unrelated (90 degrees apart), the Cosine is **$0$**. This is exactly how vector databases and LLMs search for relevant text!
