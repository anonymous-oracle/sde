## **Phase 2: Precalculus Bridge**
### **Module 3: Trigonometry (The Geometry of Cycles)**

#### **1. The Ferris Wheel Intuition**
Up until now, our mathematical universe has been mostly linear or polynomial—things either shoot off to infinity or crash to negative infinity. But the real world is full of cycles: pendulums swing, seasons change, and sound travels in waves.

Imagine you are on a Ferris wheel. As the wheel turns, your height off the ground goes up, hits a maximum, comes back down, hits a minimum, and repeats indefinitely. You are going in a circle, but if we plot just your *height* over time, it draws a perfect, repeating wave. **Trigonometry** is the mathematical bridge between circles and waves.

#### **2. Rigorous Theory: The Unit Circle**
Forget the old "SOH CAH TOA" triangle stuff for a moment. To build ML systems, we define trigonometry using the **Unit Circle**.
Imagine our 2D Cartesian plane. We draw a circle perfectly centered at the Origin $(0,0)$ with a radius of exactly 1. 

If you start at $(1,0)$ and walk counter-clockwise around the edge of this circle, you create an angle (let's call it $\theta$) with the center. At any point on that circle, your exact $(x, y)$ coordinates are defined by two functions:
* **Cosine ($\cos(\theta)$):** Your $x$-coordinate (your horizontal position).
* **Sine ($\sin(\theta)$):** Your $y$-coordinate (your vertical position).

Because the radius is 1, and these coordinates form a right triangle with the origin, we can plug them directly into the Pythagorean Theorem from Phase 1 to get the most famous trigonometric identity in history:
$$\sin^2(\theta) + \cos^2(\theta) = 1$$

#### **3. Radians: The True Measure of Angles**
Degrees (360° in a circle) are completely arbitrary—invented by ancient Babylonians because they liked the number 60. Mathematics does not care about Babylonians. 
We measure angles in **Radians**. A radian is the physical distance you have walked around the edge of the circle. Since the circumference of a circle is $2\pi r$, and our Unit Circle has a radius of $1$, a full $360^\circ$ rotation is exactly **$2\pi$ radians**. A half-circle ($180^\circ$) is $\pi$ radians.

#### **4. Machine Learning Tie-In: Positional Encoding**
Why do Neural Networks care about Ferris wheels? If you use ChatGPT, how does it know the order of the words in your sentence? Words are fed in all at once. To give the model a sense of "position," the engineers add a mathematical timestamp to every word using interlaced Sine and Cosine waves of different frequencies. This is called **Positional Encoding** (specifically the RoPE variant we will build from scratch in Phase 11).
