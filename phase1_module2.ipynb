```python
import numpy as np

def degrees_to_radians(degrees):
    # The core mathematical truth: 180 degrees is exactly equal to Pi radians.
    # Therefore, 1 degree = Pi / 180 radians.
    radians = degrees * (np.pi / 180)
    return radians

# Let's test the 4 main checkpoints of a circle
angles_in_degrees = [0, 90, 180, 270, 360]

print("--- The Unit Circle Angles ---")
for angle in angles_in_degrees:
    rad = degrees_to_radians(angle)
    # We print the result as a clean fraction of Pi for readability
    print(f"{angle:3} degrees = {rad/np.pi:.1f} * Pi radians")
```
