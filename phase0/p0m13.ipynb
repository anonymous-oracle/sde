### **Module 13, Node 1: Arithmetic Progressions (AP)**

**Textbook Reference Block (Exhaustive Parallel Citation Protocol v14.2)**
* **Selina Concise Mathematics:** Class 10, Chapter 10: Arithmetic Progression -> Page Range: Standard Edition Span.
* **M.L. Aggarwal Understanding ICSE:** Class 10, Arithmetic Progression -> Page Range: Standard Edition Span.

**1. Supermarket Intuition**
Imagine building a staircase. The first step is 15 cm high. Every step after that adds exactly 15 cm of elevation. If you want to know the height of the 100th step, you *could* add 15 to itself 100 times. But in computer science, iterating a loop 100 times to find a deterministic value is a waste of compute. An **Arithmetic Progression (AP)** is a sequence of numbers where the jump between any two consecutive terms is strictly constant. By mapping the mathematical formula for that jump, you can instantly predict the value of the 1,000,000th term, or the sum of all terms, in a single CPU cycle.

**2. Bounded Rigorous Derivation**
Let us establish the strict parameters of an AP:
* **First Term ($a$):** The starting number of the sequence.
* **Common Difference ($d$):** The fixed delta added to each term to get the next. It is always calculated as $\text{Term}_2 - \text{Term}_1$. (It can be positive, negative, or zero).
* **The General Term ($t_n$):** The value of the term at the $n^{\text{th}}$ position.
  To get to the $2^{\text{nd}}$ term, we add $d$ exactly $1$ time. To get to the $3^{\text{rd}}$ term, we add $d$ exactly $2$ times. Therefore, to get to the $n^{\text{th}}$ term, we add $d$ exactly $(n-1)$ times.
  $$t_n = a + (n - 1)d$$
* **The Sum of $n$ Terms ($S_n$):** The total accumulated value if you add the first $n$ terms together.
  $$S_n = \frac{n}{2} [2a + (n - 1)d]$$
  *(Alternative formula if the last term $l$ is known: $S_n = \frac{n}{2} [a + l]$)*

***

### **Example Illustrations (Override Rule v14.3)**

* **Basic (Finding a Term):** Find the $15^{\text{th}}$ term of the AP: $3, 7, 11, 15, \dots$
    * *Step 1:* Identify components. $a = 3$. The common difference $d = 7 - 3 = 4$. We need $n = 15$.
    * *Step 2:* Apply formula $t_n = a + (n-1)d$.
    * *Step 3:* $t_{15} = 3 + (15 - 1)4 = 3 + (14 \times 4) = 3 + 56 = 59$.
    * *Answer:* The $15^{\text{th}}$ term is $59$.

* **Medium (Simultaneous Equations Bypass):** The $4^{\text{th}}$ term of an AP is $14$, and the $8^{\text{th}}$ term is $30$. Find the $1^{\text{st}}$ term ($a$) and the common difference ($d$).
    * *Step 1:* Frame two linear equations using $t_n$.
        $t_4 = a + 3d = 14$
        $t_8 = a + 7d = 30$
    * *Step 2:* Apply Elimination (Module 11). Subtract the first equation from the second.
        $(a + 7d) - (a + 3d) = 30 - 14 \implies 4d = 16 \implies d = 4$.
    * *Step 3:* Substitute $d$ back to find $a$: $a + 3(4) = 14 \implies a = 2$.
    * *Answer:* $a = 2$, $d = 4$.

* **Difficult (Position of a Value):** Which term of the AP: $21, 18, 15, \dots$ is strictly $-81$?
    * *Step 1:* Identify parameters. $a = 21$. $d = 18 - 21 = -3$. The target value is $t_n = -81$.
    * *Step 2:* Set up the equation: $-81 = 21 + (n - 1)(-3)$.
    * *Step 3:* Solve for $n$ (which must be a strictly positive integer).
        $-102 = -3(n - 1) \implies 34 = n - 1 \implies n = 35$.
    * *Answer:* It is the $35^{\text{th}}$ term.

* **Extension (Sum & Quadratic intersection):** How many terms of the AP $9, 17, 25, \dots$ must be taken so that their sum is exactly $636$?
    * *Step 1:* $a = 9$, $d = 17 - 9 = 8$, $S_n = 636$. Find $n$.
    * *Step 2:* Apply sum formula: $636 = \frac{n}{2} [2(9) + (n - 1)8]$.
    * *Step 3:* Simplify into a quadratic equation:
        $1272 = n [18 + 8n - 8]$
        $1272 = n [10 + 8n] \implies 8n^2 + 10n - 1272 = 0$
        Divide by 2: $4n^2 + 5n - 636 = 0$.
    * *Step 4:* Factorize or use the Quadratic Formula. 
        Factors of $4 \times -636 = -2544$ that add to $5$ are $53$ and $-48$.
        $4n^2 - 48n + 53n - 636 = 0 \implies 4n(n - 12) + 53(n - 12) = 0$.
        Roots: $n = 12$ and $n = -53/4$.
    * *Step 5:* Filter. A position index $n$ cannot be negative or fractional. Reject $-53/4$.
    * *Answer:* Exactly $12$ terms must be taken.

### **Module 13, Node 2: Geometric Progressions (GP)**

**Textbook Reference Block (Exhaustive Parallel Citation Protocol v14.2)**
* **Selina Concise Mathematics:** Class 10, Chapter 11: Geometric Progression -> Page Range: Standard Edition Span.
* **M.L. Aggarwal Understanding ICSE:** Class 10, Geometric Progression -> Page Range: Standard Edition Span.

**1. Supermarket Intuition**
Arithmetic Progressions add a constant number. They are linear. **Geometric Progressions (GP)** multiply by a constant number. They are exponential. If a virus infects $3$ people, and each of those infects $3$ more, the sequence is $1, 3, 9, 27, 81$. This is the exact underlying mathematics of the Compound Interest formulas you learned in Module 08, population growth models, and the scaling laws of neural networks. In a GP, numbers do not just grow; they explode.

**2. Bounded Rigorous Derivation**
Let us establish the strict parameters of a GP:
* **First Term ($a$):** The starting number.
* **Common Ratio ($r$):** The fixed multiplier. It is always calculated strictly by division: $\frac{\text{Term}_2}{\text{Term}_1}$. 
* **The General Term ($t_n$):** The value at the $n^{\text{th}}$ position.
  To get to the $2^{\text{nd}}$ term, we multiply by $r$ once. For the $3^{\text{rd}}$, we multiply by $r$ twice ($r^2$). Therefore, for the $n^{\text{th}}$ term:
  $$t_n = a \cdot r^{n-1}$$
* **The Sum of $n$ Terms ($S_n$):** Because exponential growth behaves differently whether you are multiplying by a number greater than $1$ (growth) or less than $1$ (decay), the formula forks into two mathematically identical but structurally different branches to avoid negative denominators:
  * If $r > 1$: $$S_n = \frac{a(r^n - 1)}{r - 1}$$
  * If $r < 1$: $$S_n = \frac{a(1 - r^n)}{1 - r}$$

***

### **Example Illustrations (Override Rule v14.3)**

* **Basic (Finding a Term):** Find the $7^{\text{th}}$ term of the GP: $2, 6, 18, 54, \dots$
    * *Step 1:* Identify components. $a = 2$. The common ratio $r = \frac{6}{2} = 3$. We need $n = 7$.
    * *Step 2:* Apply formula $t_n = a \cdot r^{n-1}$.
    * *Step 3:* $t_7 = 2 \cdot (3)^{7-1} = 2 \cdot (3)^6$.
    * *Step 4:* Compute: $3^6 = 729$. Therefore, $2 \times 729 = 1458$.
    * *Answer:* The $7^{\text{th}}$ term is $1458$.

* **Medium (Simultaneous GP Equations):** The $3^{\text{rd}}$ term of a GP is $12$ and the $6^{\text{th}}$ term is $96$. Find $a$ and $r$.
    * *Step 1:* Frame equations using $t_n$.
        $t_3 = a \cdot r^2 = 12$
        $t_6 = a \cdot r^5 = 96$
    * *Step 2:* Instead of subtracting (like in AP), we strictly **divide** the equations to eliminate $a$.
        $$\frac{a \cdot r^5}{a \cdot r^2} = \frac{96}{12}$$
        $$r^3 = 8$$
    * *Step 3:* Solve for $r$: $r = \sqrt[3]{8} \implies r = 2$.
    * *Step 4:* Substitute $r$ back to find $a$: $a(2)^2 = 12 \implies 4a = 12 \implies a = 3$.
    * *Answer:* $a = 3$, $r = 2$.

* **Difficult (Position of a Value with alternating signs):** Which term of the GP $3, -6, 12, -24, \dots$ is $192$?
    * *Step 1:* $a = 3$. $r = \frac{-6}{3} = -2$. (A negative ratio means signs will oscillate).
    * *Step 2:* Set up equation: $192 = 3 \cdot (-2)^{n-1}$.
    * *Step 3:* Divide by $3$: $64 = (-2)^{n-1}$.
    * *Step 4:* Express $64$ as a base of $-2$. Since $64$ is positive and the base is negative, the power must be even. $(-2)^6 = 64$.
    * *Step 5:* Equate exponents: $6 = n - 1 \implies n = 7$.
    * *Answer:* It is the $7^{\text{th}}$ term.

* **Extension (Sum of a Finite Sequence):** Find the sum of the first $8$ terms of the GP: $3, 6, 12, \dots$
    * *Step 1:* $a = 3$, $r = \frac{6}{3} = 2$. We need $S_8$.
    * *Step 2:* Since $r > 1$, use the first sum formula: $S_n = \frac{a(r^n - 1)}{r - 1}$.
    * *Step 3:* $S_8 = \frac{3(2^8 - 1)}{2 - 1}$.
    * *Step 4:* Compute: $2^8 = 256$. 
        $S_8 = \frac{3(256 - 1)}{1} = 3(255) = 765$.
    * *Answer:* The sum is $765$.

If its present value is Rs $50,000$, mathematically frame the GP and find its exact value at the end of $3$ years. *(Hint: A $10\%$ decrease means the common multiplier $r$ is $0.9$.
