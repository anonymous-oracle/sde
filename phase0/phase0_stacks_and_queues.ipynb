### Programming Side Quest: Stacks & Queues

#### 1. The Stack (LIFO)
Imagine a stack of heavy plates at a buffet. When a clean plate is added, it goes on the **top**. When a customer takes a plate, they take it from the **top**. The last plate put on the stack is the first one taken off. 
This is called **LIFO** (Last-In, First-Out).
* **Push:** Adding to the top.
* **Pop:** Removing from the top.



**The ML Tie-in:** Remember that `StackOverflow` error we talked about during the Recursion Side Quest? When a Python function calls itself, the computer literally uses a Stack to remember where it was. It "pushes" the current function onto the call stack, goes deeper, and when it finishes, it "pops" back up to where it left off. If it pushes too many times without popping, the memory overflows!

#### 2. The Queue (FIFO)
Imagine the checkout line at the supermarket. The first person to get in line is the first person to get served and leave. 
This is called **FIFO** (First-In, First-Out).
* **Enqueue (Push):** Adding to the back of the line.
* **Dequeue (Pop):** Removing from the front of the line.



**The ML Tie-in: TDNN Audio Buffers**
In speech recognition, a Time-Delay Neural Network (TDNN) processes live streaming audio. As you speak into the microphone, audio frames arrive sequentially. The network places these frames into a Queue. It processes the oldest frame first (Dequeue), and as new milliseconds of audio arrive, they go to the back of the line (Enqueue). 

#### 3. Your Next Action: Code & Mastery

If we use a standard Python list for a Queue, removing the first item (`list.pop(0)`) is an incredibly slow `O(N)` operation because the computer has to shift every single other item forward by one space. In ML, we need `O(1)` speed. So, we use a specialized tool called a Double-Ended Queue (`deque`).

Because this is a Programming Side Quest, I am providing the full commented implementation. 

**Part 1: The Code**
Copy this into a Jupyter cell, run it, and observe how the order of items changes based on the structure.

```python
from collections import deque

print("--- TESTING THE STACK (LIFO) ---")
# A standard Python list works perfectly as a Stack because appending/popping from the END is very fast.
my_stack = []
my_stack.append("Audio_Frame_1") # Push
my_stack.append("Audio_Frame_2")
my_stack.append("Audio_Frame_3")

# We pop from the end (Top of the stack)
print(f"Popped from Stack: {my_stack.pop()}") 
print(f"Remaining in Stack: {my_stack}\n")

print("--- TESTING THE QUEUE (FIFO) ---")
# We use 'deque' for Queues so that popping from the FRONT is lightning fast.
my_queue = deque()
my_queue.append("Audio_Frame_1") # Enqueue
my_queue.append("Audio_Frame_2")
my_queue.append("Audio_Frame_3")

# We pop from the left (Front of the line)
print(f"Popped from Queue: {my_queue.popleft()}")
print(f"Remaining in Queue: {list(my_queue)}")
```
