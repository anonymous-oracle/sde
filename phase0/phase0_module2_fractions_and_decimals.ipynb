### Phase 0 | Module 2: The Mechanics of Fractions and Decimals

We know how to add and multiply whole numbers. But how do we formally manipulate the Rational numbers ($\mathbb{Q}$)?

#### 1. The Definition of a Fraction
A fraction $\frac{a}{b}$ is not just two numbers stacked on top of each other. Mathematically, it represents the solution to the equation:
$$b \cdot x = a$$
(where $b \neq 0$). The number $x$ is what we call $\frac{a}{b}$.

#### 2. Theorem: Why do we need a Common Denominator to Add?
In primary school, you are told you cannot add $\frac{1}{2} + \frac{1}{3}$ straight across to get $\frac{2}{5}$. You must find a common denominator. Why? 
Because fractions are governed by the **Distributive Property**.

**Rigorous Proof:**
Let's add $\frac{a}{c} + \frac{b}{c}$. 
By definition, dividing by $c$ is the same as multiplying by the inverse $\frac{1}{c}$.
So, we can write this as:
$$a\left(\frac{1}{c}\right) + b\left(\frac{1}{c}\right)$$
Because both terms share the factor $\left(\frac{1}{c}\right)$, we can "pull it out" using the Distributive Property:
$$(a + b)\left(\frac{1}{c}\right)$$
Which is exactly defined as:
$$\frac{a + b}{c}$$
This proves that you can *only* add the top numbers (numerators) if the bottom numbers (denominators) are exactly the same, allowing you to factor them out! $\blacksquare$

To add $\frac{a}{b} + \frac{c}{d}$, we must force the denominators to match by multiplying each fraction by $1$ (which changes the look, but not the value).
We multiply the first by $\frac{d}{d}$ and the second by $\frac{b}{b}$:
$$\frac{ad}{bd} + \frac{bc}{bd} = \frac{ad + bc}{bd}$$

#### 3. Decimals as Infinite Series
A decimal is simply a fraction where the denominator is a power of 10 ($10^1, 10^2, 10^3...$). 
The number $0.375$ is rigorously defined as:
$$\frac{3}{10} + \frac{7}{100} + \frac{5}{1000}$$

But what about repeating decimals like $0.333...$? 
This is an **Infinite Geometric Series**. 
It is defined as $\frac{3}{10} + \frac{3}{100} + \frac{3}{1000} + ...$
The starting term is $a_1 = \frac{3}{10}$, and the common ratio (the step size) is $r = \frac{1}{10}$. 
Using our convergence formula $S = \frac{a_1}{1 - r}$:
$$S = \frac{\frac{3}{10}}{1 - \frac{1}{10}} = \frac{\frac{3}{10}}{\frac{9}{10}} = \frac{3}{9} = \frac{1}{3}$$
This is the rigorous proof that $0.333...$ is exactly equal to $\frac{1}{3}$!
