## Intro to calculus and sympy

```python
import sympy as sp

x = sp.Symbol("x")
y = x**2
derivative_y = sp.diff(y, x)
print(derivative_y)
```

```python
import sympy as sp

# 1. We define our symbolic variables.
# 'w' is our model weight, 'x' is our input feature, 'y' is our true target label.
w, x, y = sp.symbols('w x y')

# 2. We define our prediction (y_hat). A simple linear model: weight * input
y_hat = w * x

# 3. We define our Mean Squared Error (MSE) loss function for a single point.
# We multiply by 0.5 as a mathematical convenience so when the power '2' drops down, they cancel out!
loss = 0.5 * (y_hat - y)**2

# 4. We use SymPy's 'diff' function to calculate the derivative of the loss with respect to the weight 'w'.
# This automatically applies the chain rule behind the scenes, parsing the nested functions.
gradient = sp.diff(loss, w)

# 5. Print the result to see the magical gradient that powers neural network learning.
# Notice how the output matches our manual pen-and-paper derivation!
print("The gradient of the MSE loss with respect to w is:", gradient)
```

```python
import sympy as sp

# 1. Define our two weight variables
w1, w2 = sp.symbols('w1 w2')

# 2. Define our multivariate cost function J
J = 3 * (w1**2) * w2 + 5 * (w2**3) + 2 * w1

# 3. Calculate the partial derivative with respect to w1
# SymPy automatically treats w2 as a constant here.
grad_w1 = sp.diff(J, w1)

# 4. Calculate the partial derivative with respect to w2
# SymPy automatically treats w1 as a constant here.
grad_w2 = sp.diff(J, w2)

# 5. Print the results
print("Partial derivative wrt w1:", grad_w1)
print("Partial derivative wrt w2:", grad_w2)
```
