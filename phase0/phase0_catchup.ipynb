## Sets and Summation

```python
# 1. SETS IN PYTHON
set_A = {1, 2, 3}
set_B = {3, 4, 5}

# Python natively supports mathematical Set operations!
set_union = set_A.union(set_B)
set_intersection = set_A.intersection(set_B)

print(f"Union: {set_union}")
print(f"Intersection: {set_intersection}")

# 2. SUMMATION NOTATION IN PYTHON
# Let's compute the mathematical sum: Sigma (from i=1 to 4) of (i * 2)
total_sum = 0
for i in range(1, 5):  # Python range stops BEFORE the last number, so we use 5 to stop at 4.
    total_sum += (i * 2)

print(f"The result of the Summation is: {total_sum}")
```

## Recursion (using the factorial example) and tree intuition

```python
def calculate_factorial(n):
    # 1. THE BASE CASE: This stops the infinite loop.
    if n == 1 or n == 0:
        return 1
    
    # 2. THE RECURSIVE STEP: The function calls ITSELF with n-1.
    else:
        return n * calculate_factorial(n - 1)

# Let's test it!
number = 5
result = calculate_factorial(number)
print(f"The factorial of {number} is {result}")
```
