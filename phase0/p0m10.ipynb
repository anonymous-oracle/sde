**1. Supermarket Intuition**
If you need to calculate $105 \times 105$, you could execute standard long multiplication. It works, but it takes time. What if you knew a universal pattern that let you solve it instantly? In computer science, doing things "the long way" when a mathematical pattern exists wastes CPU cycles. An **Identity** is an algebraic equation that is strictly true for *all* possible values of its variables. It is a pre-computed computational shortcut. Instead of multiplying polynomials term-by-term every single time, we memorize the structural outcome and jump straight to the answer.

**2. Bounded Rigorous Derivation**
We build these identities strictly using the fundamental multiplication of algebraic expressions we unlocked in Module 09. By the Distributive Law, $(a+b)(c+d) = a(c+d) + b(c+d) = ac + ad + bc + bd$.

* **Identity 1 (Square of a Binomial Sum):**
  $$(a + b)^2 = (a + b)(a + b) = a^2 + ab + ba + b^2 = a^2 + 2ab + b^2$$
* **Identity 2 (Square of a Binomial Difference):**
  $$(a - b)^2 = (a - b)(a - b) = a^2 - ab - ba + b^2 = a^2 - 2ab + b^2$$
* **Identity 3 (Difference of Two Squares):**
  This is the exact inverse of what you used to "rationalize the denominator" when you questioned my hint in Module 02.
  $$(a + b)(a - b) = a^2 - ab + ba - b^2 = a^2 - b^2$$
* **Identity 4 (Standard Expansion):**
  $$(x + a)(x + b) = x^2 + bx + ax + ab = x^2 + (a + b)x + ab$$
* **Identity 5 (Square of a Trinomial - Class 9):**
  $$(a + b + c)^2 = a^2 + b^2 + c^2 + 2ab + 2bc + 2ca$$

**1. Supermarket Intuition**
Expansion is taking a neat, packaged box and unpacking everything onto the floor. **Factorisation** is looking at the mess on the floor and figuring out how to pack it perfectly back into the original box. In machine learning and computer science, data often comes to you "expanded" as a massive matrix of numbers. Algorithms like Principal Component Analysis (PCA) essentially factorize that data to find its core multipliers—compressing the file size and reducing noise without losing the actual information. 

**2. Bounded Rigorous Derivation**
Factorisation is the strict algebraic inverse of expansion. We break expressions down into a product of simpler terms (factors). 

* **Type 1: Taking out the Highest Common Factor (HCF)**
    This is the reverse of the Distributive Law we unlocked in Module 09. You scan all terms for the largest shared integer and the lowest common power of variables.
    $$ab + ac = a(b + c)$$

* **Type 2: Grouping**
    When an expression has an even number of terms (usually four) but no single HCF for all of them, we group them into pairs and factorize the pairs.
    $$ax + ay + bx + by = a(x + y) + b(x + y) = (a + b)(x + y)$$

* **Type 3: Difference of Two Squares**
    Reversing Identity 3 from Node 1. Both terms must be perfect squares separated strictly by a minus sign.
    $$a^2 - b^2 = (a + b)(a - b)$$

* **Type 4: Splitting the Middle Term (Quadratic Trinomials)**
    To factorize an expression of the form $ax^2 + bx + c$:
    1.  Find the product of the first and last coefficients ($a \times c$).
    2.  Find two numbers that multiply to give $ac$ AND add up to give the middle coefficient $b$.
    3.  Rewrite the middle term $bx$ using those two numbers, expanding the trinomial into four terms.
    4.  Factorize by Grouping (Type 2).

**1. Supermarket Intuition**
In Module 04, you learned how to divide integers and find factors. But how do you divide a complex algebraic polynomial like $3x^3 - 4x^2 + 5x - 7$ by another polynomial like $(x - 2)$? You *could* use algebraic long division, but that is computationally expensive and error-prone. 
In computer science, particularly in cryptography and error-correcting algorithms (like the CRC checks that verify downloaded files are not corrupted), systems rely entirely on calculating the *remainder* of polynomial division. The **Remainder Theorem** is a mathematical exploit: it allows you to find the exact remainder of a division instantly, completely bypassing the actual long division process.

**2. Bounded Rigorous Derivation**
Let $f(x)$ be a polynomial expression.

* **Function Notation:** If $f(x) = x^2 + 3x - 4$, then $f(2)$ strictly means replacing every $x$ with $2$ and evaluating it. $f(2) = (2)^2 + 3(2) - 4 = 6$.
* **The Remainder Theorem:** If a polynomial $f(x)$ is divided by a linear expression $(x - a)$, the exact remainder is $f(a)$.
    * *Proof Logic:* By the basic division algorithm, $\text{Dividend} = \text{Divisor} \times \text{Quotient} + \text{Remainder}$.
    * $$f(x) = (x - a) \times q(x) + r$$
    * If we substitute $x = a$, the divisor becomes $(a - a) = 0$. The quotient is annihilated.
    * $$f(a) = (0) \times q(a) + r \implies f(a) = r$$
    * *Constraint Note:* If the divisor is $(x + a)$, you substitute $x = -a$ to annihilate it. If the divisor is $(ax - b)$, you substitute $x = \frac{b}{a}$.
* **The Factor Theorem (The Logical Corollary):**
    If a number divides another perfectly, the remainder is strictly $0$. Therefore, $(x - a)$ is a factor of $f(x)$ *if and only if* $f(a) = 0$.
