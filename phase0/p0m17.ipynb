If two chords of a circle intersect, either internally or externally when produced, the product of the lengths of the segments of one chord is equal to the product of the lengths of the segments of the other chord.

$$AP \times PB = CP \times PD$$

#### Theorem 29: Pythagoras' Theorem

In a right-angled triangle, the square of the length of the hypotenuse is exactly equal to the sum of the squares of the lengths of the other two sides.


$$a^2 + b^2 = c^2$$

**Proof:**
Let $\triangle ABC$ have $\angle ABC = 90^\circ$. Drop a perpendicular altitude $BD \perp AC$ from the right-angle vertex to the hypotenuse.
Isolate and compare the sub-triangle $\triangle ADB$ with the overarching large triangle $\triangle ABC$:

* $\angle ADB = \angle ABC = 90^\circ$ (By construction and definition).
* $\angle A = \angle A$ (Common shared angle).

By AAA Similarity (Theorem 26), $\triangle ADB \sim \triangle ABC$.
Express their proportional side weights:


$$\frac{AD}{AB} = \frac{AB}{AC} \implies AB^2 = AC \times AD \quad \text{— (Equation 1)}$$

Now isolate and compare the adjacent sub-triangle $\triangle潛BDC$ with the large triangle $\triangle ABC$:

* $\angle BDC = \angle ABC = 90^\circ$ (By construction).
* $\angle C = \angle C$ (Common shared angle).

By AAA Similarity (Theorem 26), $\triangle BDC \sim \triangle ABC$.
Express their proportional side weights:


$$\frac{DC}{BC} = \frac{BC}{AC} \implies BC^2 = AC \times DC \quad \text{— (Equation 2)}$$

Execute an algebraic summation of Equation 1 and Equation 2:


$$AB^2 + BC^2 = (AC \times AD) + (AC \times DC)$$


Factor out the common scalar value $AC$:


$$AB^2 + BC^2 = AC(AD + DC)$$


From the continuous geometric layout of the hypotenuse line segment, $AD + DC = AC$. Substitute this back into the expression:


$$AB^2 + BC^2 = AC(AC) \implies AB^2 + BC^2 = AC^2$$

#### Theorem 30: The Cartesian Distance Formula

Let $P(x_1, y_1)$ and $Q(x_2, y_2)$ be two distinct points on a Cartesian plane. Construct a right-angled triangle by drawing a line through $P$ parallel to the $x$-axis and a line through $Q$ parallel to the $y$-axis, intersecting at point $R(x_2, y_1)$.

* The horizontal length $PR = |x_2 - x_1|$
* The vertical length $RQ = |y_2 - y_1|$

Since $\angle PRQ = 90^\circ$ by construction, apply Theorem 29 (Pythagoras' Theorem) to $\triangle PRQ$:

$$PQ^2 = PR^2 + RQ^2$$

Substituting the coordinate delta lengths:

$$PQ^2 = (x_2 - x_1)^2 + (y_2 - y_1)^2$$

Extracting the principal square root yields the distance equation:

$$PQ = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

#### Theorem 31: The Section Formula (Internal Division)

Let a line segment join point $A(x_1, y_1)$ and point $B(x_2, y_2)$. Let point $P(x, y)$ lie internally on the segment $AB$ such that it divides the line in the precise scalar ratio $m_1 : m_2$, meaning:

$$\frac{AP}{PB} = \frac{m_1}{m_2}$$

**Proof:**
Drop perpendicular lines from points $A$, $P$, and $B$ to the horizontal $x$-axis, meeting it at $A', P',$ and $B'$ respectively. Draw horizontal lines from $A$ and $P$ parallel to the $x$-axis to form two right-angled triangles with the vertical lines.

* The base of the first triangle is $x - x_1$
* The base of the second triangle is $x_2 - x$

Because the vertical drop lines are strictly parallel to one another, the intercept matching mechanics of AAA Similarity (Theorem 26) lock the triangles into matching ratios:

$$\frac{x - x_1}{x_2 - x} = \frac{m_1}{m_2}$$

Cross-multiply the scalar ratio values:

$$m_2(x - x_1) = m_1(x_2 - x)$$

$$m_2 x - m_2 x_1 = m_1 x_2 - m_1 x$$

Isolate the target variable $x$ on the left side:

$$m_1 x + m_2 x = m_1 x_2 + m_2 x_1$$

$$x(m_1 + m_2) = m_1 x_2 + m_2 x_1 \implies x = \frac{m_1 x_2 + m_2 x_1}{m_1 + m_2}$$

Executing an identical projection sequence parallel to the vertical axis for the $y$-coordinates yields:

$$y = \frac{m_1 y_2 + m_2 y_1}{m_1 + m_2}$$

$$\therefore P(x, y) = \left(\frac{m_1 x_2 + m_2 x_1}{m_1 + m_2}, \frac{m_1 y_2 + m_2 y_1}{m_1 + m_2}\right)$$

*Status: [UNLOCKED]*

#### Corollary 31.1: The Midpoint Coordinate Formula

Let point $P(x,y)$ bisect the line segment $AB$ evenly. The structural ratio simplifies to $1 : 1$, meaning $m_1 = m_2 = 1$. Substituting these values directly into Theorem 31 yields:

$$P(x, y) = \left(\frac{1 \cdot x_2 + 1 \cdot x_1}{1 + 1}, \frac{1 \cdot y_2 + 1 \cdot y_1}{1 + 1}\right) = \left(\frac{x_1 + x_2}{2}, \frac{y_1 + y_2}{2}\right)$$
