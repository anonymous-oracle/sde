**Lexical Geometry Postulates (D-Nodes):**
1.  **D-Node 8: Angle Classifications & Perpendicularity**
    * *Acute Angle:* $0^\circ < \theta < 90^\circ$.
    * *Right Angle:* $\theta = 90^\circ$.
    * *Obtuse Angle:* $90^\circ < \theta < 180^\circ$.
    * *Perpendicular ($\perp$):* Two intersecting lines that form a Right Angle.
    *Status: [UNLOCKED]*
2.  **D-Node 9: Quadrilateral**
    A closed, two-dimensional plane figure bounded by exactly four straight line segments intersecting at four vertices. 
    *Status: [UNLOCKED]*
3.  **D-Node 10: Parallelogram**
    A quadrilateral wherein both pairs of opposite sides are strictly parallel.
    *Status: [UNLOCKED]*

**Supermarket Intuition:**
Imagine a rigid frame built from four metal rods, hinged at the corners. If you lock the opposing rods so they are always pointing in the exact same direction (parallel), the physical mechanics of the hinges force the opposite rods to be identical in length, and the opposite hinges to open to the exact same angle. If you string two bungee cords connecting the opposite corners (diagonals), they will always slice each other exactly in half, no matter how much you skew the frame.

**Rigorous Derivations (Bottom-Up Traversal):**

Let $ABCD$ be a parallelogram defined by D-Node 10, meaning $AB \parallel CD$ and $AD \parallel BC$. Draw the diagonal $AC$.

1.  **Theorem 12: Opposite Sides and Angles of a Parallelogram are Equal**
    * In $\triangle ABC$ and $\triangle CDA$:
    * $\angle BAC = \angle DCA$ (Alternate interior angles, since $AB \parallel CD$).
    * $\angle BCA = \angle DAC$ (Alternate interior angles, since $AD \parallel BC$).
    * $AC = CA$ (Common side).
    * By ASA Congruence, $\triangle ABC \cong \triangle CDA$.
    * By CPCTC, $AB = CD$ and $BC = DA$ (Opposite sides are equal).
    * By CPCTC, $\angle B = \angle D$.
    * Similarly, drawing diagonal $BD$ proves $\angle A = \angle C$.
    *Status: [UNLOCKED]*

2.  **Theorem 13: Diagonals of a Parallelogram Bisect Each Other**
    * Draw both diagonals $AC$ and $BD$, intersecting at point $O$.
    * In $\triangle AOB$ and $\triangle COD$:
    * $AB = CD$ (Proven in Theorem 12).
    * $\angle OAB = \angle OCD$ (Alternate interior, $AB \parallel CD$).
    * $\angle OBA = \angle ODC$ (Alternate interior, $AB \parallel CD$).
    * By ASA Congruence, $\triangle AOB \cong \triangle COD$.
    * By CPCTC, $OA = OC$ and $OB = OD$. Therefore, the diagonals bisect each other.
    *Status: [UNLOCKED]*

**Intuition-Building Illustrative Examples (IBIE):**

**IBIE 1: The One-Pair Parallel Trap**
* **[The Problem]:** In quadrilateral $PQRS$, $PQ \parallel RS$ and $PQ = RS$. Is $PQRS$ a parallelogram?
* **[The Heuristic Trap]:** Assuming only one pair of parallel sides is insufficient because D-Node 10 requires *both* pairs.
* **[The Failure State]:** Stating "No, we don't know if $PS \parallel QR$."
* **[The Resolution Heuristic]:** We can prove the second pair mathematically. One pair of equal and parallel sides guarantees a parallelogram.
* **[Solution Pathway]:** Draw diagonal $PR$. In $\triangle PQR$ and $\triangle RSP$: $PQ = RS$ (Given), $\angle QPR = \angle SRP$ (Alternate interior), $PR = RP$ (Common). By SAS, $\triangle PQR \cong \triangle RSP$. By CPCTC, $\angle PRQ = \angle SPR$. Since these are equal alternate interior angles, $PS \parallel QR$. Both pairs are parallel, thus it is a parallelogram.

**IBIE 2: The Equal Diagonal Assumption (Visual Trap)**
* **[The Problem]:** A parallelogram $ABCD$ has diagonals $AC$ and $BD$ intersecting at $O$. $AC = 10$. Can we conclude $BD = 10$?
* **[The Heuristic Trap]:** Assuming symmetry in a parallelogram extends to diagonal length.
* **[The Failure State]:** $BD = 10$, $OB = 5$.
* **[The Resolution Heuristic]:** Theorem 13 proves they bisect each other, *not* that they equal each other. Skewing the frame lengthens one diagonal and compresses the other.
* **[Solution Pathway]:** The length of $BD$ cannot be determined. We only know $OA = OC = 5$.

**IBIE 3: The Algebraic Angle Trap**
* **[The Problem]:** In parallelogram $WXYZ$, $\angle W = (3x + 15)^\circ$ and adjacent $\angle X = (2x - 10)^\circ$. Find $x$.
* **[The Heuristic Trap]:** Equating adjacent angles.
* **[The Failure State]:** $3x + 15 = 2x - 10 \implies x = -25$.
* **[The Resolution Heuristic]:** D-Node 10 states adjacent sides are parallel. A line intersecting two parallel lines forces consecutive interior angles to be supplementary (Theorem 3).
* **[Solution Pathway]:** $$(3x + 15) + (2x - 10) = 180$$
    $$5x + 5 = 180 \implies 5x = 175 \implies x = 35$$

**IBIE 4: The Bisection Synthesis Trap**
* **[The Problem]:** In parallelogram $ABCD$, diagonals intersect at $O$. A line passing through $O$ intersects $AD$ at $P$ and $BC$ at $Q$. Prove $OP = OQ$.
* **[The Heuristic Trap]:** Trying to use parallel line proportions before proving triangle congruency.
* **[The Failure State]:** Stating $OP = OQ$ "because of symmetry."
* **[The Resolution Heuristic]:** Use the diagonal bisection invariant (Theorem 13) to unlock ASA congruency.
* **[Solution Pathway]:** In $\triangle AOP$ and $\triangle COQ$:
    $\angle PAO = \angle QCO$ (Alternate interior, $AD \parallel BC$).
    $AO = CO$ (Theorem 13).
    $\angle AOP = \angle COQ$ (Vertically opposite).
    By ASA, $\triangle AOP \cong \triangle COQ$. By CPCTC, $OP = OQ$.

### MODULE 15: PURE GEOMETRY II & COORDINATE GEOMETRY
**Active ADT Node:** Specialized Parallelograms (Rectangle, Rhombus, Square)

**Axiomatic Dependency Tree (ADT) Construction:**
* Level 0: D-Node: Rectangle, Rhombus, Square [LOCKED]
* Level 0: Theorem: Diagonals of a Rectangle [LOCKED]
* Level 0: Theorem: Diagonals of a Rhombus [LOCKED]
* Level -1: D-Node: Parallelogram, Perpendicular, Right Angle [UNLOCKED]
* Level -1: Parallelogram Invariant Theorems 12 & 13 [UNLOCKED]
* Level -1: SAS & SSS Congruence [UNLOCKED]

**Exhaustive Citation Block:**
* Selina Concise Mathematics Class 8: Chapter 13 (Understanding Quadrilaterals), pp. 148-156.
* Selina Concise Mathematics Class 9: Chapter 14 (Rectilinear Figures), pp. 165-178.

**Supermarket Intuition:**
A rectangle is simply a parallelogram that has been forced to stand up perfectly straight. A rhombus is a parallelogram that has been forced to have all four sides of equal length. A square is the genetic offspring of both: perfectly straight and perfectly equal. Because they are all mathematically classified as parallelograms, they inherit *every* parallelogram invariant (opposite sides equal, diagonals bisect). But their specialized constraints unlock new, highly specific diagonal behaviors.

**Rigorous Derivations (Bottom-Up Traversal):**

1.  **D-Node 11: Rectangle**
    A parallelogram in which at least one interior angle is a Right Angle ($90^\circ$).
    *(Note: By Theorem 3, consecutive interior angles are supplementary. If one is $90^\circ$, its neighbor must be $180^\circ - 90^\circ = 90^\circ$. Therefore, all four angles are forced to $90^\circ$).*
    *Status: [UNLOCKED]*

2.  **D-Node 12: Rhombus**
    A parallelogram in which at least one pair of adjacent sides are equal in length.
    *(Note: By Theorem 12, opposite sides are equal. If $AB = BC$, and $AB = CD$, and $BC = AD$, then $AB = BC = CD = AD$. All four sides are forced to be equal).*
    *Status: [UNLOCKED]*

3.  **D-Node 13: Square**
    A quadrilateral that fulfills the parameters of both a Rectangle and a Rhombus (all angles $90^\circ$, all sides equal).
    *Status: [UNLOCKED]*

4.  **Theorem 14: Diagonals of a Rectangle are Equal**
    * Let $ABCD$ be a rectangle. Draw diagonals $AC$ and $BD$.
    * In $\triangle DAB$ and $\triangle CBA$:
        * $AD = BC$ (Opposite sides of a parallelogram, Theorem 12).
        * $\angle DAB = \angle CBA = 90^\circ$ (D-Node 11).
        * $AB = BA$ (Common side).
    * By SAS Congruence, $\triangle DAB \cong \triangle CBA$.
    * By CPCTC, $BD = AC$. 
    *Status: [UNLOCKED]*

5.  **Theorem 15: Diagonals of a Rhombus intersect at Right Angles (Perpendicular)**
    * Let $ABCD$ be a rhombus with diagonals $AC$ and $BD$ intersecting at $O$.
    * In $\triangle AOB$ and $\triangle AOD$:
        * $AB = AD$ (D-Node 12, all sides equal).
        * $BO = DO$ (Diagonals of a parallelogram bisect each other, Theorem 13).
        * $AO = AO$ (Common side).
    * By SSS Congruence, $\triangle AOB \cong \triangle AOD$.
    * By CPCTC, $\angle AOB = \angle AOD$.
    * By the Linear Pair Postulate, $\angle AOB + \angle AOD = 180^\circ$.
    * Therefore, $2\angle AOB = 180^\circ \implies \angle AOB = 90^\circ$. 
    * The diagonals are perpendicular.
    *Status: [UNLOCKED]*

**Intuition-Building Illustrative Examples (IBIE):**

**IBIE 1: The Diagonal Equivalence Trap**
* **[The Problem]:** In Rhombus $PQRS$, diagonal $PR = 12$ and diagonal $QS = 16$. Are the diagonals equal?
* **[The Heuristic Trap]:** Assuming the extreme symmetry of a rhombus forces diagonal equality.
* **[The Failure State]:** Stating $PR = QS$ because all sides are equal.
* **[The Resolution Heuristic]:** Only Rectangles (and their offspring, Squares) guarantee equal diagonals (Theorem 14). A rhombus is a skewed frame; one diagonal stretches while the other compresses.
* **[Solution Pathway]:** The diagonals are unequal ($12 \neq 16$). They bisect at $90^\circ$ (Theorem 15), but their lengths remain independent.

**IBIE 2: The Pythagorean Temptation Trap**
* **[The Problem]:** In Rectangle $ABCD$, $AB = 3$ and $BC = 4$. Find the length of diagonal $AC$.
* **[The Heuristic Trap]:** Instantly attempting to use $a^2 + b^2 = c^2$.
* **[The Failure State]:** $3^2 + 4^2 = 25 \implies AC = 5$. 
* **[The Resolution Heuristic]:** Pythagoras is [LOCKED]. Any answer derived using it is structurally illegal. 
* **[Solution Pathway]:** Under the current SUL, the numerical length of $AC$ cannot be calculated. The problem is invalid under strict chronological constraints. (This trap exists to test your adherence to the ADT).

**IBIE 3: The Rhombus Angle Bisector Trap**
* **[The Problem]:** In Rhombus $ABCD$, prove that diagonal $AC$ bisects $\angle A$.
* **[The Heuristic Trap]:** Visually assuming the diagonal splits the angle perfectly because the shape is symmetrical.
* **[The Failure State]:** Stating "It's a property of a rhombus."
* **[The Resolution Heuristic]:** Execute the SSS pathway from Theorem 15 to prove the angles equal.
* **[Solution Pathway]:** We already proved $\triangle AOB \cong \triangle AOD$ (Theorem 15). By CPCTC, $\angle OAB = \angle OAD$. Therefore, $AC$ bisects $\angle A$.

**IBIE 4: The Overlapping Definition Trap**
* **[The Problem]:** A quadrilateral has perpendicular diagonals. Is it definitely a rhombus?
* **[The Heuristic Trap]:** Reading Theorem 15 backwards (assuming the converse without proof).
* **[The Failure State]:** Stating "Yes, because rhombus diagonals are perpendicular."
* **[The Resolution Heuristic]:** A kite (unlocked later) also has perpendicular diagonals. To be a rhombus, it MUST first be a parallelogram (diagonals bisecting each other).
* **[Solution Pathway]:** No. The diagonals must be both perpendicular *and* bisect each other to guarantee a rhombus.
