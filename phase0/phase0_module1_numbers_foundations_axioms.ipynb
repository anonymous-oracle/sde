### Phase 0 | Module 1: The Axiomatic Foundation of Numbers

#### 1. The Natural Numbers ($\mathbb{N}$) and The Peano Axioms
**The Intuition:** When we are in primary school, counting is just moving our finger from one block to the next. But mathematically, how do we define counting to a computer that knows absolutely nothing? In the late 1800s, Giuseppe Peano formalized counting using just a starting point and a "Step Forward" function.

**The Formal Definition (Peano Axioms):**
Let $\mathbb{N}$ be a set.
1. $0$ is in $\mathbb{N}$. (We have a starting point).
2. For every number $n$ in $\mathbb{N}$, there is a "Successor" $S(n)$ in $\mathbb{N}$. (Every number has a "next" number).
3. There is no number $n$ whose successor is $0$. ($0$ is the absolute floor).
4. If $S(a) = S(b)$, then $a = b$. (Two different numbers cannot have the same "next" number).
5. **The Principle of Induction:** If a statement is true for $0$, and being true for $n$ implies it is true for $S(n)$, then it is true for all natural numbers.



**Theorem 1: Addition**
We don't just "know" how to add. We must define it recursively using our Successor function $S(n)$.
* Rule 1 (Base): $a + 0 = a$
* Rule 2 (Step): $a + S(b) = S(a + b)$

**Rigorous Proof: Prove that $1 + 1 = 2$.**
*Let us define $1$ as $S(0)$, and $2$ as $S(1)$.*
We want to evaluate $1 + 1$.
Substitute the definition of $1$: $1 + S(0)$.
By Addition Rule 2, $a + S(b) = S(a + b)$. Let $a=1$ and $b=0$:
$$1 + S(0) = S(1 + 0)$$
By Addition Rule 1, $1 + 0 = 1$. Substitute this inside the parenthesis:
$$S(1)$$
By our definition, $S(1) = 2$.
Therefore, $1 + 1 = 2$. $\blacksquare$ *(This is the symbol for a completed proof).*

#### 2. The Integers ($\mathbb{Z}$) and Negative Numbers
**The Intuition:** Subtraction is just asking "what number do I add to $b$ to get $a$?" But if we ask $3 - 5$, there is no Natural Number that solves this. So, mathematicians expanded the universe to create the **Integers ($\mathbb{Z}$)** by introducing an additive inverse (negatives) such that $a + (-a) = 0$.

**Theorem 2: Why does a negative times a negative equal a positive?**
Primary school teachers say "two wrongs make a right." An academic requires proof. We will prove that $(-1) \times (-1) = 1$ using the **Distributive Property** ($a(b+c) = ab + ac$).

**Rigorous Proof:**
Start with a known truth about addition:
$$1 + (-1) = 0$$
Multiply both sides of the equation by $(-1)$:
$$(-1) \times [1 + (-1)] = (-1) \times 0$$
We know that anything multiplied by $0$ is $0$:
$$(-1) \times [1 + (-1)] = 0$$
Now, apply the Distributive Property to the left side:
$$[(-1) \times 1] + [(-1) \times (-1)] = 0$$
We know that $(-1) \times 1 = -1$:
$$-1 + [(-1) \times (-1)] = 0$$
To isolate our target term, add $1$ to both sides of the equation:
$$1 - 1 + [(-1) \times (-1)] = 1 + 0$$
$$(-1) \times (-1) = 1$$
$\blacksquare$

#### 3. The Rational Numbers ($\mathbb{Q}$) and The Real Continuum ($\mathbb{R}$)
**The Intuition:** Division is asking "what do I multiply by $b$ to get $a$?" To solve $2 \div 3$, we had to invent **Rational Numbers ($\mathbb{Q}$)**, which are fractions in the form $\frac{p}{q}$ where $p$ and $q$ are integers, and $q \neq 0$.

But are fractions enough to describe the physical universe? The ancient Greeks thought so, until they looked at a square with sides of length $1$. By the Pythagorean theorem, the diagonal is $c^2 = 1^2 + 1^2 \implies c = \sqrt{2}$. 



**Theorem 3: $\sqrt{2}$ is Irrational.**
There is no possible fraction $\frac{p}{q}$ that exactly equals the length of that diagonal. The Rational number line has "holes" in it.

**Rigorous Proof (By Contradiction):**
Assume the opposite is true: Assume $\sqrt{2}$ *is* a rational number.
Then $\sqrt{2} = \frac{p}{q}$, where $p$ and $q$ are whole integers that share no common factors (the fraction is fully simplified).
Square both sides:
$$2 = \frac{p^2}{q^2}$$
Multiply by $q^2$:
$$p^2 = 2q^2$$
Because $p^2$ is $2$ times something, $p^2$ must be an **even** number. If a square is even, its root $p$ must also be **even**.
If $p$ is even, we can write it as $p = 2k$ for some integer $k$. Substitute this back into our equation:
$$(2k)^2 = 2q^2$$
$$4k^2 = 2q^2$$
Divide by 2:
$$2k^2 = q^2$$
This implies that $q^2$ is *also* $2$ times something, meaning $q^2$ is even, which means $q$ is **even**.
**The Contradiction:** We just proved that both $p$ and $q$ are even (divisible by 2). But our very first assumption was that $\frac{p}{q}$ was fully simplified and shared no common factors! This breaks the rules of logic. Therefore, our initial assumption was false. $\sqrt{2}$ cannot be written as a fraction. $\blacksquare$

To fill these "holes" (like $\sqrt{2}$ and $\pi$), mathematicians created the **Real Numbers ($\mathbb{R}$)** using infinite limits (a concept called Dedekind Cuts). When a Neural Network trains, it operates almost exclusively in the continuous realm of $\mathbb{R}$.

**2. (Medium): $a \times 0 = 0$**

**3. (Hard): Rational + Irrational = Irrational**
Your proof is highly creative, and your algebraic manipulation to show an Even = Odd + Even contradiction is excellent for your specific example ($3/2$ and $\sqrt{2}$). 
*However, an academic proof requires generalization.* You proved it for *one* specific case, but a formal proof must show it holds for *all* cases. 
Here is the formal generalized proof:
* Assume a rational number $x$ and an irrational number $y$ sum to a rational number $z$.
* So, $x + y = z$.
* We can isolate the irrational $y$ by subtracting $x$: $y = z - x$.
* Because $z$ is rational and $x$ is rational, their difference $(z - x)$ must also be a rational number (a fraction minus a fraction is just another fraction). 
* But our equation now says: $\text{Irrational } (y) = \text{Rational } (z - x)$.
* **Contradiction!** An irrational number cannot equal a rational number. Therefore, our assumption was false. The sum MUST be irrational.
