**Supermarket intuition first:** Imagine you are shopping for groceries with a budget of exactly ₹500. Instead of calculating the exact total of 17 items to the last paisa, you round each price up or down to the nearest ten rupees, add the rounded figures in your head, and instantly know whether you are safely under budget or need to put something back. Estimation turns unwieldy exact arithmetic into quick, mentally manageable decisions while keeping the maximum possible error small and predictable—exactly the same principle you will later use to bound floating-point errors in machine-learning gradient descent or truncation errors in Fourier-series approximations.

**Rigorous development (step-by-step from first principles):**  
1. **Rounding as the foundation of estimation (review & extension from Ch. 1):** To estimate, first round each number to a convenient place (nearest 10, 100, 1000, etc.). The rounding rule is derived from the midpoint principle: for target place value \(10^k\), if the next digit \(d\) satisfies \(0 \leq d < 5\), retain the target digit; if \(5 \leq d \leq 9\), increment it. All lower digits become zero. Proof of closeness: the error per number is strictly less than \(0.5 \times 10^k\).  
2. **Estimation of sums and differences:** Replace each addend/subtrahend by its rounded version, then add/subtract the rounded values. The total error is bounded by (number of terms) × (maximum rounding error per term).  
3. **Estimation of products and quotients:** Round each factor to the nearest convenient multiple of a power of 10, then multiply/divide. For quotients, use the fact that rounding both numerator and denominator in the same direction often cancels some error (compatibility rule).  
4. **General strategy (derived):** Choose the rounding place so that the rounded numbers have only one or two non-zero digits; this maximizes mental computability while controlling relative error.  
5. **Verification step (always required):** Compute the actual value (when feasible) and compare with the estimate to quantify the absolute and percentage error. This step trains the skepticism you will need for later proofs in number theory and calculus.
