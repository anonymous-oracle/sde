#### 1. Intuition: Parts of a Whole and Base-10 Extensions
Imagine you have a single contiguous block of memory, and you must allocate parts of it to different processes. You cannot use whole numbers for this. 
* A **Fraction** is simply a way to represent dividing a whole into equal parts. If you divide a matrix into 4 equal quadrants and process 3 of them, you have processed $\frac{3}{4}$ of the matrix.
* A **Decimal** is an extension of our base-10 number system. Just as moving left from the decimal point multiplies place value by 10 (ones, tens, hundreds), moving right divides by 10 (tenths, hundredths, thousandths). It is a fraction whose denominator is a power of 10.

#### 2. Theory & Rigorous Mechanics

**Fractions ($\frac{p}{q}$):**
* **Proper Fraction:** Numerator $<$ Denominator (e.g., $\frac{3}{5}$). Represents a value strictly less than 1.
* **Improper Fraction:** Numerator $\geq$ Denominator (e.g., $\frac{7}{4}$). Can be written as a mixed number ($1 \frac{3}{4}$).
* **Equivalence:** Multiplying or dividing the numerator and denominator by the same non-zero integer yields an equivalent fraction. Mathematically: $\frac{a}{b} = \frac{a \times k}{b \times k}$.

**Operations on Fractions:**
* **Addition/Subtraction:** You cannot add parts of different sizes. You must find a common denominator, ideally the LCM of the denominators.
  $$\frac{a}{b} + \frac{c}{d} = \frac{ad + bc}{bd}$$
* **Multiplication:** Multiply numerators together, and denominators together.
  $$\frac{a}{b} \times \frac{c}{d} = \frac{ac}{bd}$$
* **Division:** Dividing by a fraction is equivalent to multiplying by its reciprocal.
  $$\frac{a}{b} \div \frac{c}{d} = \frac{a}{b} \times \frac{d}{c}$$

**Decimals and Rationality:**
Any fraction $\frac{p}{q}$ in its simplest form (where $p$ and $q$ are co-prime) will result in a **terminating decimal** if and only if the prime factorization of $q$ contains *only* powers of $2$ and/or $5$. 
* Why? Because our number system is base-10, and $10 = 2 \times 5$. 
* If $q$ contains any other prime factor (like 3 or 7), the decimal will be **non-terminating and recurring** (repeating).
