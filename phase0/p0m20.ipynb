### Lexical Geometry Postulates (D-Nodes)

#### D-Node 36: Trigonometric Ratios

Let a right-angled triangle possess an acute tracking angle $\theta$.

* **Perpendicular ($opp$):** The side strictly opposite to angle $\theta$.
* **Base ($adj$):** The side adjacent to angle $\theta$ (excluding the hypotenuse).
* **Hypotenuse ($hyp$):** The side opposite the $90^\circ$ right angle.

We define the six fundamental scale-invariant metric ratios as:

$$\sin \theta = \frac{opp}{hyp} \quad \cos \theta = \frac{adj}{hyp} \quad \tan \theta = \frac{opp}{adj}$$

$$\csc \theta = \frac{hyp}{opp} \quad \sec \theta = \frac{hyp}{adj} \quad \cot \theta = \frac{adj}{opp}$$

*Status: [UNLOCKED]*

### Rigorous Derivations (Bottom-Up Traversal)

By applying Theorem 26 (AAA Similarity), any right-angled triangles sharing the identical acute angle $\theta$ are scaling replicas. Therefore, these six fractional ratios are entirely independent of the physical size of the triangle—they are functions purely locked to the angular state $\theta$.

#### Theorem 48: The Fundamental Pythagorean Identity

We must prove that for any acute angle $\theta$, $\sin^2 \theta + \cos^2 \theta = 1$.

**Proof:**
Isolate a right-angled triangle with sides labeled relative to angle $\theta$ as $opp$, $adj$, and $hyp$. By Theorem 29 (Pythagoras' Theorem), the side lengths are bound by the following quadratic constraint:

$$(opp)^2 + (adj)^2 = (hyp)^2$$

Divide every scalar block in the equation by the squared hypotenuse value $(hyp)^2$:

$$\frac{(opp)^2}{(hyp)^2} + \frac{(adj)^2}{(hyp)^2} = \frac{(hyp)^2}{(hyp)^2}$$

Group the fractional exponents using standard algebraic index laws:

$$\left(\frac{opp}{hyp}\right)^2 + \left(\frac{adj}{hyp}\right)^2 = 1$$

Substitute the direct definition formulas mapped in D-Node 36:

$$(\sin \theta)^2 + (\cos \theta)^2 = 1 \implies \sin^2 \theta + \cos^2 \theta = 1$$

*Status: [UNLOCKED]*

#### Corollary 48.1: Secondary Identity Derivations

1. Divide the fundamental identity ($\sin^2 \theta + \cos^2 \theta = 1$) completely by $\cos^2 \theta$:

$$\frac{\sin^2 \theta}{\cos^2 \theta} + \frac{\cos^2 \theta}{\cos^2 \theta} = \frac{1}{\cos^2 \theta} \implies \tan^2 \theta + 1 = \sec^2 \theta \implies \sec^2 \theta - \tan^2 \theta = 1$$


2. Divide the fundamental identity completely by $\sin^2 \theta$:

$$\frac{\sin^2 \theta}{\sin^2 \theta} + \frac{\cos^2 \theta}{\sin^2 \theta} = \frac{1}{\sin^2 \theta} \implies 1 + \cot^2 \theta = \csc^2 \theta \implies \csc^2 \theta - \cot^2 \theta = 1$$



*Status: [UNLOCKED]*

### Intuition-Building Illustrative Examples (IBIE)

#### IBIE 1: The Algebraic Conjugate Multiplication Trap

* **[The Problem]:** Prove the following identity layout: $\frac{1}{1 + \sin \theta} + \frac{1}{1 - \sin \theta} = 2\sec^2 \theta$.
* **[The Heuristic Trap]:** Getting stuck trying to substitute fractions directly before executing basic polynomial algebraic expansions.
* **[The Failure State]:** Cross-multiplying poorly and losing the sign structure of the denominator.
* **[The Resolution Heuristic]:** Treat trigonometric components as standard algebraic binomial expressions. Execute common denominator grouping first, then apply Theorem 48 to crush the remaining terms.
* **[Solution Pathway]:**
Find the common denominator of the Left-Hand Side (LHS):

$$\text{LHS} = \frac{(1 - \sin \theta) + (1 + \sin \theta)}{(1 + \sin \theta)(1 - \sin \theta)}$$



Simplify the numerator and distribute the denominator via difference of squares:

$$\text{LHS} = \frac{2}{1 - \sin^2 \theta}$$



By Theorem 48, $\sin^2 \theta + \cos^2 \theta = 1 \implies 1 - \sin^2 \theta = \cos^2 \theta$. Substitute this block:

$$\text{LHS} = \frac{2}{\cos^2 \theta} = 2 \cdot \left(\frac{1}{\cos \theta}\right)^2 = 2\sec^2 \theta$$


$$\text{LHS} = \text{RHS. Fully verified.}$$



#### IBIE 2: The Non-Right Angle Ratio Trap

* **[The Problem]:** In an obtuse triangle $ABC$, side $AB = 13$, $BC = 14$, and altitude from $A$ to $BC$ is 12. A student calculates $\sin B = \frac{12}{13}$ and immediately concludes that $\cos B = \frac{14}{13}$. Correct this.
* **[The Heuristic Trap]:** Blindly picking random full side dimensions of a non-right triangle to insert into trigonometric functions.
* **[The Failure State]:** Stating $\cos B = \frac{14}{13}$, which is mathematically impossible since the output range of a cosine function can never exceed $1$.
* **[The Resolution Heuristic]:** Trigonometric ratios are bound *strictly* to right-angled sub-triangles. You must isolate the specific right-angled zone containing the altitude step before running the ratio calculation.
* **[Solution Pathway]:** Let the altitude from $A$ meet $BC$ at a perpendicular intersection point $D$. This constructs right $\triangle ADB$. In $\triangle ADB$, hypotenuse $AB = 13$, and opposite leg $AD = 12$. By Pythagoras, base leg $BD = \sqrt{13^2 - 12^2} = 5$. Now evaluate within this isolated right system:

$$\sin B = \frac{AD}{AB} = \frac{12}{13} \quad \cos B = \frac{BD}{AB} = \frac{5}{13}$$
