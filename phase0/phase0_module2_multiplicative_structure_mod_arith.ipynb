## Phase 0 | Module 2: The Multiplicative Structure & Modular Arithmetic

We are moving from "counting" (addition) to "scaling" (multiplication). In primary school, $2 \times 3$ is just $2 + 2 + 2$. But to a computer—and to a Number Theorist—multiplication is the gateway to **Cryptography** and **Tokenization** in LLMs.

### 1. Defining Multiplication Recursively
Just as we defined addition using the Successor function $S(n)$, we define multiplication ($\times$) for $\mathbb{N}$:
1. **Base Case:** $a \times 0 = 0$
2. **Recursive Step:** $a \times S(b) = (a \times b) + a$

> **Supermarket Intuition:** If you have 3 bags with 2 apples each, and I add one more bag ($S(3)$), you just take your current total and add one more "unit" of 2 apples.

### 2. The Jump to Modular Arithmetic (The "Clock" Math)
In Deep Learning, we often deal with fixed vocabularies. If an LLM has a vocabulary of 50,000 tokens and we add 1 to the 50,000th token, we "wrap around" to 0. This is **Modular Arithmetic**.

**Definition:** We say $a \equiv b \pmod{n}$ (read: $a$ is congruent to $b$ modulo $n$) if $n$ divides the difference $(a - b)$ exactly. Essentially, $a$ and $b$ leave the same remainder when divided by $n$.
