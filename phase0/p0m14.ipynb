### **Module 14, Node 1: Reflection (Coordinate Geometry)**

**Textbook Reference Block (Exhaustive Parallel Citation Protocol v14.2)**
* **Selina Concise Mathematics:** Class 10, Chapter 12: Reflection -> Page Range: Standard Edition Span.
* **M.L. Aggarwal Understanding ICSE:** Class 10, Reflection -> Page Range: Standard Edition Span.

**1. Supermarket Intuition**
Up to this point, our numbers have existed either in equations or on a single 1-D number line. Coordinate Geometry introduces a 2-D grid: the Cartesian Plane ($X$ and $Y$ axes). Every point in reality is given a strict GPS coordinate $(x, y)$. 
**Reflection** is the mathematical engine behind computer graphics and digital rendering. When you move a character in a video game and they look into a mirror, or when you flip an image horizontally in Photoshop, the software is executing a Reflection transformation algorithm on thousands of $(x, y)$ coordinates simultaneously. 

**2. Bounded Rigorous Derivation**
Let $P(x, y)$ be an object point on the Cartesian plane. A reflection maps it to an image point $P'(x', y')$.

* **Reflection in the X-axis ($M_x$):**
  The X-axis acts as the mirror. The horizontal position ($x$) stays identical, but the vertical position ($y$) flips to the exact opposite side of the axis.
  $$M_x(x, y) \rightarrow (x, -y)$$
* **Reflection in the Y-axis ($M_y$):**
  The Y-axis acts as the mirror. The vertical position ($y$) stays identical, but the horizontal position ($x$) flips.
  $$M_y(x, y) \rightarrow (-x, y)$$
* **Reflection in the Origin ($M_o$):**
  The origin $(0,0)$ acts as a point mirror. Both coordinates flip their signs. (This is mathematically equivalent to reflecting in the X-axis, and then taking that result and reflecting it in the Y-axis).
  $$M_o(x, y) \rightarrow (-x, -y)$$
* **Invariant Points:**
  If a point lies *exactly on* the mirror line, its reflection is itself. It does not move. Such a point is called invariant. For example, $(5, 0)$ is invariant under reflection in the X-axis.

    * *Answer:* The final coordinates are $(3, 4)$.

* **Medium (Finding the Mirror):** A point $A(4, -5)$ is reflected to $A'(-4, -5)$. Identify the mirror line and state whether the point $B(0, -5)$ is invariant under this reflection.
    * *Step 1:* Compare $A(4, -5)$ and $A'(-4, -5)$. The $y$-coordinate remained identical, but the $x$-coordinate flipped its sign.
    * *Step 2:* According to the derivations, flipping the $x$-coordinate corresponds strictly to reflection in the **Y-axis**.
    * *Step 3:* Check $B(0, -5)$ against the Y-axis mirror. $M_y(0, -5) \rightarrow (-0, -5) \rightarrow (0, -5)$.

* **Difficult (Reflection across an arbitrary line):** Point $P(2, 3)$ is reflected in the line $x = 4$ to point $P'$. Find the coordinates of $P'$.
    * *Step 1:* The mirror is not an axis; it is the vertical line $x = 4$. The $y$-coordinate will not change because the mirror is vertical. $y' = 3$.
    * *Step 2:* Calculate horizontal distance from $P$ to the mirror. The mirror is at $x = 4$. $P$ is at $x = 2$. The distance is $4 - 2 = 2$ units.
    * *Step 3:* The image must be exactly $2$ units on the *other* side of the mirror. $4 + 2 = 6$. So $x' = 6$.
    * *Answer:* $P' = (6, 3)$.

* **Extension (Geometric Mapping):** Point $A(2, 2)$ is reflected in the X-axis to get $B$. Point $B$ is reflected in the Origin to get $C$. Find the coordinates of $C$ and calculate the exact area of the triangle formed by $A, B$, and $C$.
    * *Step 1:* $M_x(A) \rightarrow B(2, -2)$.
    * *Step 2:* $M_o(B) \rightarrow C(-2, 2)$.
    * *Step 3:* Plot the triangle. $A(2, 2)$ and $B(2, -2)$ form a vertical line segment of length $4$ units. This is the base.
    * *Step 4:* $C(-2, 2)$ forms a horizontal line segment with $A(2, 2)$ of length $4$ units. This is the height. The triangle is a right-angled triangle.
    * *Step 5:* Area $= \frac{1}{2} \times \text{base} \times \text{height} = \frac{1}{2} \times 4 \times 4 = 8$.
    * *Answer:* $C = (-2, 2)$. Area is $8$ sq units.

### **The Geometric Proof of the Section Formula**

**1. The Setup (The Cartesian Grid)**
* Plot two points on a 2-D plane: $A(x_1, y_1)$ and $B(x_2, y_2)$.
* Draw a straight line connecting them. 
* Place a point $P(x, y)$ strictly somewhere on that line segment between $A$ and $B$.
* $P$ divides the line segment $AB$ into two parts: $AP$ and $PB$. 
* We are given the ratio of these parts. Let $\frac{AP}{PB} = \frac{m_1}{m_2}$.

**2. Building the Scaffolding (Right-Angled Triangles)**
To find coordinates, we must project these points down to the X and Y axes.
* Drop a vertical line straight down from $A$ to the X-axis. 
* Drop a vertical line straight down from $P$ to the X-axis. 
* Drop a vertical line straight down from $B$ to the X-axis.
* Now, draw a horizontal line strictly parallel to the X-axis passing through $A$. Let it intersect the vertical line from $P$ at a point $Q$.
* Draw another horizontal line parallel to the X-axis passing through $P$. Let it intersect the vertical line from $B$ at a point $R$.

You have just constructed two right-angled triangles sitting on the line segment: $\triangle AQP$ and $\triangle PRB$.

**3. The Logic of Similarity**
Because the horizontal lines are parallel, and the vertical lines are parallel, the angles inside these two triangles are mathematically identical. 
By the Angle-Angle (AA) similarity criterion, $\triangle AQP$ is strictly similar to $\triangle PRB$. 

The foundational law of Similar Triangles dictates that the ratio of their corresponding sides must be perfectly equal:
$$\frac{AQ}{PR} = \frac{PQ}{RB} = \frac{AP}{PB}$$

Since we already know $\frac{AP}{PB} = \frac{m_1}{m_2}$, we can lock the equation:
$$\frac{AQ}{PR} = \frac{PQ}{RB} = \frac{m_1}{m_2}$$

**4. Translating Geometry to Coordinates**
Now, we measure the lengths of the base and height of these triangles using their $(x, y)$ coordinates.
* **Base of small triangle ($AQ$):** This is the horizontal distance from $A$ to $P$. 
  $AQ = x - x_1$
* **Base of large triangle ($PR$):** This is the horizontal distance from $P$ to $B$.
  $PR = x_2 - x$
* **Height of small triangle ($PQ$):** This is the vertical distance from $A$ to $P$.
  $PQ = y - y_1$
* **Height of large triangle ($RB$):** This is the vertical distance from $P$ to $B$.
  $RB = y_2 - y$

**5. The Algebraic Execution (Finding $x$)**
We take the horizontal ratio and equate it to our known mass ratio:
$$\frac{x - x_1}{x_2 - x} = \frac{m_1}{m_2}$$

Now, we simply cross-multiply to isolate $x$:
$$m_2(x - x_1) = m_1(x_2 - x)$$
$$m_2x - m_2x_1 = m_1x_2 - m_1x$$

Bring all terms containing $x$ strictly to the Left Hand Side:
$$m_1x + m_2x = m_1x_2 + m_2x_1$$
$$x(m_1 + m_2) = m_1x_2 + m_2x_1$$

Divide to isolate $x$:
$$x = \frac{m_1x_2 + m_2x_1}{m_1 + m_2}$$

*(The exact same algebraic process applied to the $y$-coordinates yields $y = \frac{m_1y_2 + m_2y_1}{m_1 + m_2}$).*

### **The Geometric Proof of the Centroid Formula**

**1. The Geometric Axioms (The Setup)**
* Let a triangle have vertices $A(x_1, y_1)$, $B(x_2, y_2)$, and $C(x_3, y_3)$.
* A **median** is a straight line drawn from a vertex to the exact mid-point of the opposite side. Let us draw a median from vertex $A$ to side $BC$.
* Let the mid-point of side $BC$ be denoted as $D$.
* The **Centroid** ($G$) is the specific point where all three medians of a triangle intersect. A foundational axiom of Euclidean geometry states that the centroid strictly divides every median in the exact ratio of **$2:1$** (with the larger $2$-part being closest to the vertex).

**2. The Algebraic Execution**
* **Step 1: Find the coordinates of Mid-point $D$.**
  Since $D$ is the mid-point of line segment $BC$, we apply the Mid-Point Formula derived earlier:
  $$D = \left( \frac{x_2 + x_3}{2}, \frac{y_2 + y_3}{2} \right)$$

* **Step 2: Apply the Section Formula to median $AD$.**
  We now have a straight line segment $AD$. 
  Point $A$ is $(x_1, y_1)$.
  Point $D$ is $\left( \frac{x_2 + x_3}{2}, \frac{y_2 + y_3}{2} \right)$.
  The Centroid $G(x, y)$ sits on this line and divides it strictly in the ratio $m_1 : m_2 = 2 : 1$.

* **Step 3: Calculate the $x$-coordinate of $G$.**
  Substitute into the Section Formula: $x = \frac{m_1x_D + m_2x_A}{m_1 + m_2}$
  $$x = \frac{2 \left( \frac{x_2 + x_3}{2} \right) + 1 (x_1)}{2 + 1}$$
  The $2$ in the numerator and the $2$ in the denominator of the fraction cancel out perfectly:
  $$x = \frac{(x_2 + x_3) + x_1}{3}$$
  Rearranging for standard alphabetical order:
  $$x = \frac{x_1 + x_2 + x_3}{3}$$

*(The exact same substitution logic applied to the $y$-coordinates yields $y = \frac{y_1 + y_2 + y_3}{3}$, perfectly isolating the Centroid).*

### **Module 14, Node 2: Section and Mid-Point Formula**

**Textbook Reference Block (Exhaustive Parallel Citation Protocol v14.2)**
* **Selina Concise Mathematics:** Class 10, Chapter 13: Section and Mid-Point Formula -> Page Range: Standard Edition Span.
* **M.L. Aggarwal Understanding ICSE:** Class 10, Section Formula -> Page Range: Standard Edition Span.

**1. Supermarket Intuition**
If you are building a bridge between two GPS coordinates $A$ and $B$, you need to place a support pillar exactly $\frac{1}{3}^{\text{rd}}$ of the way across. How do you find the exact GPS coordinates for that pillar? You cannot just guess. The **Section Formula** is the mathematical engine that takes a straight line segment and divides it internally into any exact ratio (like $1:2$ or $3:4$). In computer graphics, this exact algorithm is used for "interpolation"—smoothly calculating the frames between point A and point B to create fluid animation.

**2. Bounded Rigorous Derivation**
* *Axiom Recap (Class 9 Distance Formula):* The absolute distance between $A(x_1, y_1)$ and $B(x_2, y_2)$ is strictly $d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$.

* **The Section Formula (Internal Division):**
  If a point $P(x, y)$ divides the line segment joining $A(x_1, y_1)$ and $B(x_2, y_2)$ exactly in the ratio $m_1 : m_2$, the coordinates of $P$ are:
  $$x = \frac{m_1x_2 + m_2x_1}{m_1 + m_2}$$
  $$y = \frac{m_1y_2 + m_2y_1}{m_1 + m_2}$$
  *(Note the cross-multiplication structure: $m_1$ multiplies with B's coordinates, $m_2$ multiplies with A's coordinates).*

* **The Mid-Point Formula (Special Case):**
  If $P$ is exactly in the center, the ratio is strictly $1:1$. The Section Formula collapses into a simple average:
  $$x = \frac{x_1 + x_2}{2}, \quad y = \frac{y_1 + y_2}{2}$$

* **Centroid of a Triangle:**
  The centroid is the exact center of mass where the three medians of a triangle intersect. For vertices $(x_1, y_1), (x_2, y_2),$ and $(x_3, y_3)$:
  $$x = \frac{x_1 + x_2 + x_3}{3}, \quad y = \frac{y_1 + y_2 + y_3}{3}$$

***

### **Example Illustrations (Override Rule v14.3)**

* **Basic (Mid-Point):** Find the mid-point of the line segment joining $A(-5, 4)$ and $B(7, -8)$.
    * *Step 1:* Identify coordinates. $x_1=-5, y_1=4$ and $x_2=7, y_2=-8$.
    * *Step 2:* Apply Mid-Point formula.
        $x = \frac{-5 + 7}{2} = \frac{2}{2} = 1$.
        $y = \frac{4 + (-8)}{2} = \frac{-4}{2} = -2$.
    * *Answer:* The mid-point is $(1, -2)$.

* **Medium (Section Formula):** Find the coordinates of the point which divides the line joining $A(-1, 7)$ and $B(4, -3)$ in the ratio $2:3$.
    * *Step 1:* $m_1 = 2, m_2 = 3$. Coordinates: $(-1, 7)$ and $(4, -3)$.
    * *Step 2:* Apply Section Formula for $x$.
        $x = \frac{2(4) + 3(-1)}{2 + 3} = \frac{8 - 3}{5} = \frac{5}{5} = 1$.
    * *Step 3:* Apply Section Formula for $y$.
        $y = \frac{2(-3) + 3(7)}{2 + 3} = \frac{-6 + 21}{5} = \frac{15}{5} = 3$.
    * *Answer:* The point is $(1, 3)$.

* **Difficult (Finding the Ratio):** In what ratio does the X-axis divide the line segment joining the points $A(3, -4)$ and $B(1, 2)$? Find the coordinates of the point of division.
    * *Step 1:* A point strictly on the X-axis has a $y$-coordinate of exactly $0$. Let the point be $P(x, 0)$ and the ratio be $k:1$.
    * *Step 2:* Apply the Section Formula exclusively for the known $y$-coordinate to find $k$.
        $y = \frac{k(y_2) + 1(y_1)}{k + 1} \implies 0 = \frac{k(2) + 1(-4)}{k + 1}$
    * *Step 3:* Multiply $(k+1)$ by $0$.
        $0 = 2k - 4 \implies 2k = 4 \implies k = 2$.
        The ratio is strictly $2:1$.
    * *Step 4:* Use $k=2$ to find the $x$-coordinate.
        $x = \frac{2(1) + 1(3)}{2 + 1} = \frac{2 + 3}{3} = \frac{5}{3}$.
    * *Answer:* The ratio is $2:1$, and the point is $(\frac{5}{3}, 0)$.

* **Extension (Parallelogram Unknown Vertex):** Three consecutive vertices of a parallelogram are $A(1, -2), B(3, 6),$ and $C(5, 10)$. Find the fourth vertex $D$.
    * *Step 1:* The foundational geometric property of a parallelogram is that its diagonals bisect each other. Therefore, the Mid-Point of diagonal $AC$ is exactly the same coordinate as the Mid-Point of diagonal $BD$.
    * *Step 2:* Let $D$ be $(x, y)$.
    * *Step 3:* Find Mid-Point of $AC$.
        $M_{AC} = (\frac{1 + 5}{2}, \frac{-2 + 10}{2}) = (\frac{6}{2}, \frac{8}{2}) = (3, 4)$.
    * *Step 4:* Find Mid-Point of $BD$ in terms of variables.
        $M_{BD} = (\frac{3 + x}{2}, \frac{6 + y}{2})$.
    * *Step 5:* Equate the coordinates.
        $\frac{3 + x}{2} = 3 \implies 3 + x = 6 \implies x = 3$.
        $\frac{6 + y}{2} = 4 \implies 6 + y = 8 \implies y = 2$.
    * *Answer:* The fourth vertex $D$ is $(3, 2)$.

### **Module 14, Node 3: Equation of a Straight Line**

**1. Supermarket Intuition**
A straight line is not just a drawing; it is a strict mathematical law that binds an infinite number of $(x, y)$ coordinates together. If you know the law (the equation), you can predict exactly where the line will be a million miles away. In machine learning, a "Linear Regression" model is simply an algorithm trying to find the best possible equation of a straight line to fit a scatter plot of data. 

**2. The Dependency & Derivation Block (Override Rules v14.5 & v14.6)**
*Note: We are introducing $4$ core formulas/conditions here. We will strictly execute $4$ mathematical derivations.*

**Formula 1: The Slope (Gradient) of a Line**
* *Concept:* The slope ($m$) measures the steepness of a line. It is strictly defined as the ratio of the vertical change ("rise") to the horizontal change ("run").
* *Derivation:* Let a line pass through $A(x_1, y_1)$ and $B(x_2, y_2)$. 
  The horizontal distance between them is strictly $(x_2 - x_1)$.
  The vertical distance between them is strictly $(y_2 - y_1)$.
  Therefore, by fundamental definition:
  $$m = \frac{y_2 - y_1}{x_2 - x_1}$$

**Formula 2: The Point-Slope Form**
* *Concept:* The equation of a line when you know one specific point on it and its slope.
* *Derivation:*
  Let a line have a known slope $m$ and pass through a specific known point $A(x_1, y_1)$. 
  Let $P(x, y)$ be *any* generic point on this line. 
  By the slope formula derived above, the slope between $P$ and $A$ must be:
  $$m = \frac{y - y_1}{x - x_1}$$
  Cross-multiply to linearize the equation:
  $$y - y_1 = m(x - x_1)$$

**Formula 3: The Slope-Intercept Form**
* *Concept:* The equation of a line when you know its slope $m$ and the exact point where it cuts the Y-axis (called the $y$-intercept, denoted as $c$).
* *Derivation:*
  If a line cuts the Y-axis at a height of $c$, the coordinate of that exact point is strictly $(0, c)$.
  We now have a known point $(0, c)$ and a known slope $m$. We substitute this strictly into the Point-Slope Form (Formula 2):
  $$y - c = m(x - 0)$$
  $$y - c = mx$$
  $$y = mx + c$$

**Formula 4: Conditions for Parallel and Perpendicular Lines**
* *Concept 4A (Parallel):* If two lines are parallel ($L_1 \parallel L_2$), they never intersect, meaning their steepness is identical.
  *Geometric Proof:* If you draw a horizontal line across two parallel lines, the corresponding angles are equal. Therefore, their rise over run is mathematically identical.
  $$m_1 = m_2$$
* *Concept 4B (Perpendicular):* If two lines intersect at exactly $90^\circ$ ($L_1 \perp L_2$), their slopes are strictly negative reciprocals.
  *Algebraic Proof (via Pythagoras):*
  Let $L_1$ and $L_2$ intersect strictly at the Origin $O(0,0)$ at a $90^\circ$ angle. 
  Let $L_1$ have slope $m_1$. A point on $L_1$ would be $A(1, m_1)$. (Since $\frac{m_1 - 0}{1 - 0} = m_1$).
  Let $L_2$ have slope $m_2$. A point on $L_2$ would be $B(1, m_2)$.
  Since $\triangle AOB$ is a right-angled triangle at $O$, Pythagoras strictly applies:
  $$OA^2 + OB^2 = AB^2$$
  Using the distance formula:
  $OA^2 = (1-0)^2 + (m_1-0)^2 = 1 + m_1^2$
  $OB^2 = (1-0)^2 + (m_2-0)^2 = 1 + m_2^2$
  $AB^2 = (1-1)^2 + (m_1-m_2)^2 = (m_1 - m_2)^2 = m_1^2 - 2m_1m_2 + m_2^2$
  Substitute back into Pythagoras:
  $$(1 + m_1^2) + (1 + m_2^2) = m_1^2 - 2m_1m_2 + m_2^2$$
  $$2 + m_1^2 + m_2^2 = m_1^2 - 2m_1m_2 + m_2^2$$
  Cancel the identical squared terms on both sides:
  $$2 = -2m_1m_2$$
  Divide by $2$:
  $$m_1m_2 = -1$$
  *(Proof complete. No external dependencies required).*

### **Example Illustrations (Override Rule v14.3)**

* **Basic (Equation from Two Points):** Find the equation of the line passing through $A(2, 5)$ and $B(-1, -4)$.
    * *Step 1 (Find $m$):* $m = \frac{-4 - 5}{-1 - 2} = \frac{-9}{-3} = 3$.
    * *Step 2 (Apply Point-Slope Form):* Use either point. Let's use $A(2,5)$.
      $y - 5 = 3(x - 2)$
    * *Step 3 (Standardize):* $y - 5 = 3x - 6 \implies 3x - y - 1 = 0$.

* **Medium (Parallel Line):** Find the equation of a line passing through $(-3, 2)$ and parallel to the line $4x - 2y + 7 = 0$.
    * *Step 1 (Extract Slope of Given Line):* Convert $4x - 2y + 7 = 0$ strictly into $y = mx + c$ form.
      $2y = 4x + 7 \implies y = 2x + \frac{7}{2}$.
      The slope $m = 2$.
    * *Step 2 (Apply Parallel Condition):* Since our target line is parallel, $m_1 = m_2$. Our slope is also strictly $2$.
    * *Step 3 (Apply Point-Slope Form):* We have $m = 2$ and point $(-3, 2)$.
      $y - 2 = 2(x - (-3)) \implies y - 2 = 2(x + 3)$
      $y - 2 = 2x + 6 \implies 2x - y + 8 = 0$.

* **Difficult (Perpendicular Bisector):** Find the equation of the perpendicular bisector of the line segment joining $P(4, 2)$ and $Q(-2, 6)$.
    * *Step 1 (Find the Mid-Point):* A perpendicular *bisector* strictly passes through the center.
      $M = \left( \frac{4 + (-2)}{2}, \frac{2 + 6}{2} \right) = \left( \frac{2}{2}, \frac{8}{2} \right) = (1, 4)$.
    * *Step 2 (Find Slope of PQ):* $m_{PQ} = \frac{6 - 2}{-2 - 4} = \frac{4}{-6} = -\frac{2}{3}$.
    * *Step 3 (Find Perpendicular Slope):* By derivation, $m_1m_2 = -1$.
      $\left(-\frac{2}{3}\right) m_2 = -1 \implies m_2 = \frac{3}{2}$.
    * *Step 4 (Form the Equation):* Use slope $3/2$ and the mid-point $(1, 4)$.
      $y - 4 = \frac{3}{2}(x - 1)$
      $2(y - 4) = 3(x - 1) \implies 2y - 8 = 3x - 3 \implies 3x - 2y + 5 = 0$.

* **Extension (Intersection of Lines):** Find the coordinates of the point of intersection of the lines $2x - y = 5$ and $x + y = 4$. *(Dependency Note: Uses Simultaneous Equations from Module 11).*
    * *Step 1:* We have a linear system:
      (1) $2x - y = 5$
      (2) $x + y = 4$
    * *Step 2:* Apply Elimination. Add equations directly to annihilate $y$.
      $3x = 9 \implies x = 3$.
    * *Step 3:* Substitute $x$ back into equation (2).
      $3 + y = 4 \implies y = 1$.
    * *Answer:* The lines intersect exactly at $(3, 1)$.
