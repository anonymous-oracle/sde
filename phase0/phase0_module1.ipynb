## **Phase 0: Elementary Foundations**
### **Module 1: The Number Line and Basic Counting**

#### **1. The Supermarket Intuition**
Imagine you are standing perfectly still in the middle of a long supermarket aisle. Let's call your starting position the "Origin." 

If you take one step forward to grab an apple, you are at position "1". If you take another step forward, you are at position "2". But what if you forgot the milk and need to take steps backward? If you step back exactly to where you started, you are at the Origin (0). If you take one *more* step backward, you are in negative territory. 

Mathematics is just a way to formally describe these steps.

#### **2. Rigorous Theory: Building the Integers**
We cannot just use words like "numbers" without defining them. Let us formally define what we are working with.

* **The Origin ($0$):** We must always have a reference point. In mathematics, we call the concept of "nothing" or our starting point zero, denoted as $0$.
* **The Successor Function:** Let's define a rule. For any position you are in, you can take exactly one standard step to the right. We call this the *successor*. The successor of $0$ is $1$. The successor of $1$ is $2$. 
* **The Natural Numbers ($\mathbb{N}$):** If we start at $1$ and keep applying the successor rule forever, we get the set of Natural Numbers. We write this as a set (a collection of items): 
    $$\mathbb{N} = \{1, 2, 3, 4, ...\}$$
* **The Whole Numbers ($\mathbb{W}$):** If we simply include our Origin ($0$) into this set, we get the Whole Numbers:
    $$\mathbb{W} = \{0, 1, 2, 3, ...\}$$
* **The Predecessor Function:** Now, let's invent the opposite rule. For any position, you can take one step to the left. This is the *predecessor*. The predecessor of $2$ is $1$. The predecessor of $1$ is $0$.
* **The Integers ($\mathbb{Z}$):** What is the predecessor of $0$? We denote this as $-1$. Continuing left gives us $-2$, $-3$, and so on. The complete set of all whole steps backward, the Origin, and all whole steps forward is called the Integers, denoted by the symbol $\mathbb{Z}$ (from the German word *Zahlen*, meaning numbers):
    $$\mathbb{Z} = \{..., -3, -2, -1, 0, 1, 2, 3, ...\}$$


#### **3. Addition and Subtraction (Geometric Definition)**
* **Addition ($+$):** Moving to the right on our line. For example, $2 + 3$ means "Start at $2$, and apply the successor function $3$ times." You land at $5$.
* **Subtraction ($-$):** Moving to the left on our line. For example, $3 - 4$ means "Start at $3$, and apply the predecessor function $4$ times." You land at $-1$.
