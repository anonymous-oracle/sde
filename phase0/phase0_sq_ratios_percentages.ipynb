### CORRECTIVE TRACK MODULE 3: COMMERCIAL ARITHMETIC (RATIOS, PROPORTIONS, & PERCENTAGES)

#### 1. Intuition (The Supermarket View)
**Ratios** are simply recipes. If a recipe calls for 2 cups of water to 1 cup of rice, the ratio is $2:1$. If you want to feed 10 people, you scale the recipe to $20:10$. Notice that mathematically, the fraction $\frac{2}{1}$ is identical to $\frac{20}{10}$. A ratio is just a line in 2D space; no matter how far you walk along that line, the relationship between the x-axis (rice) and y-axis (water) remains constant. This is your first introduction to a **linear transformation**.

**Percentages** are just a standardized ratio. Comparing $\frac{7}{16}$ to $\frac{5}{12}$ is annoying because the denominators (the wholes) are different. A percentage forces the denominator to always be exactly $100$. "Percent" literally means "per centum" (out of a hundred). It is a universal translation layer.

#### 2. Rigorous Theory & Derivations

**A. Ratios and Equivalence**
A ratio $a:b$ is the fraction $\frac{a}{b}$. 
A **proportion** is an equation stating that two ratios are equivalent: $a:b :: c:d$, which translates algebraically to $\frac{a}{b} = \frac{c}{d}$. 
By multiplying both sides by $bd$, we get the foundational rule of cross-multiplication: 
$$ad = bc$$
(The product of the extremes equals the product of the means).

**B. Direct and Inverse Variation**
* **Direct Variation:** If $y$ scales directly with $x$, then $\frac{y}{x} = k$ (where $k$ is a constant). If $x$ doubles, $y$ doubles. 
* **Inverse Variation:** If $y$ scales inversely with $x$, then $x \cdot y = k$. If you double the number of workers ($x$), the time to finish the job ($y$) is cut in half. 

**C. Percentage Mechanics**
A percentage is defined as $x\% = x \cdot 10^{-2} = \frac{x}{100}$.
To find what percentage $A$ is of $B$, we set up the proportion:
$$\frac{A}{B} = \frac{P}{100} \implies P = \left(\frac{A}{B}\right) \times 100$$
When scaling a value by a percentage, we use a multiplier. A $20\%$ increase on a base value $x$ is not just adding $20$; it is $x + 0.20x$, which factors neatly to $x(1 + 0.20) = 1.20x$.
