### **Module 02, Node 1: Integers & The Geometrical Number Line**

**Textbook Reference:** Selina Concise Mathematics (Class 6) -> Chapter 6: Negative Numbers And Integers & Chapter 7: Number Line -> Pages 29-45 (Standard Edition Span).

**1. Supermarket Intuition**
Under our previous constraints (Whole Numbers, $\mathbb{W}$), the operation $5 - 8$ causes a system crash. It is undefined. In computer science, if an algorithm attempts to subtract a larger array index from a smaller one without a framework for negative states, you get a buffer underflow. To prevent this, we must expand our numerical sandbox by creating a perfect, inverted mirror of the Natural Numbers. This allows for directional magnitude—representing concepts like financial debt, temperatures below freezing, or reverse vectors.

**2. Bounded Rigorous Derivation**
We now introduce the set of **Integers ($\mathbb{Z}$)**. 

* **The Set Definition:** $$\mathbb{Z} = \{\dots, -3, -2, -1, 0, 1, 2, 3, \dots\}$$
  This set restores the Closure Property for subtraction. For any $a, b \in \mathbb{Z}$, the result of $a - b$ is guaranteed to exist within $\mathbb{Z}$.

* **Geometrical Representation (The Number Line):**
  We map $\mathbb{Z}$ onto a one-dimensional continuous line. 
  1. We establish an origin point, defined as $0$.
  2. Values to the right of the origin are strictly positive ($+$).
  3. Values to the left of the origin are strictly negative ($-$).
  4. For any two numbers $x$ and $y$ on the line, if $y$ lies to the right of $x$, then $y > x$.

* **Absolute Value (Magnitude):**
  The absolute value $|x|$ represents the geometrical distance of the integer $x$ from the origin $0$, irrespective of its direction. Therefore, it is always non-negative.
  $$|-5| = 5 \quad \text{and} \quad |5| = 5$$

* **Arithmetic on the Number Line:**
  * Adding a positive integer $n$: Move $n$ units right.
  * Adding a negative integer $-n$: Move $n$ units left.
  * Subtracting a negative integer $-n$: The double negative reverses the directional vector. $a - (-n) = a + n$ (Move right).

### **Module 02, Node 2: Rational Numbers**

**Textbook Reference:** Selina Concise Mathematics (Class 7 & Class 8) -> Rational Numbers -> Pages 46-75 (Standard Edition Span).

**1. Supermarket Intuition**
Integers ($\mathbb{Z}$) gave us directional magnitude—the ability to move left and right of zero. But what exists *between* $0$ and $1$? In the integer framework, nothing. However, in physical reality, and certainly in data science, we must quantify fragments of a whole. When you train a neural network, your probabilities and learning rates are almost exclusively fractional values between $0$ and $1$. To process these, we must expand our numerical sandbox once again.

**2. Bounded Rigorous Derivation**
We now introduce the set of **Rational Numbers ($\mathbb{Q}$)**. We build this directly using the Integers ($\mathbb{Z}$) we just formalized.

* **The Set Definition:** A rational number is any number that can be expressed in the form $\frac{p}{q}$, where:
  1. $p, q \in \mathbb{Z}$ (Both are integers).
  2. $q \neq 0$ (The denominator cannot be zero).
  
  Therefore, the set is defined as: $$\mathbb{Q} = \left\{ \frac{p}{q} \mid p, q \in \mathbb{Z}, q \neq 0 \right\}$$

* **Rational Numbers vs. Fractions:**
  Historically, a fraction $\frac{a}{b}$ restricts $a$ and $b$ to Whole Numbers ($\mathbb{W}$). Rational numbers expand this by allowing $p$ and $q$ to be integers, thereby introducing negative ratios, such as $\frac{-3}{4}$ or $\frac{5}{-7}$.

* **Standard Form:**
  A rational number $\frac{p}{q}$ is in standard (or simplest) form if:
  1. $q$ is strictly positive ($q > 0$).
  2. $p$ and $q$ share no common divisors other than $1$.
  *If given $\frac{5}{-7}$, we multiply both numerator and denominator by $-1$ to get $\frac{-5}{7}$.*

* **Density Property:**
  Unlike integers, which have "empty space" between them, rational numbers are infinitely dense. Between any two rational numbers $x$ and $y$, there exists another rational number. The simplest one to find is their average: $\frac{x + y}{2}$.

### **Module 02, Node 3: Irrational Numbers & The Real Continuum**

**Textbook Reference:** Selina Concise Mathematics (Class 9) -> Chapter 1: Rational and Irrational Numbers -> Pages 1-22 (Standard Edition Span).

**1. Supermarket Intuition**
We built Integers ($\mathbb{Z}$) to handle direction, and Rational Numbers ($\mathbb{Q}$) to handle fractions. You might think the number line is now completely full. It is not. It is full of microscopic "holes." If you draw a square with sides of length 1, the diagonal across that square has a length of $\sqrt{2}$. No matter how hard you try, you can *never* represent $\sqrt{2}$ as a neat fraction $\frac{p}{q}$. These are the "Irrational" numbers. In machine learning, the most critical numbers you will use—like $e$ (Euler's number) for exponential decay and log-normal distributions, or $\pi$ for periodic signal processing—are purely irrational. 

**2. Bounded Rigorous Derivation**
We now introduce the set of **Irrational Numbers** and complete the **Real Number Line ($\mathbb{R}$)**.

* **The Definition:** An irrational number is a number that cannot be expressed in the form $\frac{p}{q}$, where $p, q \in \mathbb{Z}$ and $q \neq 0$.
* **Decimal Expansion Theorem:** * Rational numbers have decimal expansions that either *terminate* (e.g., $\frac{1}{4} = 0.25$) or *repeat* (e.g., $\frac{1}{3} = 0.333\dots$).
    * Irrational numbers have decimal expansions that are strictly **non-terminating and non-repeating** (e.g., $\pi = 3.14159265\dots$).
* **Surds:** Roots of positive rational numbers that cannot be simplified into rational numbers are irrational. 
    * $\sqrt{4} = 2$ (Rational).
    * $\sqrt{3} \approx 1.732\dots$ (Irrational).
* **The Real Numbers ($\mathbb{R}$):** The union of the set of Rational Numbers and the set of Irrational Numbers forms the set of Real Numbers. 
    $$\mathbb{R} = \mathbb{Q} \cup \{ \text{Irrational Numbers} \}$$
    Every point on the geometrical number line now corresponds to exactly one Real number, and every Real number corresponds to exactly one point. The continuum is complete.
