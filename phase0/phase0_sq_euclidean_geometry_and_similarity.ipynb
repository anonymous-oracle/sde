#### 1. Intuition (The Supermarket View)
Imagine taking a photograph on your phone and pinching the screen to zoom in. The objects in the picture get larger, but they do not distort. A person's height-to-width ratio remains exactly the same. The angles of the buildings remain exactly $90^\circ$. 

This is **Similarity**. Two shapes are similar if one is a perfectly scaled, rotated, or translated version of the other. **Congruence** is just a special case of similarity where the zoom scale factor is exactly $1$.

#### 2. Rigorous Theory & Derivations

Two triangles, $\Delta ABC$ and $\Delta DEF$, are **similar** (denoted as $\Delta ABC \sim \Delta DEF$) if and only if:
1.  Their corresponding angles are equal: $\angle A = \angle D$, $\angle B = \angle E$, $\angle C = \angle F$.
2.  The ratio of their corresponding sides is a constant scale factor $k$:
    $$\frac{AB}{DE} = \frac{BC}{EF} = \frac{AC}{DF} = k$$

**The Axioms of Similarity:**
* **AAA (Angle-Angle-Angle):** If two angles of one triangle are respectively equal to two angles of another triangle, then the two triangles are similar. (The third angle must be equal because they sum to $180^\circ$).
* **SAS (Side-Angle-Side):** If one angle of a triangle is equal to one angle of another triangle and the sides including these angles are proportional, the triangles are similar.
* **SSS (Side-Side-Side):** If the corresponding sides of two triangles are proportional, they are similar.

**The Basic Proportionality Theorem (Thales's Theorem):**
If a line is drawn parallel to one side of a triangle to intersect the other two sides in distinct points, the other two sides are divided in the same ratio. This is the cornerstone of scaling.
