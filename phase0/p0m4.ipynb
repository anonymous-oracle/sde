Why do we care about factoring numbers? In computer science, everything is bounded by memory and processing speed. **2. Bounded Rigorous Derivation**
Let $a, b, c \in \mathbb{W}$ (Whole Numbers).

* **Definitions:**
    * If $a \times b = c$, then $a$ and $b$ are **factors** of $c$. A factor divides the number exactly, leaving a remainder of $0$.
    * $c$ is a **multiple** of $a$ and $b$.
    * **Prime Number:** A natural number strictly greater than $1$ that has exactly two distinct factors: $1$ and itself.
    * **Composite Number:** A natural number with more than two distinct factors. ($1$ is neither prime nor composite).

* **The Divisibility Rules (Algorithmic Syntax Parsing):**
    Instead of performing long division, we exploit base-10 algebra to predict divisibility.
    Let a 3-digit number be $N = 100a + 10b + c$.
    * **By 2:** The last digit ($c$) must be even ($0, 2, 4, 6, 8$). *Proof: $100a + 10b$ is always even. The parity depends entirely on $c$.*
    * **By 3:** The sum of all digits must be a multiple of 3. *Proof: $100a + 10b + c = 99a + a + 9b + b + c = 3(33a + 3b) + (a + b + c)$. The first term is strictly divisible by 3, so $(a+b+c)$ must also be.*
    * **By 4:** The number formed by the last two digits ($10b + c$) must be divisible by 4. *Proof: $100a$ is always divisible by 4 (since $100/4 = 25$).*
    * **By 5:** The last digit must be $0$ or $5$.
    * **By 6:** The number must pass the rules for BOTH 2 and 3.
    * **By 8:** The number formed by the last three digits must be divisible by 8 (because $1000$ is strictly divisible by 8).
    * **By 9:** The sum of all digits must be a multiple of 9.
    * **By 11:** The difference between the sum of digits in odd positions and the sum of digits in even positions must be $0$ or a multiple of $11$.

### **Module 04, Node 2: Prime Factorization, HCF & LCM**

**Textbook Reference:** Selina Concise Mathematics (Class 6) -> Chapter 9: Playing With Numbers -> Pages 78-95 (Standard Edition Span).

**1. Supermarket Intuition**
Prime factorization is the process of breaking a composite number down into its atomic elements. In cryptography, multiplying two massive prime numbers is computationally trivial, but reversing that product back into its prime factors is so difficult it protects global financial networks. 
Once we have these atomic elements, we can compare datasets. The Highest Common Factor (HCF) finds the largest shared building block between them. The Least Common Multiple (LCM) finds the smallest common container that can fit all of them. You already used LCM implicitly to find common denominators in Module 03; now we formalize the engine behind it.

**2. Bounded Rigorous Derivation**
Let $A$ and $B$ be natural numbers. 
* **Prime Factorization:** Expressing a number strictly as the product of primes, typically written using the exponent syntax we unlocked in Module 01. 
    Example: $72 = 2 \times 2 \times 2 \times 3 \times 3 = 2^3 \times 3^2$.
* **Highest Common Factor (HCF):** The largest integer that divides both $A$ and $B$ without a remainder. 
    *Algorithm:* Find the prime factorization of both numbers. The HCF is the product of the lowest powers of all common prime factors.
* **Least Common Multiple (LCM):** The smallest integer that is strictly divisible by both $A$ and $B$.
    *Algorithm:* Find the prime factorization of both numbers. The LCM is the product of the highest powers of all prime factors present in either number.
* **The Relational Theorem:** For any two positive integers $a$ and $b$, the product of the numbers is equal to the product of their HCF and LCM.
    $$HCF(a, b) \times LCM(a, b) = a \times b$$

### **Module 04, Node 3: The Hierarchy of Operations (BODMAS)**

**Textbook Reference:** Selina Concise Mathematics (Class 6) -> Chapter 9: Playing With Numbers -> Pages 80-84 (Standard Edition Span).

**1. Supermarket Intuition**
Language requires syntax. If I say "eat your dinner and wash the dishes or go to bed," the instruction is ambiguous. Do you eat and wash, OR go to bed? Or do you eat, and then choose between washing and sleeping? Mathematics cannot tolerate ambiguity. When you feed a string of operations into a Python compiler, the compiler needs a deterministic hierarchy to execute them. If it processes addition before multiplication when it shouldn't, your loss function will output garbage. We formalize this compiler hierarchy as BODMAS.

**2. Bounded Rigorous Derivation**
For any complex arithmetic expression, operations must be evaluated in the following strict descending hierarchy:

* **B - Brackets:** Innermost to outermost. 
    1.  `( )` Simple or Circular Brackets
    2.  `{ }` Curly Brackets or Braces
    3.  `[ ]` Square Brackets
* **O - Of:** Functionally acts as multiplication, but holds higher priority than standard division. (e.g., "half of 10").
* **D - Division** ($\div$)
* **M - Multiplication** ($\times$)
* **A - Addition** ($+$)
* **S - Subtraction** ($-$)

*(Note: Division and Multiplication possess equal functional weight, as do Addition and Subtraction. When they appear consecutively, we read strictly left-to-right.)*
