#### 1. Intuition: The Sense of Numberness
Before we can train billion-parameter models, we must understand how to count. 
* Imagine you are counting apples. You need a starting point: 1, 2, 3. These are **Natural Numbers**. 
* What if you have no apples? We introduce 0. Natural numbers + 0 become **Whole Numbers**.
* What if you owe someone an apple? We need a concept of direction or debt. This introduces negative numbers. Whole numbers + negative numbers become **Integers**.
* What if you cut an apple in half? You need parts of a whole. This introduces fractions, and formally, **Rational Numbers** (numbers that can be expressed as a ratio).

#### 2. Theory & Definitions
Let us define our sets strictly. 
* **Natural Numbers ($\mathbb{N}$):** $\{1, 2, 3, 4, ...\}$
* **Whole Numbers ($\mathbb{W}$):** $\{0, 1, 2, 3, 4, ...\}$
* **Integers ($\mathbb{Z}$):** $\{..., -3, -2, -1, 0, 1, 2, 3, ...\}$
* **Rational Numbers ($\mathbb{Q}$):** Any number that can be written in the form $\frac{p}{q}$, where $p$ and $q$ are integers, and $q \neq 0$.

**Place Value Systems:**
A number's magnitude depends on the position of its digits. 
* *Indian System:* Groups digits from the right as 3, 2, 2, 2... (Hundreds, Thousands, Lakhs, Crores). Example: $1,23,45,678$.
* *International System:* Groups digits from the right as 3, 3, 3... (Hundreds, Thousands, Millions). Example: $12,345,678$.

**Estimation & Rounding:**
Approximating a number to its nearest tens, hundreds, or thousands to simplify calculations. If the digit to the right of the target place is 5 or greater, round up. Otherwise, keep it the same.

**HCF (Highest Common Factor) & LCM (Lowest Common Multiple):**
* **Factors** are numbers you multiply together to get another number. The HCF of two numbers is the largest factor they share.
* **Multiples** are what you get after multiplying a number by an integer. The LCM is the smallest non-zero multiple they share.
* *Core Theorem:* For any two positive integers $a$ and $b$, $a \times b = \text{HCF}(a,b) \times \text{LCM}(a,b)$.
