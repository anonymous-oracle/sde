### CORRECTIVE TRACK MODULE 2: PRIMES, RATIONAL EQUIVALENCE, AND DECIMAL PROJECTIONS

If divisibility is the mechanics of cutting, prime numbers are the indivisible atoms of mathematics. You must understand how they dictate the behavior of fractions and decimals before we can map them to the continuous spaces of Machine Learning.

#### 1. Intuition (The Supermarket View)
Imagine every number is a chemical compound. Prime numbers are the elements on the periodic table. The number 60 is not just "60"; it is a molecule made of $2 \times 2 \times 3 \times 5$. When you simplify a fraction like $\frac{42}{60}$, you are essentially crossing out the common atoms (the HCF) from the numerator and denominator until the two molecules share no common elements. 

Furthermore, our number system is Base-10 (built on the atoms 2 and 5). If a fraction's denominator contains any "alien" atoms (like 3, 7, or 11) after being fully simplified, it will never fit perfectly into a Base-10 grid. It will bleed over the edges forever, creating a repeating decimal.

#### 2. Rigorous Theory & Derivations

**A. The Fundamental Theorem of Arithmetic (FTA)**
Every integer greater than 1 either is a prime number itself or can be represented as a product of prime numbers, and this representation is unique, up to the order of the factors.
* Let $N = p_1^{a_1} \cdot p_2^{a_2} \cdots p_k^{a_k}$. 
* This prime factorization instantly gives us the HCF and LCM. The HCF is the product of the lowest powers of common primes. The LCM is the product of the highest powers of all prime factors.

**B. Rational Numbers ($\mathbb{Q}$) and Equivalence Classes**
A rational number is any number that can be expressed as $\frac{p}{q}$, where $p, q \in \mathbb{Z}$ and $q \neq 0$. 
Two fractions $\frac{a}{b}$ and $\frac{c}{d}$ are in the same *equivalence class* (they are the same number) if $ad = bc$. To add fractions $\frac{a}{b} + \frac{c}{d}$, we must translate them into a shared dimensional space. The most efficient shared space is the $\text{LCM}(b, d)$.

**C. Theorem of Terminating Decimals**
Let $x = \frac{p}{q}$ be a rational number, such that $p$ and $q$ are coprime (their HCF is 1). The decimal expansion of $x$ terminates *if and only if* the prime factorization of $q$ is of the form $2^n 5^m$, where $n, m$ are non-negative integers.
* **Proof Sketch:** If $q = 2^n 5^m$, we can multiply both numerator and denominator by $2^m 5^n$. The denominator becomes $2^{n+m} 5^{n+m} = 10^{n+m}$. Any integer divided by a power of 10 is a terminating decimal. If $q$ has any other prime factor, no integer multiplier will ever turn the denominator into a perfect power of 10.
