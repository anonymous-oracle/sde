### **Module 15, Node 1: Trigonometric Ratios & Identities**

**Inter-Bucket Dependency Ledger (To be derived in Pure Geometry Bucket):**
* *The Pythagorean Theorem:* $p^2 + b^2 = h^2$ (Requires Similar Triangles proof).

**1. Supermarket Intuition**
In Module 14, we measured straight lines on a grid. But the universe operates in angles, rotations, and waves. If you want to calculate the height of a mountain without climbing it, program the physics of a swinging pendulum, or encode audio waves for digital transmission, algebra alone fails. **Trigonometry** is the mathematical bridge between linear distance (sides) and rotational degrees (angles). It establishes a strict, predictable ratio: if an angle is fixed, the ratio between the sides of a right-angled triangle is absolute, regardless of how massive the triangle gets.

**2. The Dependency & Derivation Block (Override Rules v14.5 & v14.6)**
*Note: We are introducing $4$ core identity concepts here. We will strictly execute $4$ mathematical derivations.*

**The Setup (The Right-Angled Triangle):**
Let $\triangle ABC$ be right-angled at $B$. Let the angle at $C$ be $\theta$. 
From the perspective of $\theta$:
* The side opposite to it ($AB$) is the **Perpendicular ($p$)**.
* The side adjacent to it ($BC$) is the **Base ($b$)**.
* The longest side ($AC$) is the **Hypotenuse ($h$)**.

**Definitions (The 6 Fundamental Ratios):**
1. $\sin\theta = \frac{p}{h}$
2. $\cos\theta = \frac{b}{h}$
3. $\tan\theta = \frac{p}{b}$
*Their strict reciprocals:*
4. $\csc\theta = \frac{h}{p}$ (Reciprocal of $\sin$)
5. $\sec\theta = \frac{h}{b}$ (Reciprocal of $\cos$)
6. $\cot\theta = \frac{b}{p}$ (Reciprocal of $\tan$)

**Derivation 1: The Quotient Identity**
* *Concept:* Tangent is mathematically equal to Sine divided by Cosine.
* *Proof:* $$\frac{\sin\theta}{\cos\theta} = \frac{p/h}{b/h}$$
  The hypotenuse ($h$) cancels out perfectly in the division:
  $$= \frac{p}{b}$$
  By definition, $\frac{p}{b} = \tan\theta$. Therefore, $\tan\theta = \frac{\sin\theta}{\cos\theta}$.

**Derivation 2: The Core Pythagorean Identity**
* *Concept:* The square of Sine plus the square of Cosine strictly equals $1$.
* *Proof:*
  $$\sin^2\theta + \cos^2\theta = \left(\frac{p}{h}\right)^2 + \left(\frac{b}{h}\right)^2$$
  $$= \frac{p^2}{h^2} + \frac{b^2}{h^2} = \frac{p^2 + b^2}{h^2}$$
  By our logged axiom (The Pythagorean Theorem), we know strictly that $p^2 + b^2 = h^2$. Substitute this into the numerator:
  $$= \frac{h^2}{h^2} = 1$$
  Therefore, $\sin^2\theta + \cos^2\theta = 1$.

**Derivation 3: The Secant-Tangent Identity**
* *Concept:* $1 + \tan^2\theta = \sec^2\theta$.
* *Proof:*
  $$1 + \tan^2\theta = 1 + \left(\frac{p}{b}\right)^2 = 1 + \frac{p^2}{b^2}$$
  Apply LCM to merge the fraction:
  $$= \frac{b^2 + p^2}{b^2}$$
  By Pythagoras, $b^2 + p^2 = h^2$:
  $$= \frac{h^2}{b^2} = \left(\frac{h}{b}\right)^2$$
  By definition, $\frac{h}{b} = \sec\theta$. Therefore, $1 + \tan^2\theta = \sec^2\theta$.

**Derivation 4: The Cosecant-Cotangent Identity**
* *Concept:* $1 + \cot^2\theta = \csc^2\theta$.
* *Proof:*
  $$1 + \cot^2\theta = 1 + \left(\frac{b}{p}\right)^2 = \frac{p^2 + b^2}{p^2}$$
  By Pythagoras, $p^2 + b^2 = h^2$:
  $$= \frac{h^2}{p^2} = \left(\frac{h}{p}\right)^2$$
  By definition, $\frac{h}{p} = \csc\theta$. Therefore, $1 + \cot^2\theta = \csc^2\theta$.

### **Example Illustrations (Override Rule v14.3)**

* **Basic (Ratio Mapping):** Given $\sin\theta = \frac{3}{5}$, find the exact values of $\cos\theta$ and $\tan\theta$.
    * *Step 1:* Use the core identity $\sin^2\theta + \cos^2\theta = 1$.
    * *Step 2:* $\left(\frac{3}{5}\right)^2 + \cos^2\theta = 1 \implies \frac{9}{25} + \cos^2\theta = 1$.
    * *Step 3:* $\cos^2\theta = 1 - \frac{9}{25} = \frac{16}{25} \implies \cos\theta = \frac{4}{5}$.
    * *Step 4:* Use the quotient identity: $\tan\theta = \frac{\sin\theta}{\cos\theta} = \frac{3/5}{4/5} = \frac{3}{4}$.

* **Medium (Basic Algebraic Proof):** Prove that $\sec\theta \sqrt{1 - \sin^2\theta} = 1$.
    * *Step 1:* From the core identity, $1 - \sin^2\theta = \cos^2\theta$.
    * *Step 2:* Substitute this inside the root: $\sec\theta \sqrt{\cos^2\theta}$.
    * *Step 3:* Evaluate the root: $\sec\theta \cdot \cos\theta$.
    * *Step 4:* Since Secant is the strict reciprocal of Cosine ($\sec\theta = \frac{1}{\cos\theta}$), they multiply to exactly $1$. $1 = 1$. Proved.

* **Difficult (Fractional Identity Proof):** Prove that $\frac{\cos A}{1 + \sin A} + \frac{1 + \sin A}{\cos A} = 2 \sec A$.
    * *Step 1:* Apply LCM to add the fractions on the Left Hand Side (LHS).
      $\text{LHS} = \frac{\cos^2 A + (1 + \sin A)^2}{\cos A (1 + \sin A)}$
    * *Step 2:* Expand the binomial square $(1 + \sin A)^2$.
      $= \frac{\cos^2 A + 1 + 2\sin A + \sin^2 A}{\cos A (1 + \sin A)}$
    * *Step 3:* Group $\cos^2 A + \sin^2 A$ together. We know this strictly equals $1$.
      $= \frac{1 + 1 + 2\sin A}{\cos A (1 + \sin A)} = \frac{2 + 2\sin A}{\cos A (1 + \sin A)}$
    * *Step 4:* Factor out the $2$ from the numerator.
      $= \frac{2(1 + \sin A)}{\cos A (1 + \sin A)}$
    * *Step 5:* The $(1 + \sin A)$ terms cancel out.
      $= \frac{2}{\cos A} = 2 \sec A$. Proved.

* **Extension (Substitution Logic):** If $x = a \sec\theta + b \tan\theta$ and $y = a \tan\theta + b \sec\theta$, mathematically prove that $x^2 - y^2 = a^2 - b^2$.
    * *Step 1:* Square $x$.
      $x^2 = (a \sec\theta + b \tan\theta)^2 = a^2\sec^2\theta + b^2\tan^2\theta + 2ab\sec\theta\tan\theta$
    * *Step 2:* Square $y$.
      $y^2 = (a \tan\theta + b \sec\theta)^2 = a^2\tan^2\theta + b^2\sec^2\theta + 2ab\tan\theta\sec\theta$
    * *Step 3:* Execute the subtraction ($x^2 - y^2$). The $2ab$ middle terms cancel out perfectly.
      $x^2 - y^2 = a^2\sec^2\theta + b^2\tan^2\theta - (a^2\tan^2\theta + b^2\sec^2\theta)$
    * *Step 4:* Group by constants $a^2$ and $b^2$.
      $= a^2(\sec^2\theta - \tan^2\theta) - b^2(\sec^2\theta - \tan^2\theta)$
    * *Step 5:* Apply the Secant-Tangent Identity ($1 + \tan^2\theta = \sec^2\theta \implies \sec^2\theta - \tan^2\theta = 1$).
      $= a^2(1) - b^2(1) = a^2 - b^2$. Proved.
