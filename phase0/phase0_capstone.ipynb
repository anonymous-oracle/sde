**Your Task:** Design a programmatic logic engine (in Python or structured pseudo-code) that performs fundamental arithmetic *without* relying on native hardware multiplication or division operators. 

**Requirements:**

* `multiply(a, b)`: Must compute the product of two positive integers using *only* a loop of your `add(a, b)` logic. (No using the `*` operator).
* `divide_with_remainder(a, b)`: Must compute how many times $b$ goes into $a$ (the quotient) and what is left over, using *only* repeated subtraction. Must explicitly handle the fatal $b = 0$ error.

```python
def add(a, b):
    return a + b
def multiply(a, b):
    if b==0 or a==0:
        return 0
    if b == 1:
        return add(a, 0)
    if a==1:
        return add(0, b)
    return a + multiply(a, b - 1)

def divide_with_remainder(a, b):
    if b==0:
        print("Undefined")
        return
    quotient = 0
    while a >= b:
        a -= b
        quotient += 1
    return quotient, a

print(add(20, 2))
print(multiply(20, 2))
print(multiply(20, 1))
print(multiply(20, 0))
print(multiply(0, 33))
print(divide_with_remainder(20, 3))
print(divide_with_remainder(20, 0))
```
