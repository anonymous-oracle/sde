When moving *with* the medium (downstream), the forces add together ($S_{\text{net}} = v_{\text{boat}} + v_{\text{current}}$).
* **[Solution Pathway]:**

$$S_{\text{upstream}} = v_{\text{boat}} - v_{\text{current}} = 15 - 3 = 12 \text{ km/h}$$


$$T = \frac{\text{Distance}}{S_{\text{upstream}}} = \frac{72}{12} = 6 \text{ hours}$$
