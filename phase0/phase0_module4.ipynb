## **Phase 0: Elementary Foundations**
### **Module 4: Scaling the Scales - Exponents and the Order of Operations**

#### **1. The Supermarket Intuition**
In Module 2, we learned that multiplication is just a shortcut for repeated addition. Instead of writing `3 + 3 + 3 + 3`, we write `3 * 4`. 

But what if you are tracking something that grows exponentially, like a population of bacteria on a piece of spoiled supermarket fruit? If the bacteria doubles every hour, after 4 hours you have `2 * 2 * 2 * 2` times the original amount. We need a new mathematical shortcut for **repeated multiplication**. 

#### **2. Rigorous Theory: Exponents and PEMDAS**

* **Exponents (Powers):** An exponent tells you how many times to multiply a "base" number by itself. We write this as $x^n$. 
    * $x$ is the **base**.
    * $n$ is the **exponent** (or power).
    * Example: $2^4 = 2 * 2 * 2 * 2 = 16$.
* **The Laws of Exponents:** Because exponents are just repeated multiplication, they follow strict logical rules when we combine them:
    1.  **Multiplication Law:** When multiplying the exact same bases, you *add* the exponents: $x^a \cdot x^b = x^{a+b}$. (Because `(x*x) * (x*x*x) = x^5`).
    2.  **Power of a Power:** When raising an exponent to another exponent, you *multiply* them: $(x^a)^b = x^{ab}$.
    3.  **Negative Exponents:** A negative exponent is the inverse operation—it means repeated *division*. Formally, it drops the base to the denominator: $x^{-a} = 1 / x^a$.
* **The Order of Operations (The Grammar of Math):** If I write `2 + 3 * 4`, do I add first or multiply first? If I add first, I get 20. If I multiply first, I get 14. Math needs strict grammar so we all get the same answer.
    * We evaluate from most powerful/complex to least powerful.
    * **P/B** - Parentheses/Brackets (Overrides everything else; do what is inside first).
    * **E/O** - Exponents/Orders (Repeated multiplication).
    * **M/D** - Multiplication and Division (Standard scaling; evaluate left-to-right).
    * **A/S** - Addition and Subtraction (Standard stepping; evaluate left-to-right).
