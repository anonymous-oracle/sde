```python
import numpy as np
p1 = np.array([1, 1])
p2 = np.array([4, 5])

diff = p2 - p1

# Algebraic method
squared_diff = diff ** 2
summed_dimensions = np.sum(squared_diff)
distance_alg = np.sqrt(summed_dimensions)

# Dot product method
distance_dot = np.sqrt(np.dot(diff, diff))

print(distance_alg)
print(distance_dot)
```

## **Phase 0: Elementary Foundations**
### **Module 3: The Spaces Between - Fractions and Rational Numbers ($\mathbb{Q}$)**

#### **1. The Supermarket Intuition**
Up until now, our number line only allowed us to take discrete, whole steps. If you buy apples, you buy $1, 2,$ or $3$ apples. But what if you go to the bakery and want half a cake? The Integers ($\mathbb{Z}$) fail us here. There are no integers between $0$ and $1$. We need to invent a way to describe the spaces *between* the integers.

#### **2. Rigorous Theory: Building the Rational Numbers**
We will now expand our universe by defining a new set of numbers based on division.

* **The Rational Numbers ($\mathbb{Q}$):** The symbol $\mathbb{Q}$ stands for "Quotient". A rational number is defined strictly as any number that can be written in the form of a fraction $p / q$, where $p$ and $q$ are integers (from our set $\mathbb{Z}$), and crucially, $q \neq 0$ (as you just proved in Module 2!).
    * $p$ is the **numerator** (how many parts we have).
    * $q$ is the **denominator** (how many equal parts make up a whole).
* **Every Integer is a Rational Number:** Is $5$ a rational number? Yes, because it can be written as $5 / 1$. 
* **Equivalence Classes (Simplifying Fractions):** The fraction $1/2$ represents the exact same location on the number line as $2/4$ and $3/6$. If you multiply or divide both the numerator and denominator by the exact same non-zero integer, the value of the rational number does not change.
* **Addition and Subtraction:** To add or subtract fractions, they *must* represent the same sized slices of the whole—meaning they must have a **Common Denominator**. 
    * Example: $1/2 + 1/3$. We cannot add halves and thirds directly. We must find a common multiple for $2$ and $3$ (which is $6$).
    * $1/2 = 3/6$
    * $1/3 = 2/6$
    * $3/6 + 2/6 = 5/6$.
* **Multiplication:** Multiplication is much simpler. You just scale the numerators and the denominators directly: $(a/b) \times (c/d) = (a \times c) / (b \times d)$.
* **Division:** To divide by a fraction, you multiply by its **reciprocal** (flip the numerator and denominator): $(a/b) / (c/d) = (a/b) \times (d/c)$.
