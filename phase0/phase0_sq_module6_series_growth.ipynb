#### 1. Intuition (The Supermarket View)
Imagine you are saving money. 
**Scenario A (Arithmetic):** You put a $10$ bill in a jar every single week. The growth is constant, absolute, and linear. Week 1 is $10$, Week 2 is $20$, Week 3 is $30$. You are adding a fixed *difference*.
**Scenario B (Geometric):** You put your money in a magical bank that increases your balance by 10% every week. You are not adding a fixed amount; you are multiplying by a fixed *ratio* (1.10). The growth accelerates because the new interest generates its own interest.

This difference between linear addition and exponential multiplication is the bedrock of all dynamic systems.

#### 2. Rigorous Theory & Derivations

**A. Arithmetic Progression (AP)**
A sequence where the difference between consecutive terms is a constant $d$.
* **The $n$-th term:** $T_n = a + (n-1)d$ 
* **The Sum of $n$ terms ($S_n$):** Legend has it that a young Carl Friedrich Gauss derived this in primary school to instantly sum the numbers from 1 to 100. You pair the first and last terms, the second and second-to-last, etc., all of which equal the same sum.
  $$S_n = \frac{n}{2}(2a + (n-1)d) = \frac{n}{2}(a + l) \ \ \ \  (where\ l\ is\ the\ last\ term).$$

**B. Geometric Progression (GP)**
A sequence where the ratio of consecutive terms is a constant $r$.
* **The $n$-th term:** $T_n = ar^{n-1}$
* **The Sum of $n$ terms ($S_n$):** $$S_n = a\frac{r^n - 1}{r - 1}$$ (for $r > 1$)
* **The Sum of an Infinite GP:** If $|r| < 1$ (the terms are shrinking), the sequence converges to a finite sum even though it has infinite terms. This is your first taste of Calculus limits.
  $$S_\infty = \frac{a}{1 - r}$$

**C. Compound Interest as a GP**
Compound interest is simply a Geometric Progression where the common ratio is $\left(1 + \frac{R}{100}\right)$. The formula for the total amount $A$ after $n$ compounding periods is:
$$A = P\left(1 + \frac{R}{100}\right)^n$$
