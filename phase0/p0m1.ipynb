### 1. Intuition (The Supermarket to the Syntax)
Before we talk about memory allocation in Python (`int8`, `float64`) or tensor dimensions in NumPy, we must define the atomic units of our universe: numbers. 

Imagine you are tallying inventory. You start counting: 1, 2, 3. These are the **Natural Numbers** ($\mathbb{N}$). Eventually, your warehouse empties. You need a symbol for "nothing." The Hindus and Arabs formalized $0$, giving us the **Whole Numbers** ($\mathbb{W}$). 

What happens when you borrow inventory? You owe me 5 units. How do we mathematically represent a deficit? We invent **Integers** ($\mathbb{Z}$), moving backward across the zero boundary into negative territory. 
Next, you sell half an item. Now we are dividing integers, creating ratios: $p/q$ where $q$ is not zero. These are **Rational Numbers** ($\mathbb{Q}$). For a long time, the Greeks believed everything in the universe could be written as a neat ratio. But what is the length of the diagonal of a square with sides of length 1? By the Pythagorean theorem, it's $\sqrt{2}$. You cannot write this as a simple ratio. This breaks the rational continuum and forces the creation of **Irrational Numbers**, filling the microscopic gaps on our number line to form the complete set of **Real Numbers** ($\mathbb{R}$).

### 2. Rigorous Derivations & First Principles
Let us codify the rules governing these sets. 

**Closure Property:** A set is "closed" under an operation if performing that operation on two elements always produces an element within the same set. 
* $\mathbb{N}$ is closed under addition ($a + b \in \mathbb{N}$) but *not* under subtraction ($3 - 5 = -2$, which is not in $\mathbb{N}$).
* $\mathbb{Z}$ is closed under subtraction.
* $\mathbb{Q}$ is closed under division (except by 0).

**Commutative and Distributive Laws:**
* Commutativity of Addition: $a + b = b + a$
* Distributive Law of Multiplication over Addition: $a(b + c) = ab + ac$. This is the mechanical engine behind expanding algebraic brackets later on.

**The Foundational Proof (Class 9 threshold): Prove $\sqrt{2}$ is irrational.**
I expect you to know how to prove by contradiction.
1. Assume the opposite: $\sqrt{2}$ is rational.
2. Therefore, $\sqrt{2} = \frac{p}{q}$, where $p$ and $q$ are integers with *no common factors* (they are coprime).
3. Square both sides: $2 = \frac{p^2}{q^2} \implies 2q^2 = p^2$.
4. Since $p^2$ is a multiple of 2, $p^2$ is even. This implies $p$ must also be even. Let $p = 2k$.
5. Substitute back: $2q^2 = (2k)^2 \implies 2q^2 = 4k^2 \implies q^2 = 2k^2$.
6. This means $q^2$ is even, so $q$ is even.
7. If both $p$ and $q$ are even, they share a factor of 2. This contradicts our initial assumption in Step 2. Therefore, $\sqrt{2}$ cannot be rational.

### **Module 01, Node 1: Number System Foundations**

**Textbook Reference:** Selina Concise Mathematics (Class 6) -> Chapter 1: Number System -> Pages 1-10 (Standard Edition Span).

**1. Supermarket Intuition**
Before we can construct multi-dimensional tensors or optimize loss functions, we must agree on the fundamental syntax of quantification. A number system is simply a standardized way to represent quantities. In our standard Hindu-Arabic system, the position of a digit dictates its magnitude. If you cannot reliably order, construct, and deconstruct these base-10 representations, your future algorithms will fail at the parsing stage.

**2. Bounded Rigorous Derivation**
At this Class 6 chronologically bounded stage, our rigor focuses on positional logic and strict inequalities. 
We operate in base-10. A number consisting of digits $d_n, d_{n-1}, \dots, d_1, d_0$ represents a total value calculated by:
$$V = (d_n \times 10^n) + (d_{n-1} \times 10^{n-1}) + \dots + (d_1 \times 10^1) + (d_0 \times 10^0)$$
* **Place Value:** The value of a digit based on its position ($d_i \times 10^i$).
* **Face Value:** The actual value of the digit itself ($d_i$).
* **Ordering:** To arrange a set of discrete values $S$ in ascending order means establishing a sequence $x_1 < x_2 < x_3 \dots < x_n$. Descending order is the strict inverse: $x_1 > x_2 > x_3 \dots > x_n$.

### **Module 01, Node 3: Place Value Operations & Domain Properties**

**Textbook Reference:** Selina Concise Mathematics (Class 6) -> Chapter 4: Place Value & Chapter 5: Natural Numbers And Whole Numbers -> Pages 11-28 (Standard Edition Span).

**1. Supermarket Intuition**
When you eventually write code to multiply massive matrices or optimize search algorithms, your computer does not perform calculations "the hard way" every time. It exploits symmetries and structural shortcuts. If you know that calculating $A \times B$ yields the same result as $B \times A$, you can program the system to choose the path that requires less memory. To do this legally, we must formally define the "sandbox" we are playing in (our number sets) and the "rules of the game" (the algebraic properties of those sets).

**2. Bounded Rigorous Derivation**
At our current chronological bound, we formalize two discrete sets and their operational laws:

* **The Sets:**
    * **Natural Numbers ($\mathbb{N}$):** The set of counting numbers. $\mathbb{N} = \{1, 2, 3, \dots\}$
    * **Whole Numbers ($\mathbb{W}$):** The set of Natural numbers unioned with zero. $\mathbb{W} = \{0, 1, 2, 3, \dots\}$

* **The Axiomatic Properties of $\mathbb{W}$ (under Addition and Multiplication):**
    Let $a, b, c \in \mathbb{W}$.
    * **Closure Property:** If you operate on two elements in a set, the result must also be in that set.
        * Addition: $a + b \in \mathbb{W}$
        * Multiplication: $a \times b \in \mathbb{W}$
        * *(Note: $\mathbb{W}$ is NOT closed under subtraction or division. $3 - 5$ does not yield a Whole Number).*
    * **Commutative Law:** Order of operands does not change the result.
        * Addition: $a + b = b + a$
        * Multiplication: $a \times b = b \times a$
    * **Associative Law:** Grouping of operands does not change the result.
        * Addition: $(a + b) + c = a + (b + c)$
        * Multiplication: $(a \times b) \times c = a \times (b \times c)$
    * **Distributive Law of Multiplication over Addition:** This is the bridge between the two operators.
        * $a \times (b + c) = (a \times b) + (a \times c)$

* **Place Value Operations (Chapter 4 context):** We manipulate local values algorithmically. The "difference between place values" requires identifying $d_i \times 10^i$ and $d_j \times 10^j$, then executing standard subtraction.

### **Module 02, Node 1: Integers & The Geometrical Number Line**

**Textbook Reference:** Selina Concise Mathematics (Class 6) -> Chapter 6: Negative Numbers And Integers & Chapter 7: Number Line -> Pages 29-45 (Standard Edition Span).

**1. Supermarket Intuition**
Under our previous constraints (Whole Numbers, $\mathbb{W}$), the operation $5 - 8$ causes a system crash. It is undefined. In computer science, if an algorithm attempts to subtract a larger array index from a smaller one without a framework for negative states, you get a buffer underflow. To prevent this, we must expand our numerical sandbox by creating a perfect, inverted mirror of the Natural Numbers. This allows for directional magnitude—representing concepts like financial debt, temperatures below freezing, or reverse vectors.

**2. Bounded Rigorous Derivation**
We now introduce the set of **Integers ($\mathbb{Z}$)**. 

* **The Set Definition:** $$\mathbb{Z} = \{\dots, -3, -2, -1, 0, 1, 2, 3, \dots\}$$
  This set restores the Closure Property for subtraction. For any $a, b \in \mathbb{Z}$, the result of $a - b$ is guaranteed to exist within $\mathbb{Z}$.

* **Geometrical Representation (The Number Line):**
  We map $\mathbb{Z}$ onto a one-dimensional continuous line. 
  1. We establish an origin point, defined as $0$.
  2. Values to the right of the origin are strictly positive ($+$).
  3. Values to the left of the origin are strictly negative ($-$).
  4. For any two numbers $x$ and $y$ on the line, if $y$ lies to the right of $x$, then $y > x$.

* **Absolute Value (Magnitude):**
  The absolute value $|x|$ represents the geometrical distance of the integer $x$ from the origin $0$, irrespective of its direction. Therefore, it is always non-negative.
  $$|-5| = 5 \quad \text{and} \quad |5| = 5$$

* **Arithmetic on the Number Line:**
  * Adding a positive integer $n$: Move $n$ units right.
  * Adding a negative integer $-n$: Move $n$ units left.
  * Subtracting a negative integer $-n$: The double negative reverses the directional vector. $a - (-n) = a + n$ (Move right).

### **Module 01, Node 4: Exponents (Indices)**

To answer your first question: "Rationalizing the denominator" is the algebraic process of eliminating an irrational surd from the bottom of a fraction. For $\frac{1}{2 + \sqrt{3}}$, you multiply both the numerator and the denominator by its "conjugate", which is $2 - \sqrt{3}$. 
This exploits an algebraic identity we have not yet formally derived: $(a + b)(a - b) = a^2 - b^2$. 
By doing this: 
$$\frac{1 \times (2 - \sqrt{3})}{(2 + \sqrt{3})(2 - \sqrt{3})} = \frac{2 - \sqrt{3}}{2^2 - (\sqrt{3})^2} = \frac{2 - \sqrt{3}}{4 - 3} = 2 - \sqrt{3}$$
Thus, $x + \frac{1}{x} = (2 + \sqrt{3}) + (2 - \sqrt{3}) = 4$.

**Textbook Reference:** Selina Concise Mathematics (Class 7 & Class 8) -> Exponents -> Pages 76-95 (Standard Edition Span).

**1. Supermarket Intuition**
Multiplication was invented as a shortcut for repeated addition ($5 + 5 + 5 = 5 \times 3$). Exponents are the logical next step: a shortcut for repeated multiplication ($5 \times 5 \times 5 = 5^3$). When you scale algorithms, time complexity is often exponential. Understanding how to mathematically compress, expand, and shift these powers is mandatory for tasks ranging from calculating compound interest to analyzing the convergence of gradient descent.

**2. Bounded Rigorous Derivation**
Let $a \in \mathbb{Q}$ (where $a$ is the base) and let $m, n \in \mathbb{Z}$ (where $m, n$ are the exponents/indices).

* **The Core Definition:**
  $$a^n = a \times a \times a \dots \text{(n times, if } n > 0 \text{)}$$

* **The Laws of Exponents:**
  1. **Product Law:** When multiplying identical bases, add the indices.
     $$a^m \times a^n = a^{m+n}$$
  2. **Quotient Law:** When dividing identical bases, subtract the indices.
     $$a^m \div a^n = a^{m-n} \quad \text{(where } a \neq 0\text{)}$$
  3. **Power of a Power Law:** When raising a power to another power, multiply the indices.
     $$(a^m)^n = a^{mn}$$
  4. **The Zero Index:** Any non-zero base raised to the power of zero equals $1$. We prove this using the Quotient Law: $1 = \frac{a^m}{a^m} = a^{m-m} = a^0$.
     $$a^0 = 1 \quad \text{(where } a \neq 0\text{)}$$
  5. **The Negative Index:** A negative exponent indicates the reciprocal of the base raised to the positive of that exponent. We prove this using the Quotient Law: $\frac{1}{a^n} = \frac{a^0}{a^n} = a^{0-n} = a^{-n}$.
     $$a^{-n} = \frac{1}{a^n} \quad \text{and} \quad \left(\frac{a}{b}\right)^{-n} = \left(\frac{b}{a}\right)^n$$

### **Module 01, Node 5: Logarithms**

**Textbook Reference:** Selina Concise Mathematics (Class 9) -> Chapter: Indices (Exponents) and Logarithms -> Pages 96-120 (Standard Edition Span).

**1. Supermarket Intuition**
Exponents ask: "If I multiply a base $a$ by itself $x$ times, what is the result $N$?" ($a^x = N$). 
Logarithms ask the exact reverse question: "To what power $x$ must I raise the base $a$ to produce the result $N$?" 
A logarithm is simply an exponent written in a different syntax. It is the mathematical tool we use to pull a variable down from an exponent so that we can isolate and solve for it.

**2. Bounded Rigorous Derivation**
* **The Core Definition:**
    Let $a$ be a positive real number ($a > 0$) where $a \neq 1$, and let $N$ be a positive real number ($N > 0$).
    If $a^x = N$, then we define this logarithmically as:
    $$\log_a N = x$$
    *(Read as: "The logarithm of $N$ to the base $a$ is $x$").*

* **The Laws of Logarithms (Derived directly from the Laws of Indices):**
    We defined $a^m \times a^n = a^{m+n}$ in the previous node. We map this to logarithms.
    Let $M = a^x \implies \log_a M = x$.
    Let $N = a^y \implies \log_a N = y$.
    
    1.  **Product Law:** $M \times N = a^x \times a^y = a^{x+y}$. Converting back to log form: 
        $$\log_a(MN) = x + y \implies \log_a(MN) = \log_a M + \log_a N$$
    2.  **Quotient Law:** $\frac{M}{N} = \frac{a^x}{a^y} = a^{x-y}$. Converting back to log form:
        $$\log_a\left(\frac{M}{N}\right) = \log_a M - \log_a N$$
    3.  **Power Law:** $M^n = (a^x)^n = a^{xn}$. Converting back to log form:
        $$\log_a(M^n) = xn = n \log_a M$$

* **Axiomatic Base Properties:**
    * Since $a^1 = a$, it follows that $\log_a a = 1$.
    * Since $a^0 = 1$ (as derived in Node 4), it follows that $\log_a 1 = 0$.
