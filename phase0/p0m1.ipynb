### THEMATIC MODULE 1: Foundations of the Number System (Integers & Rationals)

#### 1. Intuition: The Supermarket View of Numbers
Mathematics is built on the necessity of representation. We begin with counting objects: 1, 2, 3. These are your **Natural Numbers** ($\mathbb{N}$). Eventually, humanity needed to quantify "nothing," introducing zero to create the **Whole Numbers** ($\mathbb{W}$). 

But what happens when you owe someone apples, or you walk backward? We must reflect our number line symmetrically across zero. This introduces negative counts, expanding our set to **Integers** ($\mathbb{Z}$). 

Finally, a whole unit often requires slicing. If you divide a pizza into 4 slices and take 3, you have a ratio of integers: 3 to 4. This fractional expansion creates the **Rational Numbers** ($\mathbb{Q}$). In future phases, you will see how these discrete algebraic structures form the precise data types required for computational algorithms and cryptography.

#### 2. Rigorous Derivation and Properties
We define our sets mathematically:
* $\mathbb{N} = \{1, 2, 3, \dots\}$
* $\mathbb{W} = \{0, 1, 2, 3, \dots\}$
* $\mathbb{Z} = \{\dots, -2, -1, 0, 1, 2, \dots\}$
* $\mathbb{Q} = \{\frac{p}{q} \mid p, q \in \mathbb{Z}, q \neq 0\}$

We strictly observe the following algebraic axioms on these sets (specifically $\mathbb{Z}$ and $\mathbb{Q}$):
1.  **Closure:** For any $a, b \in \mathbb{Z}$, the sum $a+b$ and product $ab$ belong to $\mathbb{Z}$.
2.  **Commutativity:** $a+b = b+a$ and $ab = ba$. (Note: Subtraction and division fail here).
3.  **Associativity:** $a+(b+c) = (a+b)+c$ and $a(bc) = (ab)c$.
4.  **Distributivity:** Multiplication distributes over addition: $a(b+c) = ab + ac$.
5.  **Identities:** $0$ is the additive identity ($a+0=a$). $1$ is the multiplicative identity ($a \times 1 = a$).
6.  **Inverses:** Every $a \in \mathbb{Z}$ has an additive inverse $-a$ such that $a + (-a) = 0$. Every non-zero $a \in \mathbb{Q}$ has a multiplicative inverse $1/a$ such that $a \times (1/a) = 1$.
