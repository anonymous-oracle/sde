### **Module 03, Node 1: Fractions & Typological Syntax**

**Textbook Reference:** Selina Concise Mathematics (Class 6) -> Chapter 14: Fractions -> Pages 128-140 (Standard Edition Span).

**1. Supermarket Intuition**
We defined Rational Numbers ($\mathbb{Q}$) as ratios of integers ($\frac{p}{q}$). However, before we abstracted to negative ratios, elementary mathematics established "Fractions" strictly as fragments of a positive whole. In data structures, when you slice an array or calculate a normalized probability between 0 and 1, you are performing fractional arithmetic. We must formally define the structural types of these fragments so we can write algorithms to convert between them without data loss.

**2. Bounded Rigorous Derivation**
At our current bound, a fraction is defined as $\frac{a}{b}$, where $a, b \in \mathbb{W}$ (Whole numbers) and $b \neq 0$.
* **$a$ (Numerator):** The count of the fragments we possess.
* **$b$ (Denominator):** The total number of equal fragments that constitute exactly one whole unit.

**Typological Classification:**
1.  **Proper Fraction:** The absolute value of the numerator is strictly less than the denominator ($a < b$). The total value is strictly $< 1$. Example: $\frac{3}{5}$.
2.  **Improper Fraction:** The numerator is greater than or equal to the denominator ($a \ge b$). The total value is $\ge 1$. Example: $\frac{7}{4}$.
3.  **Mixed Fraction (Mixed Number):** A composite syntax consisting of an integer and a proper fraction. Example: $1 \frac{3}{4}$, which mathematically means $1 + \frac{3}{4}$.

**Algorithmic Conversion:**
* **Mixed to Improper:** For $Q \frac{R}{D}$ (where $Q$ is quotient/integer, $R$ is remainder/numerator, $D$ is divisor/denominator), the improper numerator is $(Q \times D) + R$. The denominator $D$ remains unchanged.
    * $$2 \frac{3}{5} = \frac{(2 \times 5) + 3}{5} = \frac{13}{5}$$
* **Equivalent Fractions:** By the Identity Property of Multiplication (multiplying by $1$, or $\frac{n}{n}$), if we multiply or divide both $a$ and $b$ by the same non-zero scalar, the ratio's magnitude remains identical. $\frac{a}{b} = \frac{a \times n}{b \times n}$.

### **Module 03, Node 2: Fraction Operations (Addition, Subtraction, Multiplication, Division)**

**Textbook Reference:** Selina Concise Mathematics (Class 6 & Class 7) -> Chapter 14 (Class 6) / Fractions (Class 7) -> Operations on Fractions.

**1. Supermarket Intuition**
When dealing with fractions, you are dealing with pieces of a defined whole. You cannot simply add a "third" of a pizza to a "fourth" of a pizza and say you have "two sevenths." That is a catastrophic failure of unit matching. In data engineering, trying to merge two datasets with different structural normalizations without aligning them first will corrupt your entire pipeline. 
Addition and Subtraction require a *common base unit* (a common denominator). Multiplication and Division, however, are scaling operations—you are taking a fragment *of* a fragment, which scales the numerators and denominators independently.

**2. Bounded Rigorous Derivation**
Let $\frac{a}{b}$ and $\frac{c}{d}$ be fractions where $b \neq 0$ and $d \neq 0$.

* **Addition & Subtraction (The LCM Mandate):**
    To compute $\frac{a}{b} \pm \frac{c}{d}$, we must first make the denominators identical. We find the Least Common Multiple (LCM) of $b$ and $d$. Let this LCM be $L$.
    We convert both to equivalent fractions: $\frac{a \times n_1}{L} \pm \frac{c \times n_2}{L}$.
    Once the denominators are identical, the operation only applies to the numerator: 
    $$\frac{x}{L} \pm \frac{y}{L} = \frac{x \pm y}{L}$$

* **Multiplication (Scaling):**
    Multiplication is a direct mapping of numerator to numerator and denominator to denominator.
    $$\frac{a}{b} \times \frac{c}{d} = \frac{a \times c}{b \times d}$$

* **Division (The Reciprocal):**
    Division by a rational number is strictly defined as multiplication by its reciprocal.
    $$\frac{a}{b} \div \frac{c}{d} = \frac{a}{b} \times \frac{d}{c}$$
    *(Note: All mixed fractions must be converted to improper fractions before applying these operational algorithms.)*
