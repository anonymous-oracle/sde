**1. Supermarket Intuition**
In Module 09, an algebraic expression was just a theoretical blueprint ($2x + 5$). It had no specific value until we plugged data into it. An **Equation** is fundamentally different. It is a strict mathematical scale perfectly balanced by an equals sign ($=$). If I say $2x + 5 = 15$, I am stating a rigid fact about reality. In data science, you will constantly set up these "scales" to calculate weights, biases, and threshold values. The goal is to isolate the unknown variable to one side of the scale to reveal its true value. 

**2. Bounded Rigorous Derivation**
* **The Definition:** A linear equation in one variable is an equation where the highest power of the variable is strictly $1$ (e.g., $x$, not $x^2$). Its standard form is $ax + b = 0$.
* **The Golden Rule of Equations:** Whatever operation you perform on the Left Hand Side (LHS), you must perform the exact same operation on the Right Hand Side (RHS) to maintain the balance.
* **Transposition Algorithm:**
    Instead of manually adding/subtracting from both sides every time, we "transpose" terms across the equals sign. When a term crosses the equals sign, its operational sign strictly reverses:
    * $+$ becomes $-$
    * $-$ becomes $+$
    * $\times$ (multiplication) becomes $\div$ (division)
    * $\div$ (division) becomes $\times$ (multiplication)
* **The Solving Sequence:**
    1. Expand all brackets using the Distributive Law (Module 09).
    2. Transpose all variable terms to the LHS.
    3. Transpose all constant (numerical) terms to the RHS.
    4. Combine Like Terms.
    5. Isolate the variable by dividing by its coefficient.

**1. Supermarket Intuition**
A single equation with two variables (like $x + y = 10$) is not a locked scale; it is a sliding continuum. If $x=1$, $y=9$. If $x=2$, $y=8$. There are infinite solutions. To lock down the exact coordinates of reality, you need a second, intersecting truth (like $x - y = 2$). In data science, you rarely have just one unknown. When a neural network calculates optimal weights, it is essentially solving millions of "Simultaneous Equations" as matrices to find the exact point where all the data intersects. 

**2. Bounded Rigorous Derivation**
A system of simultaneous linear equations in two variables takes the standard form:
$a_1x + b_1y + c_1 = 0$
$a_2x + b_2y + c_2 = 0$

To solve this, we must mathematically compress it back into a one-variable equation (Node 1). We have two primary algorithmic methods:

* **Method 1: Substitution (The Injection Protocol)**
    1. Choose one equation and isolate one variable entirely on the LHS (e.g., $x = 10 - y$).
    2. *Substitute* this entire expression into the second equation in place of $x$. 
    3. The second equation is now strictly in terms of $y$. Solve for $y$.
    4. Inject the numerical value of $y$ back into the first equation to find $x$.

* **Method 2: Elimination (Equating Coefficients)**
    1. Multiply one or both equations by integer scalars so that the numerical coefficient of one variable (say, $x$) becomes exactly identical in both equations.
    2. Add or subtract the two equations. Because the coefficients are identical, the $x$ terms will annihilate each other ($0x$).
    3. You are left with a one-variable equation in $y$. Solve for $y$, then substitute back to find $x$.

### **Module 11, Node 3: Linear Inequations (Class 10)**

**1. Supermarket Intuition**
In Node 1 and Node 2, we dealt with perfect balance ($=$). But reality is rarely a perfect balance. A server does not crash exactly at $16$ GB of RAM usage; it crashes if usage is *greater than* $16$ GB. You do not need exactly $35\%$ to pass an exam; you need *at least* $35\%$. These are **Inequations**. In machine learning, algorithms rely heavily on inequalities to classify data (e.g., "If $P(x) > 0.5$, classify as spam") and to set optimization constraints (bounding parameters so a loss function doesn't drift to infinity).

**2. Bounded Rigorous Derivation**
An inequation uses the strict comparative operators: $<$ (less than), $>$ (greater than), $\le$ (less than or equal to), and $\ge$ (greater than or equal to).

* **The Domain (Replacement Set):**
    Because inequalities have infinite solutions, you must know what universe of numbers you are allowed to draw from.
    * $\mathbb{N}$ (Natural Numbers): $\{1, 2, 3, \dots\}$
    * $\mathbb{W}$ (Whole Numbers): $\{0, 1, 2, 3, \dots\}$
    * $\mathbb{Z}$ or $\mathbb{I}$ (Integers): $\{\dots, -2, -1, 0, 1, 2, \dots\}$
    * $\mathbb{R}$ (Real Numbers): The continuous infinite line (Integers + Rationals + Irrationals).

* **The Transposition Rules:**
    You can add, subtract, and transpose terms across an inequality sign exactly as you do with an equals sign.
    * *However, there is ONE supreme, unbreakable rule:* **The Negative Flip Mandate.**
    * If you strictly multiply or divide an inequation by a *negative* number, the inequality sign MUST flip its direction to preserve mathematical truth.
    * *Example:* $-x < 5 \implies x > -5$.

* **Compound Inequations:**
    An expression like $-3 < 2x - 1 \le 7$. You must solve this by splitting it into two distinct inequations ($-3 < 2x - 1$ AND $2x - 1 \le 7$), solving both, and finding the strict intersection of their Solution Sets.

### **Corrective Track: Simultaneous Equations (Cross-Multiplication)**

**1. Bounded Rigorous Derivation**
When dealing with massive coefficients, substitution and elimination become prone to arithmetic errors. Cross-multiplication is a deterministic formula that solves for $x$ and $y$ instantly.
Given two equations in strict standard form ($a_1x + b_1y + c_1 = 0$ and $a_2x + b_2y + c_2 = 0$), the solution is derived by:

$$\frac{x}{b_1c_2 - b_2c_1} = \frac{y}{c_1a_2 - c_2a_1} = \frac{1}{a_1b_2 - a_2b_1}$$

**Example Illustration:**
Solve: $2x + 3y - 17 = 0$ and $3x - 2y - 6 = 0$.
* *Identify Coefficients:* $a_1=2, b_1=3, c_1=-17$ | $a_2=3, b_2=-2, c_2=-6$.
* *Apply Formula for $x$:* $\frac{x}{(3)(-6) - (-2)(-17)} = \frac{1}{(2)(-2) - (3)(3)} \implies \frac{x}{-18 - 34} = \frac{1}{-4 - 9} \implies \frac{x}{-52} = \frac{1}{-13} \implies x = 4$.
* *Apply Formula for $y$:* $\frac{y}{(-17)(3) - (-6)(2)} = \frac{1}{-13} \implies \frac{y}{-51 + 12} = \frac{1}{-13} \implies \frac{y}{-39} = \frac{1}{-13} \implies y = 3$.
