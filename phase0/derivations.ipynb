### **CORRECTIVE TRACK: THE BYPASSED DERIVATIONS**

#### **1. Proof by Contradiction: $\sqrt{2}$ is Irrational (Module 02)**
*The fundamental proof that numbers exist beyond fractions.*
* **Assume the opposite:** Assume $\sqrt{2}$ is rational. Therefore, it can be written as an irreducible fraction $\frac{p}{q}$, where $p$ and $q$ are integers with *no common factors* other than 1, and $q \neq 0$.
* $$\sqrt{2} = \frac{p}{q}$$
* Square both sides: $$2 = \frac{p^2}{q^2} \implies 2q^2 = p^2$$
* **Logic step:** Since $p^2$ is exactly 2 times an integer ($q^2$), $p^2$ must be an even number. If a square is even, its square root must also be even. Therefore, $p$ is even.
* Let $p = 2k$ (where $k$ is an integer). Substitute this back into the equation:
* $$2q^2 = (2k)^2 \implies 2q^2 = 4k^2 \implies q^2 = 2k^2$$
* **Logic step:** By the exact same logic, $q^2$ is even, which means $q$ is even.
* **The Contradiction:** If both $p$ and $q$ are even, they share a common factor of 2. This directly contradicts our foundational assumption that $\frac{p}{q}$ was irreducible. The original assumption must be false. $\sqrt{2}$ is irrational.

#### **2. Sum of First $n$ Natural Numbers (Module 08 - Banking)**
*Used to derive the Equivalent Principal for Recurring Deposits.*
* Let $S = 1 + 2 + 3 + \dots + (n-1) + n$
* Write the exact same series in reverse order directly below it:
    $$S = n + (n-1) + (n-2) + \dots + 2 + 1$$
* Add the two equations vertically, term by term:
    $$2S = (n+1) + (n+1) + (n+1) + \dots + (n+1)$$
* Since there are exactly $n$ terms, the sum is $n \times (n+1)$.
    $$2S = n(n+1) \implies S = \frac{n(n+1)}{2}$$

#### **3. Completing the Square: Deriving the Quadratic Formula (Module 12)**
*The algorithmic engine of non-linear mathematics.*
* Start with the standard form: $$ax^2 + bx + c = 0$$
* Divide the entire equation by $a$ (to make the coefficient of $x^2$ strictly 1):
    $$x^2 + \frac{b}{a}x + \frac{c}{a} = 0$$
* Transpose the constant term:
    $$x^2 + \frac{b}{a}x = -\frac{c}{a}$$
* **The Core Trick (Completing the Square):** We want the LHS to look like $(x + k)^2 = x^2 + 2kx + k^2$. To achieve this, we take half of the $x$ coefficient ($\frac{b}{2a}$), square it, and add it strictly to *both* sides of the equation.
    $$x^2 + \frac{b}{a}x + \left(\frac{b}{2a}\right)^2 = -\frac{c}{a} + \left(\frac{b}{2a}\right)^2$$
* The LHS is now a perfect square. Evaluate the RHS:
    $$\left(x + \frac{b}{2a}\right)^2 = -\frac{c}{a} + \frac{b^2}{4a^2}$$
    $$\left(x + \frac{b}{2a}\right)^2 = \frac{-4ac + b^2}{4a^2}$$
* Take the square root of both sides (introducing the $\pm$ bound):
    $$x + \frac{b}{2a} = \pm\frac{\sqrt{b^2 - 4ac}}{2a}$$
* Isolate $x$:
    $$x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$$

#### **4. Deriving the Sum of an Arithmetic Progression (Module 13)**
* Let the sum of $n$ terms be $S_n = a + (a+d) + (a+2d) + \dots + (l-d) + l$ *(where $l$ is the last term, $t_n$)*.
* Write the series in reverse:
    $S_n = l + (l-d) + (l-2d) + \dots + (a+d) + a$
* Add vertically. The $d$ terms annihilate each other perfectly:
    $2S_n = (a+l) + (a+l) + (a+l) + \dots + (a+l)$
    $2S_n = n(a+l) \implies S_n = \frac{n}{2}(a+l)$
* Substitute $l = a + (n-1)d$:
    $$S_n = \frac{n}{2}[a + a + (n-1)d] = \frac{n}{2}[2a + (n-1)d]$$

#### **5. Deriving the Sum of a Geometric Progression (Module 13)**
* Let $S_n = a + ar + ar^2 + ar^3 + \dots + ar^{n-1}$
* Multiply the entire equation by the common ratio $r$:
    $r \cdot S_n = ar + ar^2 + ar^3 + \dots + ar^{n-1} + ar^n$
* Subtract the second equation from the first. Every term cancels out (telescoping) except the first term of the first equation and the last term of the second equation.
    $$S_n - r \cdot S_n = a - ar^n$$
    $$S_n(1 - r) = a(1 - r^n)$$
* Isolate $S_n$:
    $$S_n = \frac{a(1 - r^n)}{1 - r}$$

### **Module 14, Node 2: Section and Mid-Point Formula (Re-Initiation)**

*Note: The formal geometric derivation for the Section Formula was executed in my previous response. We proceed directly to the illustrations and practice.*

**1. Bounded Rigorous Derivation (Summary)**
If $P(x, y)$ divides $A(x_1, y_1)$ and $B(x_2, y_2)$ in the ratio $m_1 : m_2$:
$$x = \frac{m_1x_2 + m_2x_1}{m_1 + m_2}, \quad y = \frac{m_1y_2 + m_2y_1}{m_1 + m_2}$$

### **Example Illustrations (Override Rule v14.3)**

* **Basic (Mid-Point):** Find the mid-point of $A(-5, 4)$ and $B(7, -8)$.
    * *Step 1:* $x = \frac{-5 + 7}{2} = 1$. $y = \frac{4 + (-8)}{2} = -2$.
    * *Answer:* $(1, -2)$.

* **Medium (Section Formula):** Find the coordinates of the point dividing $A(-1, 7)$ and $B(4, -3)$ in the ratio $2:3$.
    * *Step 1:* $x = \frac{2(4) + 3(-1)}{2 + 3} = \frac{5}{5} = 1$.
    * *Step 2:* $y = \frac{2(-3) + 3(7)}{2 + 3} = \frac{15}{5} = 3$.
    * *Answer:* $(1, 3)$.

* **Difficult (Finding the Ratio):** In what ratio does the X-axis divide $A(3, -4)$ and $B(1, 2)$? Find the point.
    * *Step 1:* Point on X-axis is $(x, 0)$. Let ratio be $k:1$.
    * *Step 2:* Use the known $y$-coordinate to solve for $k$: $0 = \frac{k(2) + 1(-4)}{k + 1} \implies 2k - 4 = 0 \implies k = 2$.
    * *Step 3:* Ratio is $2:1$. Find $x$: $x = \frac{2(1) + 1(3)}{2 + 1} = \frac{5}{3}$.
    * *Answer:* Ratio $2:1$, Point $(\frac{5}{3}, 0)$.

* **Extension (Parallelogram):** Three consecutive vertices are $A(1, -2), B(3, 6),$ and $C(5, 10)$. Find the fourth vertex $D(x,y)$.
    * *Step 1:* Diagonals bisect. Mid-point of $AC$ = Mid-point of $BD$.

    * *Step 3:* $M_{BD} = (\frac{3+x}{2}, \frac{6+y}{2})$.
    * *Step 4:* Equate: $\frac{3+x}{2} = 3 \implies x=3$. $\frac{6+y}{2} = 4 \implies y=2$.
    * *Answer:* $D(3, 2)$.
