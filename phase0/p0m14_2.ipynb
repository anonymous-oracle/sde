**Supermarket Intuition:**
Imagine two sticks connected by a hinge. If you open the hinge just a tiny bit, the gap between the ends is small. As you open the hinge wider (increasing the angle), the gap (the third side) gets longer. The widest angle will always look at the longest side. Furthermore, the shortest path between two points is a straight line; therefore, taking a detour through a third point (forming a triangle) will *always* be a longer journey than walking straight there. 

**Rigorous Derivations (Bottom-Up Traversal):**

1.  **Theorem 10: Angle-Side Inequality (Longer Side $\rightarrow$ Larger Angle)**
    If two sides of a triangle are unequal, the angle opposite the longer side is greater than the angle opposite the shorter side.
    * Let $\triangle ABC$ have $AC > AB$.
    * Mark a point $D$ on $AC$ such that $AD = AB$. Connect $B$ to $D$.
    * In $\triangle ABD$, since $AD = AB$, by the Isosceles Triangle Theorem (Theorem 7), $\angle ABD = \angle ADB$.
    * In $\triangle BDC$, $\angle ADB$ is an exterior angle. By the Exterior Angle Theorem (Theorem 5), the exterior angle is strictly greater than either interior opposite angle. Thus, $\angle ADB > \angle C$.
    * Since $\angle ABD = \angle ADB$, it follows that $\angle ABD > \angle C$.
    * The total angle $\angle ABC$ is made of $\angle ABD + \angle DBC$. Therefore, $\angle ABC > \angle ABD$.
    * Combining the inequalities: $\angle ABC > \angle ABD > \angle C \implies \angle B > \angle C$.
    *Status: [UNLOCKED]*

2.  **Theorem 11: The Triangle Inequality Theorem**
    The sum of the lengths of any two sides of a triangle is strictly greater than the length of the third side.
    * Let $\triangle ABC$ be a triangle. We must prove $AB + AC > BC$.
    * Extend $BA$ to a point $D$ such that $AD = AC$. Connect $C$ to $D$.
    * In $\triangle ADC$, since $AD = AC$, by Theorem 7, $\angle ACD = \angle D$.
    * The total angle $\angle BCD = \angle BCA + \angle ACD$. Therefore, $\angle BCD > \angle ACD$.
    * Substituting $\angle D$ for $\angle ACD$, we get $\angle BCD > \angle D$.
    * Now look at the large $\triangle BCD$. We just established that $\angle BCD$ is greater than $\angle D$. 
    * By the converse of Theorem 10 (Larger Angle $\rightarrow$ Longer Side), the side opposite the larger angle must be longer.
    * The side opposite $\angle BCD$ is $BD$. The side opposite $\angle D$ is $BC$.
    * Therefore, $BD > BC$.
    * Since $BD = AB + AD$, and $AD = AC$, we substitute to get $AB + AC > BC$.
    *Status: [UNLOCKED]*

**Intuition-Building Illustrative Examples (IBIE):**

**IBIE 1: The Impossible Stick Trap**
* **[The Problem]:** Can a triangle be formed with side lengths $3$, $4$, and $8$?
* **[The Heuristic Trap]:** Assuming any three numbers can form a closed polygon.
* **[The Failure State]:** Stating "Yes" without running the inequality check.
* **[The Resolution Heuristic]:** The sum of the two smallest sides MUST be strictly greater than the largest side.
* **[Solution Pathway]:** $3 + 4 = 7$. Since $7$ is not greater than $8$, the two shorter sides will fold flat and never touch. No triangle is possible.

**IBIE 2: The Visual Angle Trap**
* **[The Problem]:** In $\triangle XYZ$, $XY = 5$, $YZ = 7$, $XZ = 6$. The diagram shows $\angle X$ as acute and $\angle Y$ as obtuse. Which angle is actually the largest?
* **[The Heuristic Trap]:** Trusting the visual representation over the mathematical data.
* **[The Failure State]:** Declaring $\angle Y$ as the largest because it looks obtuse in the drawing.
* **[The Resolution Heuristic]:** Apply Theorem 10. The largest angle sits opposite the longest side.
* **[Solution Pathway]:** The longest side is $YZ = 7$. The angle opposite $YZ$ is $\angle X$. Therefore, $\angle X$ is mathematically the largest angle, regardless of the drawing.

**IBIE 3: The Difference Bound Trap**
* **[The Problem]:** Two sides of a triangle are $9$ and $14$. What are the constraints on the third side $x$?
* **[The Heuristic Trap]:** Only checking the upper bound ($9 + 14 > x$).
* **[The Failure State]:** Concluding $x < 23$, allowing $x = 1$ or $x = 2$, which are impossible.
* **[The Resolution Heuristic]:** The inequality works in all directions. $a + c > b \implies c > b - a$. The third side must be greater than the difference of the other two.
* **[Solution Pathway]:** Upper bound: $9 + 14 > x \implies x < 23$.
    Lower bound: $9 + x > 14 \implies x > 5$.
    Constraint: $5 < x < 23$.

**IBIE 4: The Altitude vs Side Trap**
* **[The Problem]:** In $\triangle ABC$, $AD$ is an altitude perpendicular to $BC$. Prove that $AB > AD$.
* **[The Heuristic Trap]:** Trying to use SSS or SAS without having congruent triangles.
* **[The Failure State]:** Stating $AB > AD$ visually.
* **[The Resolution Heuristic]:** Apply the Angle-Side inequality in the newly formed right triangle.
* **[Solution Pathway]:** In $\triangle ABD$, $\angle ADB = 90^\circ$. By Theorem 4 (Angle sum), the other two angles must sum to $90^\circ$, meaning they are acute (less than $90^\circ$). Therefore, $\angle ADB$ is the largest angle in $\triangle ABD$. By Theorem 10, the side opposite the largest angle is the longest side. $AB$ is opposite the $90^\circ$ angle. Thus, $AB > AD$.
