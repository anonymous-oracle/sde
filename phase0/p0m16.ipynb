**Lexical Geometry Postulates (D-Nodes):**

1. **D-Node 21: Logarithm**
The *Logarithm* is defined as the strictly inverse operation to exponentiation. If a base $a$ raised to a power $x$ equals a number $N$ (i.e., $a^x = N$), then the logarithm of $N$ to the base $a$ is exactly $x$.
*Notation:* $\log_a(N) = x$.
*Constraints:* $a > 0, a \neq 1, N > 0$.
*Status: [UNLOCKED]*

**Supermarket Intuition:**
If exponents ask, "What do I get if I multiply $2$ by itself $3$ times?" ($2^3 = 8$), then logarithms ask the exact opposite: "How many times do I need to multiply $2$ by itself to get $8$?" ($\log_2(8) = 3$). It is a mathematical question seeking the exponent as the answer. Because logarithms are just disguised exponents, they perfectly inherit all the laws of exponents you proved in Module 09.

**Rigorous Derivations (Bottom-Up Traversal):**

1. **Theorem 22: The Product Law ($\log_a(MN) = \log_a(M) + \log_a(N)$)**
* Let $\log_a(M) = x$ and $\log_a(N) = y$.
* By D-Node 21 (definition), translate these into exponential form:
$a^x = M$ and $a^y = N$.
* Multiply $M$ and $N$: $M \cdot N = a^x \cdot a^y$.
* By the Fundamental Law of Exponents (Module 09), $a^x \cdot a^y = a^{x+y}$.
* Therefore, $MN = a^{x+y}$.
* Convert this back to logarithmic form (D-Node 21): $\log_a(MN) = x + y$.
* Substitute $x$ and $y$ back: $\log_a(MN) = \log_a(M) + \log_a(N)$.
*Status: [UNLOCKED]*


2. **Theorem 23: The Quotient Law ($\log_a(\frac{M}{N}) = \log_a(M) - \log_a(N)$)**
* Let $\log_a(M) = x$ and $\log_a(N) = y \implies a^x = M$ and $a^y = N$.
* Divide $M$ by $N$: $\frac{M}{N} = \frac{a^x}{a^y}$.
* By the Law of Exponents, $\frac{a^x}{a^y} = a^{x-y}$.
* Therefore, $\frac{M}{N} = a^{x-y}$.
* Convert to logarithmic form: $\log_a(\frac{M}{N}) = x - y$.
* Substitute $x$ and $y$: $\log_a(\frac{M}{N}) = \log_a(M) - \log_a(N)$.
*Status: [UNLOCKED]*


3. **Theorem 24: The Power Law ($\log_a(M^p) = p \cdot \log_a(M)$)**
* Let $\log_a(M) = x \implies a^x = M$.
* Raise both sides of the exponential equation to the power of $p$: $(a^x)^p = M^p$.
* By the Law of Exponents, $(a^x)^p = a^{xp}$.
* Therefore, $a^{xp} = M^p$.
* Convert to logarithmic form: $\log_a(M^p) = xp$.
* Substitute $x$: $\log_a(M^p) = p \cdot \log_a(M)$.
*Status: [UNLOCKED]*

**Intuition-Building Illustrative Examples (IBIE):**

**IBIE 1: The Base Reversal Trap**

* **[The Problem]:** Evaluate $\log_9(3)$.
* **[The Heuristic Trap]:** Seeing $9$ and $3$ and instantly writing $3$ because $9 / 3 = 3$.
* **[The Failure State]:** Stating $\log_9(3) = 3$. If we check this via D-Node 21, $9^3 = 729$, not $3$.
* **[The Resolution Heuristic]:** Always frame it as $a^x = N$. "What power turns $9$ into $3$?"
* **[Solution Pathway]:** Let $\log_9(3) = x$. Then $9^x = 3$. Since $9 = 3^2$, we write $(3^2)^x = 3^1 \implies 3^{2x} = 3^1$. Therefore, $2x = 1 \implies x = \frac{1}{2}$.

**IBIE 2: The Addition Distribution Trap**

* **[The Problem]:** Expand $\log_a(x + y)$.
* **[The Heuristic Trap]:** Treating the "log" operator like a variable that can be distributed across parenthesis.
* **[The Failure State]:** Writing $\log_a(x + y) = \log_a(x) + \log_a(y)$.
* **[The Resolution Heuristic]:** The Product Law (Theorem 22) dictates that multiplication *inside* the log becomes addition *outside* the log. Addition inside the log has no expansion property.
* **[Solution Pathway]:** $\log_a(x + y)$ cannot be expanded further. It is already in its simplest state.

**IBIE 3: The Unity Trap**

* **[The Problem]:** Solve $\log_7(1) = x$.
* **[The Heuristic Trap]:** Assuming the answer is $1$ or $7$.
* **[The Resolution Heuristic]:** Revert to exponential notation (D-Node 21).
* **[Solution Pathway]:** $7^x = 1$. The only exponent that turns a non-zero base into $1$ is $0$. Therefore, $\log_a(1) = 0$ for any base $a$.

**IBIE 4: The Algebraic Exponent Trap**

* **[The Problem]:** Solve for $x$: $2\log_3(x) = \log_3(25)$.
* **[The Heuristic Trap]:** Dividing by $2$ on both sides without resolving the logarithmic coefficient.
* **[The Failure State]:** $\log_3(x) = \frac{\log_3(25)}{2}$, creating an unresolvable fraction.
* **[The Resolution Heuristic]:** Use the Power Law (Theorem 24) in reverse to pull the coefficient inside the logarithm.
* **[Solution Pathway]:** $2\log_3(x) = \log_3(x^2)$.
Therefore, $\log_3(x^2) = \log_3(25)$.
Since the bases match, the arguments must match: $x^2 = 25 \implies x = 5$ (reject $x = -5$ because by D-Node 21, log arguments must be $> 0$).
