### CORRECTIVE TRACK MODULE 7: SURDS & CENTRAL TENDENCY

#### 1. Intuition (The Supermarket View)
**Surds:** Think of $\sqrt{2}$ as an "irrational" length. You can't write it as a tidy fraction, so we treat it like a variable (like $x$). If you have $2\sqrt{2} + 3\sqrt{2}$, you have $5\sqrt{2}$. It's just a way of maintaining perfect precision in a world of messy numbers.

**Central Tendency:** * **Mean:** The "Fair Share." If everyone pooled their money and split it equally.
* **Median:** The "Middle Person." If everyone stood in a line from poorest to richest, who is in the dead center? This is robust against "outliers" (like a billionaire standing in a line of paupers).
* **Mode:** The "Popular Vote." Which value appears most frequently?

#### 2. Rigorous Theory & Derivations

**A. Laws of Indices (Rational Extension)**
For $a > 0$, we define $a^{m/n} = \sqrt[n]{a^m}$.
* **Rationalizing the Denominator:** If you have $\frac{1}{\sqrt{a} + \sqrt{b}}$, you multiply by the **conjugate** $\frac{\sqrt{a} - \sqrt{b}}{\sqrt{a} - \sqrt{b}}$ to exploit the "Difference of Squares" identity $(x-y)(x+y) = x^2 - y^2$, turning the denominator into a clean integer $a-b$.

**B. Statistics Equations**
* **Mean ($\bar{x}$):** $\frac{\sum x_i}{n}$
* **Median:** The value at position $\frac{n+1}{2}$ for odd $n$. For even $n$, the average of positions $\frac{n}{2}$ and $\frac{n}{2}+1$.
