If you want to write SQL queries, build knowledge graphs, or calculate the probability of an AI model hallucinating, you must understand Sets.

#### 1. Intuition (The Supermarket View)
Imagine a "Set" as a strictly labeled shopping basket. 
* **Basket A** has `{Apple, Banana, Orange}`.
* **Basket B** has `{Banana, Kiwi, Grape}`.

If we want to know what fruits are in *both* baskets, we look for the overlap: just the `{Banana}`. This is the **Intersection**. 
If we dump both baskets on the table to see all our unique fruits, we get `{Apple, Banana, Orange, Kiwi, Grape}`. Notice we don't count the Banana twice; a set only cares about *unique* identities. This is the **Union**.

#### 2. Rigorous Theory & Derivations

Let's formalize this with strict notation.
* **Definition:** A set is a well-defined collection of distinct objects, denoted by capital letters (e.g., $A, B$). 
* **Universal Set ($U$):** The master set containing all possible elements under consideration.
* **Intersection ($A \cap B$):** The set of all elements that belong to both $A$ and $B$. 
    * Algebraic definition: $\{x \mid x \in A \text{ and } x \in B\}$
* **Union ($A \cup B$):** The set of all elements that belong to $A$, or $B$, or both.
    * Algebraic definition: $\{x \mid x \in A \text{ or } x \in B\}$
* **Difference ($A - B$):** Elements in $A$ but strictly *not* in $B$.
* **Complement ($A'$ or $A^c$):** Everything in the Universal Set $U$ that is *not* in $A$. 
    * Algebraic definition: $\{x \mid x \in U \text{ and } x \notin A\}$

**The Cardinality Formula (Principle of Inclusion-Exclusion):**
If $n(A)$ represents the number of elements in set $A$, then to find the total number of unique elements in two overlapping sets, you cannot just add them together, because you double-count the intersection. You must subtract the overlap:
$$n(A \cup B) = n(A) + n(B) - n(A \cap B)$$
