**1. Supermarket Intuition**
When you rent an apartment, you pay the landlord for the privilege of using their physical asset over time. When you borrow money, you are "renting" capital. The rent you pay is called **Interest**. In the financial algorithms you may eventually code for banking sectors, calculating the time-value of money is foundational. Simple Interest is a linear progression: you pay a fixed percentage on the *original* borrowed amount, strictly proportional to the amount of time you hold it. 

**2. Bounded Rigorous Derivation**
Let us formalize the variables:
* **Principal ($P$):** The original baseline capital borrowed or invested.
* **Rate ($R$):** The rent percentage charged per unit of time (standardized as "per annum" or per year).
* **Time ($T$):** The duration for which the capital is held, strictly converted into years to match the Rate's metric.
* **Simple Interest ($SI$):** The total rent accumulated.
* **Amount ($A$):** The final settlement value (Principal + Interest).

**The Foundational Formula:**
Derived directly from our Percentage and Variation modules, the interest is a direct variation of $P$, $R$, and $T$.
$$SI = \frac{P \times R \times T}{100}$$

* **Amount Equation:**
$$A = P + SI = P + \left(\frac{P \times R \times T}{100}\right) = P \left(1 + \frac{R \times T}{100}\right)$$

*Note on Time Conversion (The Unit Mandate):* If time is given in months, $T = \frac{\text{months}}{12}$. If given in days, $T = \frac{\text{days}}{365}$.

### **Module 08, Node 2: Compound Interest (The Exponential Curve)**

**1. Supermarket Intuition**
Simple Interest assumes capital grows linearly; you only ever pay rent on the original baseline. But capital has gravity. If you leave your money in a bank, the bank owes you interest at the end of the year. If you do not withdraw that interest, you are effectively lending the bank *more* money for the second year. Therefore, they must pay you interest on the original capital *and* interest on the previously accumulated interest. This is **Compound Interest**. It transforms linear scaling into an exponential curve. In machine learning, failing to normalize weights in deep neural networks causes "gradient explosion," which is the exact same compounding mathematical phenomenon tearing your algorithm apart.

**2. Bounded Rigorous Derivation**
Let $P$ be Principal, $R$ be Rate per annum, and $n$ be the number of conversion periods.

* **Step-by-Step Method (Without Formula):**
    We iterate using the Simple Interest formula.
    * *Year 1:* $SI_1 = \frac{P \times R \times 1}{100}$. End of Year 1 Amount $A_1 = P + SI_1$.
    * *Year 2:* The Principal for Year 2 is $A_1$. $SI_2 = \frac{A_1 \times R \times 1}{100}$. End of Year 2 Amount $A_2 = A_1 + SI_2$.
    * The total Compound Interest ($CI$) = Final Amount - Original Principal.

* **The Formula Method (Class 9):**
    If we trace the algebra of the step-by-step method:
    $A_1 = P(1 + \frac{R}{100})$.
    $A_2 = A_1(1 + \frac{R}{100}) = P(1 + \frac{R}{100})(1 + \frac{R}{100}) = P(1 + \frac{R}{100})^2$.
    By mathematical induction, for $n$ years:
    $$A = P\left(1 + \frac{R}{100}\right)^n$$
    $$CI = A - P = P\left[\left(1 + \frac{R}{100}\right)^n - 1\right]$$

* **Variable Compounding Periods:**
    If interest is compounded *half-yearly*, the time periods double ($2n$), and the rate per period is halved ($R/2$).
    $$A = P\left(1 + \frac{R/2}{100}\right)^{2n}$$

### **Module 08, Node 3: Banking (Recurring Deposits)**

**Textbook Reference Block (Exhaustive Parallel Citation Protocol v14.2):**
* **Selina Concise Mathematics:** Class 10, Chapter 2: Banking (Recurring Deposit Accounts) -> Page Range: Standard Edition Span.
* **M.L. Aggarwal Understanding ICSE:** Class 10, Chapter 2: Banking -> Page Range: Standard Edition Span.
* *(Note: Other ICSE publishers map this identically in their Class 10 volumes).*

**1. Supermarket Intuition**
Compound Interest assumes you have a large lump sum of capital to invest on day one. But most human beings, and many corporate cash flow systems, generate revenue monthly. If you deposit a fixed amount every single month, calculating the compound interest normally would require 12 separate calculations for a 1-year account, 60 calculations for a 5-year account, etc. This is computationally expensive. Banks circumvent this by using a "Recurring Deposit" (RD) formula, which mathematically compresses all those staggered monthly deposits into a single "Equivalent Principal" for exactly one month.

**2. Bounded Rigorous Derivation**
Let $P$ be the fixed monthly installment deposited.
Let $n$ be the total number of months the account is active.
Let $R$ be the rate of Simple Interest per annum.

* **The Equivalent Principal Logic:**
  The 1st installment earns interest for $n$ months.
  The 2nd installment earns interest for $(n-1)$ months.
  ...The last installment earns interest for $1$ month.
  
  Instead of calculating these separately, we sum the time periods. The sum of the first $n$ natural numbers ($1 + 2 + 3 + \dots + n$) is proven algebraically to be $\frac{n(n+1)}{2}$. 
  Therefore, the "Equivalent Principal for 1 month" is:
  $$P_{\text{eq}} = P \times \frac{n(n+1)}{2}$$

* **The Recurring Deposit Interest Formula:**
  We apply the standard Simple Interest formula to this Equivalent Principal for a time duration of exactly $1$ month (which is $\frac{1}{12}$ of a year, satisfying the Unit Mandate).
  $$I = P_{\text{eq}} \times \frac{R}{100} \times \frac{1}{12}$$
  $$I = P \times \frac{n(n+1)}{2 \times 12} \times \frac{R}{100}$$

* **Maturity Value ($MV$):**
  The final amount the bank owes you is the total money you actually deposited plus the accumulated interest.
  $$MV = (P \times n) + I$$
