### 📈 Challenge 1: Multi-Stage Algebraic Optimization

In quantitative financial engineering and automated machine learning asset allocation, models minimize risk metrics under hard resource constraints.

Let an automated trading agent allocate a total capital pool of exactly $10\text{ lakhs}$ between two discrete algorithmic strategies, Strategy $x$ and Strategy $y$. The total allocation constraint is defined by the linear system:


$$x+y=10$$

The total risk variance $V$ of this combined portfolio is modeled by the multi-variable quadratic cost function:


$$V=2x^2+y^2-4x$$

Our goal is to isolate the exact capital allocation values for both $x$ and $y$ that minimize the overall portfolio risk variance $V$, and compute that minimum variance score.

To begin this optimization without using calculus derivatives, we must first collapse the system into a single-variable quadratic function using our boundary constraint.

What is the resulting single-variable quadratic expression for $V$ in terms of $x$ once you substitute and eliminate $y$?

Let's look at how the capital allocation works between the two strategies. 📊 Setting $y=0$ gives us a look at just one specific scenario where all $10\text{ lakhs}$ go directly to Strategy $x$. To find the absolute minimum risk variance across all possible combinations, we want to keep $y$ as a variable that changes dynamically with $x$.

🧮 Since our total constraint is $x+y=10$, we can rearrange this to express $y$ completely in terms of $x$:


$$y=10-x$$

📈 Let's substitute this full expression $(10-x)$ in place of $y$ in our variance formula:


$$V=2x^2+y^2-4x$$
