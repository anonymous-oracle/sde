## **Phase 0: Elementary Foundations**
### **Module 2: Multiplication, Division, and the Properties of Operations**

#### **1. The Supermarket Intuition**
Imagine you are back in the supermarket. Instead of picking up individual apples one by one, you realize they are sold in pre-packaged bags of $4$. If you put $3$ bags into your cart, you could count every single apple ($1, 2, 3... 12$), but that is inefficient. Multiplication is simply a mathematical shortcut for **repeated addition**. 

Conversely, if you have $12$ apples and want to split them equally among $3$ friends, you are figuring out how many times you can subtract a group of $3$ before you run out. This is **division**.

#### **2. Rigorous Theory: Scaling the Number Line**
Let us formally define these operations on our set of Integers ($\mathbb{Z}$).

* **Multiplication ($\times$ or $\cdot$):** For any integers $a$ and $b$, $a \times b$ means adding $a$ to itself $b$ times. We call $a$ and $b$ the *factors*, and the result the *product*. Geometrically, multiplication *scales* your position on the number line. 
    * *Positive $\times$ Positive:* $3 \times 2 = 2 + 2 + 2 = 6$. (Take $3$ steps of size $2$ to the right).
    * *Positive $\times$ Negative:* $3 \times (-2) = (-2) + (-2) + (-2) = -6$. (Take $3$ steps of size $2$ to the left).
* **The Three Fundamental Axioms of Arithmetic:** To do higher-level algebra, we must accept three foundational rules about how addition and multiplication interact:
    1.  **Commutativity:** Order does not matter.
        * $a + b = b + a$
        * $a \times b = b \times a$
    2.  **Associativity:** Grouping does not matter.
        * $(a + b) + c = a + (b + c)$
        * $(a \times b) \times c = a \times (b \times c)$
    3.  **Distributivity:** Multiplication distributes over addition. This is the most important rule for algebraic expansion.
        * $a \times (b + c) = (a \times b) + (a \times c)$
* **Division ($/$):** Division is the inverse operation of multiplication. Asking "What is $a / b$?" is strictly equivalent to asking "What number $q$ satisfies the equation $b \times q = a$?". We call $q$ the *quotient*.
