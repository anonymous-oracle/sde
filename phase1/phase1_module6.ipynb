### Phase 1 | Module 6: Dynamic Programming & Recursion

We are now combining two concepts you have already mastered: **Recursion** (from Module 4) and **Algorithmic Complexity** (from Module 3). 

Sometimes, recursive algorithms are elegant but secretly hide an incredibly dangerous computational trap. To fix this trap, we use **Dynamic Programming (DP)**.

#### 1. The Fibonacci Trap (Intuition)
Imagine the famous Fibonacci sequence, where every number is the sum of the two before it: 0, 1, 1, 2, 3, 5, 8, 13...
If we write a recursive function to find the 5th Fibonacci number, `fib(5)`, the function says: "I need to calculate `fib(4)` and `fib(3)`."
* But `fib(4)` needs `fib(3)` and `fib(2)`.
* Now the computer is calculating `fib(3)` *twice* in completely different branches of the recursive tree. 

If you ask for `fib(40)`, the computer will calculate `fib(2)` millions of times. This creates an exponential time complexity of $O(2^N)$. Your CPU will grind to an absolute halt.

#### 2. The Dynamic Programming Antidote (Memoization)
"Memoization" is just a fancy computer science word for **writing down the answers on a notepad so you don't have to do the math twice.**

In DP, we introduce a Hash Map (Dictionary) to our recursive function. 
Every time the computer calculates a branch, like `fib(3)`, it writes the answer into the Hash Map: `{3: 2}`. 
The next time the tree asks for `fib(3)`, the computer checks the Hash Map first. It finds the answer instantly ($O(1)$) and completely skips doing the math! We just turned an algorithm that takes billions of years into one that takes milliseconds.

#### 3. The ML Tie-in: The Viterbi Algorithm (Classical ASR)
Why is Dynamic Programming a core skill for a Staff Scientist? Because of **Speech Recognition**.
In Phase 4.5, we will build a Kaldi-style speech pipeline using Hidden Markov Models (HMMs). When an HMM tries to guess what word you spoke from a spectrogram, there are billions of possible word combinations. 

If the computer tried every combination recursively, it would take years to decode a single sentence. Instead, we use the **Viterbi Algorithm**, which is a famous Dynamic Programming algorithm. It calculates the probability of the *best path* up to a certain point, saves that probability in a table (Memoization!), and uses it to rapidly decode the rest of the sentence in real-time.

1. **The Naive Approach:** Write a standard recursive function `def fib_naive(n):`. 
   * Base cases: if `n == 0`, return 0. If `n == 1`, return 1.
   * Recursive step: return `fib_naive(n-1) + fib_naive(n-2)`.
2. **The DP Approach:** Write a DP function `def fib_dp(n, memo={}):`.
   * Check if `n` is already a key in `memo`. If it is, just `return memo[n]`.
   * Base cases: same as above.
   * Recursive step: calculate the result (`fib_dp(n-1, memo) + fib_dp(n-2, memo)`), store it in `memo[n]`, and *then* return it.
3. **The Race:** Import `time`. Time how long it takes to calculate `fib_naive(35)`. Then, time how long it takes to calculate `fib_dp(35)`. 

Run your code, watch your M1 Mac struggle with the naive version and instantly solve the DP version, and paste your from-scratch code and the two execution times!

```python
import time
def fib_naive(n):
    if n==0:
        return 0
    if n==1:
        return 1
    return fib_naive(n-1) + fib_naive(n-2)

def fib_dp(n, memo={}):
    if n in memo:
        return memo[n]
    if n==0:
        return 0
    if n==1:
        return 1
    memo[n] = fib_dp(n-1, memo) + fib_dp(n-2, memo)
    return memo[n]

naive_start = time.time()
print(fib_naive(35))
print(f"time for naive to run: {time.time() - naive_start}")

dp_start = time.time()
print(fib_dp(35))
print(f"time for dp to run: {time.time() - dp_start}")
```
