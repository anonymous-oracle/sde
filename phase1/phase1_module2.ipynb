#### 1. Unrolling the Circle (Sine and Cosine)
Forget triangles for a moment. Let's put a mathematical bug on our Unit Circle at the coordinate $(1, 0)$—the rightmost edge. 

Imagine this bug starts walking counter-clockwise along the perimeter of the circle. We want to track its exact coordinates $(x, y)$ as it walks. 
* The bug's **$x$-coordinate** (how far left/right it is) is called **Cosine** ($\cos$).
* The bug's **$y$-coordinate** (how far up/down it is) is called **Sine** ($\sin$).

As the bug walks from 0 to $\frac{\pi}{2}$ radians (top of the circle), its height ($y$ or $\sin$) goes from 0 up to 1. Its horizontal position ($x$ or $\cos$) goes from 1 down to 0. 

If we "unroll" the bug's journey and plot its height ($y$) over time (radians), we get a perfectly smooth, oscillating line that goes up to 1, down through 0 to -1, and back up. **This is a Sine Wave.**

#### 2. The ML Tie-in: Why do we care?
In the world of Machine Learning, Sine and Cosine waves are the fundamental building blocks of two massive domains:

1. **Audio & Classical ASR (Kaldi/TDNNs):** When you speak, your vocal cords compress the air, creating pressure waves. A microphone turns that physical pressure wave into an electrical sine wave. To build Time-Delay Neural Networks (TDNNs) that recognize speech, we have to process these oscillating waves.
2. **Transformers (LLMs):** A standard neural network has no concept of time or sequence. If you feed it "The dog bit the man," it just sees a bag of words. To fix this, the creators of the Transformer invented **Positional Encodings**. They generate hundreds of sine and cosine waves at different frequencies and add them to the words. This gives every position in a sentence a unique mathematical "timestamp" that the attention mechanism can read!

```python
import numpy as np

def degrees_to_radians(degrees):
    # The core mathematical truth: 180 degrees is exactly equal to Pi radians.
    # Therefore, 1 degree = Pi / 180 radians.
    radians = degrees * (np.pi / 180)
    return radians

# Let's test the 4 main checkpoints of a circle
angles_in_degrees = [0, 90, 180, 270, 360]

print("--- The Unit Circle Angles ---")
for angle in angles_in_degrees:
    rad = degrees_to_radians(angle)
    # We print the result as a clean fraction of Pi for readability
    print(f"{angle:3} degrees = {rad/np.pi:.1f} * Pi radians")
```

```python
import numpy as np
import matplotlib.pyplot as plt

linspaced_radians = np.linspace(0, 4 * np.pi, 100)
y_sin = np.sin(linspaced_radians)
y_cos = np.cos(linspaced_radians)
print(y_sin[:25])
print(y_cos[:25])
plt.plot(y_sin, y_cos)
plt.plot(linspaced_radians, y_sin)
plt.plot(linspaced_radians, y_cos)
```

### Phase 1 | Module 2: Equations, Inequalities & 3D Planes

#### 1. Equations vs. Inequalities (The Budget Intuition)
* **Equation ($=$):** You go to the supermarket with exactly $20. You must spend exactly $20. If apples cost $2 each, the equation is $2x = 20$. The only valid answer is $x = 10$. It is a strict boundary.
* **Inequality ($<, >, \le, \ge$):** You go to the supermarket with $20, but you *don't have to spend it all*. The inequality is $2x \le 20$. Now, $x$ can be 10, but it can also be 9, 8, 3, or 0. The answer isn't a single point; it is a **Region of valid space**.

**The ML Tie-in:** In Machine Learning Optimization (like Support Vector Machines or Linear Programming), we frequently deal with "Constrained Optimization." For example, "Maximize the model's accuracy, *subject to* the inequality that the GPU memory used must be $\le 16$ GB." We are searching for the best answer inside a shaded region of valid possibilities!

#### 2. Extending to 3D: The Plane
You already figured out 3D distance. Now let's look at 3D surfaces.
In 2D, $y = mx + b$ draws a 1-dimensional Line on a flat piece of paper.
If we add a third dimension (the $z$-axis for depth), the equivalent equation is:
$$ax + by + cz = d$$
This does not draw a line. It draws a **2-Dimensional Plane** (like a flat sheet of glass) slicing through a 3D room. 

**The ML Tie-in:** If you want to separate cats from dogs using just their height and weight (2 features), you draw a 1D Line between them. If you add a third feature (e.g., ear length), you are now in 3D space, and your neural network must draw a 2D Plane to separate the cats from the dogs. This separating boundary is called a **Hyperplane**!

### Deep Dive Remediation: Intercepts, Quadratics, & Normal Vectors

#### 1. The Geometry of Intercepts (Addressing Q2)
**The Concept:** An "intercept" is simply where a line or a 3D plane crashes into one of the main axes (x, y, or z). 
**The Intuition:** Imagine you are a drone flying in a 3D room. 
* If you are touching the floor (the x-y plane), your height ($z$) is exactly 0. 
* If you crash directly into the vertical $z$-axis pole, you haven't moved left/right ($x$) and you haven't moved forward/backward ($y$). Therefore, on the $z$-axis, **$x=0$ and $y=0$**.

**The Math:** To find where the plane $2x + 3y + z = 12$ intersects the $z$-axis, you mathematically enforce that reality by plugging in 0 for the other variables:
$$2(0) + 3(0) + z = 12$$
$$0 + 0 + z = 12 \implies z = 12$$
So the exact coordinate in 3D space is **$(0, 0, 12)$**. You guessed $z=12$ intuitively, which was brilliant, but this is the rigorous mathematical *why*.

#### 2. Solving Quadratic Inequalities (Addressing Q3)
You correctly factored the polynomial to find the roots: $-1$ and $5$. But your conclusion ($x < -1$ and $x < 5$) mixed up the logic of how a curve behaves. 

**The Concept:** The inequality $x^2 - 4x - 5 < 0$ is asking: *"For what values of $x$ does this parabola dip BELOW the zero-line (the x-axis)?"*

**The Derivation (The Number Line Test):**
1. Factor to find the roots: $(x - 5)(x + 1) = 0$. The roots are $x = 5$ and $x = -1$. These are the exact points where the curve crosses the zero-line.
2. These two roots slice the x-axis into three distinct regions:
   * **Region A:** $x < -1$ (e.g., -2)
   * **Region B:** between $-1$ and $5$ (e.g., 0)
   * **Region C:** $x > 5$ (e.g., 6)
3. We test one random number from each region by plugging it into $(x - 5)(x + 1)$:
   * Test -2: $(-2 - 5)(-2 + 1) = (-7)(-1) = \mathbf{7}$. (Positive. The curve is ABOVE zero here).
   * Test 0: $(0 - 5)(0 + 1) = (-5)(1) = \mathbf{-5}$. (Negative. The curve is BELOW zero here).
   * Test 6: $(6 - 5)(6 + 1) = (1)(7) = \mathbf{7}$. (Positive. The curve is ABOVE zero here).

**The Conclusion:** The inequality asked where the curve is $< 0$. Our test proves it only happens in the middle region. Therefore, the mathematically correct answer is **$-1 < x < 5$**.

#### 3. The Normal Vector (Addressing Q4)
Even without me teaching you, you guessed: *"The vector is perpendicular to the plane?"* **That is an incredible mathematical deduction, and you are 100% correct.**

**The Concept:** Imagine pitching a flat tent (a plane). To define exactly how that tent is tilted, you just need to know the direction of the main tent pole sticking straight out of it. 
In Linear Algebra, a vector that sticks strictly perpendicular (at a 90-degree angle) to a surface is called the **Normal Vector**.

**The Math:** If a 3D plane has the equation $ax + by + cz = d$, the Normal Vector (let's call it $\vec{n}$) is magically just the coefficients of the variables!
$$\vec{n} = [a, b, c]$$
So for your plane $2x + 3y + z = 12$, the vector pointing straight out of it is exactly $[2, 3, 1]$. In ML, when a Support Vector Machine (SVM) draws a hyperplane to separate data, it learns this exact Normal Vector to determine the angle of the boundary!
