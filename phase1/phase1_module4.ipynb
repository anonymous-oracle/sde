#### 1. Anatomy of a Tree
Imagine an upside-down family tree. 
* **The Root:** The single starting point at the very top.
* **Nodes:** Every individual item in the tree.
* **Edges:** The invisible lines connecting the nodes.
* **Leaves:** The nodes at the very bottom that have no children.

The most common type of tree in computer science is the **Binary Tree**, where every node is allowed a maximum of two children (a "Left" child and a "Right" child).

#### 2. The ML Tie-in: Why Trees?
Trees are everywhere in Machine Learning:
* **Decision Trees & Random Forests:** Classical ML models are literally just massive binary trees where every node is a True/False question (e.g., "Is the pixel value > 128?"), and the leaves are the final predictions.
* **LLM Code Generation:** When an LLM writes Python code for you, it doesn't just guess text; it often builds an Abstract Syntax Tree (AST) under the hood to ensure the code's structural logic is perfectly valid before it translates it back into English characters.
* **Tokenizers (Trie):** Advanced tokenizers often use a specific type of tree called a Trie (Prefix Tree) to rapidly search for sub-words.

#### 3. Your Next Action: Main Track Guided Implementation

Because we are back on the Main Track, you are going to write this from scratch! We will use Object-Oriented Python to build a node, link them together, and use your new recursive skills to read them.

**The Task:**
Open a new Jupyter cell and write pure Python code to do the following:

1. **Define the Node:** Create a Python class called `TreeNode`. 
    * Give it an `__init__(self, value)` method. 
    * Inside the init, store the `value` (`self.value = value`), and initialize `self.left` and `self.right` to `None`.
2. **Build the Tree:** * Instantiate a root node: `root = TreeNode("AI")`.
    * Give the root a left child: `root.left = TreeNode("Machine Learning")`.
    * Give the root a right child: `root.right = TreeNode("Deep Learning")`.
3. **The Recursive Traversal:** Write a standalone function called `print_tree(node)`.
    * **Base Case:** If the `node` is `None`, just `return`.
    * **Action:** Print the `node.value`.
    * **Recursive Step 1:** Call `print_tree(node.left)`.
    * **Recursive Step 2:** Call `print_tree(node.right)`.
4. Finally, call `print_tree(root)` to trigger the whole process.

Run your code, watch the recursion elegantly walk through your tree, and paste your code and output here!

```python
class TreeNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

def print_tree(node):
    if node is None:
        return
    print(node.value)
    print_tree(node.left)
    print_tree(node.right)

root = TreeNode("AI")
root.left = TreeNode("Machine Learning")
root.right = TreeNode("Deep Learning")

print_tree(root)
```
