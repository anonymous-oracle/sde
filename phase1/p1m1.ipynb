#### Step 1: Intuition (The Supermarket Layer)
Imagine you are tracking inventory in a massive warehouse. A simple tally mark system collapses the moment you reach a thousand items. To scale, humanity invented a positional numeral system. A number's position determines its weight. By adding a negative sign, we mirror this system to represent debt or deficit. To manipulate these values consistently without chaos, we establish a strict hierarchy of operations—a grammatical syntax for arithmetic. 

#### Step 2: First-Principles Derivations & Mechanisms
**A. Base-10 Positional Notation:**
Every integer $N$ in our system is a linear combination of the powers of 10.
$$N = d_k 10^k + d_{k-1} 10^{k-1} + \dots + d_1 10^1 + d_0 10^0$$
Where $d_i$ are the digits $\{0, 1, \dots, 9\}$. The "Place Value" of a digit $d_i$ is its literal value multiplied by its positional weight $10^i$.

**B. Set Definitions & Properties:**
* **Natural Numbers:** $\mathbb{N} = \{1, 2, 3, \dots\}$.
* **Whole Numbers:** $\mathbb{W} = \{0, 1, 2, 3, \dots\}$.
* **Integers:** $\mathbb{Z} = \{\dots, -2, -1, 0, 1, 2, \dots\}$.

We must define the structural laws governing these sets:
1.  **Closure Property:** A set is closed under an operation if applying that operation to any two elements yields an element within the same set. For example, if $a, b \in \mathbb{W}$, then $(a + b) \in \mathbb{W}$.
2.  **Commutative Law of Multiplication:** $a \cdot b = b \cdot a$.
3.  **Distributive Law:** Multiplication distributes over addition, unlocking algebraic expansion: $a \cdot (b + c) = a \cdot b + a \cdot c$.

**C. The Syntactical Engine: BODMAS:**
To prevent ambiguity in compound expressions, we enforce BODMAS (Brackets, Orders, Division, Multiplication, Addition, Subtraction). This is a deterministic parsing tree. Without it, an expression like $8 - 2 \cdot 3$ could yield 18 or 2. Under BODMAS, multiplication precedes subtraction, rendering the absolute truth: $8 - 6 = 2$.

The integer line is vast, but it has gaps. What exists between 1 and 2? To quantify fragments—like dividing a pizza or measuring precise lengths—we invented fractions. But fractions are messy; comparing $3/7$ and $4/9$ isn't intuitively obvious. So, we developed a system to standardise these fragments using common denominators or base-10 decimals. 

Furthermore, as multiplication is just repeated addition, what happens when we repeatedly multiply the same number? Writing $2 \times 2 \times 2 \times 2 \times 2$ is inefficient. We need a higher-order syntactical engine to compress this notation, leading to the invention of Exponents.

#### Step 2: First-Principles Derivations & Mechanisms

**A. Rational Numbers:**
We define a new set, the Rational Numbers ($\mathbb{Q}$), as any number that can be expressed as a ratio of two integers.
$$\mathbb{Q} = \left\{ \frac{p}{q} \mid p, q \in \mathbb{Z}, q \neq 0 \right\}$$
*Derivation of the $q \neq 0$ constraint:* Why can't we divide by zero? If $p/0 = c$, then by inverse operations, $0 \cdot c = p$. If $p$ is any non-zero number (say, 5), there is no number $c$ that makes $0 \cdot c = 5$. It breaks the fundamental logic of multiplication, hence it is strictly "undefined".

**B. Fractional Operations & The LCM:**
To add or subtract fractions, you cannot simply combine numerators and denominators. They must share a common base of measurement.
To compute $\frac{a}{b} + \frac{c}{d}$, we must find a common denominator, ideally the Least Common Multiple (LCM) of $b$ and $d$. Algebraically, the cross-multiplication derivation is:
$$\frac{a}{b} + \frac{c}{d} = \frac{ad + bc}{bd}$$

**C. The Laws of Exponents:**
Exponents denote repeated multiplication: $a^n = a \cdot a \cdot a \dots$ ($n$ times). From this basic definition, we derive the fundamental laws:

1.  **Product Law:** $a^m \cdot a^n = a^{m+n}$
    *Proof:* $(a \cdot a \dots m \text{ times}) \cdot (a \cdot a \dots n \text{ times}) = a \cdot a \dots (m+n) \text{ times}$.
2.  **Quotient Law:** $\frac{a^m}{a^n} = a^{m-n}$ (for $a \neq 0$)
3.  **Power of a Power:** $(a^m)^n = a^{m \cdot n}$
4.  **The Zero Exponent Identity:** $a^0 = 1$
    *Proof:* By the quotient law, $\frac{a^m}{a^m} = a^{m-m} = a^0$. Since any non-zero quantity divided by itself is 1, it dictates that $a^0 = 1$.

You know it's 3, but how do we isolate $x$ algebraically when it is trapped in the superscript? We must invent a new operational inverse: the Logarithm. 

#### Step 2: First-Principles Derivations & Mechanisms

**A. The Real Number System & Rationalization:**
The union of Rational numbers ($\mathbb{Q}$) and Irrational numbers ($\mathbb{I}$) forms the set of Real Numbers ($\mathbb{R}$). 
When dealing with irrational "surds" (like $\sqrt{a}$) in the denominator of a fraction, we execute a protocol called **Rationalization** to shift the irrationality to the numerator for cleaner computation. We use the algebraic identity $(a - b)(a + b) = a^2 - b^2$.
*Mechanism:* To rationalize $\frac{1}{\sqrt{a} + \sqrt{b}}$, we multiply the numerator and denominator by the conjugate ($\sqrt{a} - \sqrt{b}$):
$$\frac{1}{\sqrt{a} + \sqrt{b}} \cdot \frac{\sqrt{a} - \sqrt{b}}{\sqrt{a} - \sqrt{b}} = \frac{\sqrt{a} - \sqrt{b}}{a - b}$$

**B. Fractional Indices:**
We extend the laws of exponents to include rational fractions. 
*Definition:* $x^{m/n} = \sqrt[n]{x^m} = (\sqrt[n]{x})^m$. The denominator $n$ dictates the root, and the numerator $m$ dictates the power.

**C. Logarithms (The Inverse of the Index):**
The logarithm answers the question: *"To what power must I raise the base to get this number?"*
*Fundamental Definition:* If $a^x = N$ (where $a > 0$, $a \neq 1$), then $\log_a N = x$.

From this definition, we derive the core **Laws of Logarithms**:
1. **The Product Law:** $\log_a(M \cdot N) = \log_a M + \log_a N$
   *Rigorous Proof:* Let $\log_a M = x \implies a^x = M$. Let $\log_a N = y \implies a^y = N$. 
   Multiply them: $M \cdot N = a^x \cdot a^y = a^{x+y}$. 
   Convert back to logarithmic form: $\log_a(M \cdot N) = x + y$. 
   Substitute $x$ and $y$ back: $\log_a(M \cdot N) = \log_a M + \log_a N$.
2. **The Quotient Law:** $\log_a\left(\frac{M}{N}\right) = \log_a M - \log_a N$
3. **The Power Law:** $\log_a(M^n) = n \log_a M$

Total Interest ($I$) is the sum of an arithmetic progression ($n + (n-1) + \dots + 1 = \frac{n(n+1)}{2}$).
$$I = P \times \frac{n(n+1)}{2 \times 12} \times \frac{R}{100}$$
Maturity Value $= (\text{Total deposited}) + I = P \cdot n + I$.

* **Investment:** $\text{Number of Shares} \times \text{Market Value (MV)}$.
* **Income (Dividend):** $\text{Number of Shares} \times \text{Nominal Value (NV)} \times \text{Dividend Percentage}$. 
    *(Crucial logic check: Dividends are ALWAYS calculated on the NV, never the MV).*
* **Return on Investment (%):** $\left( \frac{\text{Total Income}}{\text{Total Investment}} \right) \times 100$.

#### Step 2: First-Principles Derivations & Mechanisms

**B. The Unitary Method:**
If $n$ items cost $C$, the cost of 1 item is $C/n$. The cost of $k$ items is therefore $k \cdot (C/n)$. This relies directly on the assumption of direct variation (if quantity increases, cost increases linearly).

**C. Percentage and Commercial Syntax:**
* **Percent ($x\%$):** Strictly defined as the fraction $x/100$.
* **Profit / Loss:** Always calculated on the Cost Price (CP), not the Selling Price (SP). 
    * $\text{Profit \%} = \left(\frac{\text{SP} - \text{CP}}{\text{CP}}\right) \times 100$.
* **Simple Interest (SI):** Interest calculated strictly on the original principal year after year. 
    * Derivation: If interest $R\%$ is charged on Principal $P$ for 1 year, interest is $P \cdot (R/100)$. Over $T$ years, it scales linearly: $SI = \frac{P \cdot R \cdot T}{100}$.

#### Step 1: Intuition (The Supermarket Layer)
Simple interest is linear; you earn money strictly on your original deposit. But in reality, wealth generates wealth. If your interest earns interest, the growth curve bends upwards. This is Compound Interest. It is the mathematical engine of global finance.

#### Step 2: First-Principles Derivations & Mechanisms

**A. Compound Interest (The Recursive Engine):**
* **Class 8/9 Method (Without Formula):** We calculate Simple Interest year by year. The Amount at the end of Year 1 becomes the strictly enforced Principal for Year 2. 
* **Class 9 Method (The Exponential Formula):**
    Let Principal be $P$, Rate be $R\%$, and Time be $n$ years.
    Year 1 Amount: $A_1 = P + P\left(\frac{R}{100}\right) = P\left(1 + \frac{R}{100}\right)$
    Year 2 Amount: $A_2 = A_1\left(1 + \frac{R}{100}\right) = P\left(1 + \frac{R}{100}\right)\left(1 + \frac{R}{100}\right) = P\left(1 + \frac{R}{100}\right)^2$
    **General Formula:** $A = P\left(1 + \frac{R}{100}\right)^n$
    *Compound Interest (CI)* is therefore $A - P$.

**B. Advanced Proportion Laws (Class 10 - Componendo & Dividendo):**
If we are given the proportion $\frac{a}{b} = \frac{c}{d}$, we can derive powerful structural properties:
1.  **Componendo:** Add 1 to both sides. $\frac{a}{b} + 1 = \frac{c}{d} + 1 \implies \frac{a+b}{b} = \frac{c+d}{d}$
2.  **Dividendo:** Subtract 1 from both sides. $\frac{a}{b} - 1 = \frac{c}{d} - 1 \implies \frac{a-b}{b} = \frac{c-d}{d}$
3.  **Componendo and Dividendo (C&D):** Divide the Componendo result by the Dividendo result.
    $$\frac{a+b}{a-b} = \frac{c+d}{c-d}$$
    *Mechanism:* This law is an algebraic chainsaw. It allows you to instantly simplify fractions that look like $\frac{\sqrt{x} + 2}{\sqrt{x} - 2}$.

**C. Inverse Variation (Time and Work):**
If $x$ and $y$ are inversely related (e.g., more workers = less time), their *product* is constant: $x_1 y_1 = x_2 y_2$. If $A$ can do a job in $n$ days, A's 1-day work is strictly $1/n$.
