### **Phase 1 Capstone Project: The Brute-Force Gradient Descent Engine**
**Your Task:** You said that to find the minimum of a valley, you would "step to check which side the slope is downhill." Let's code that exact intuition from scratch in Python.

We have a loss function (a parabola): $y = x^2 - 8x + 16$. 
*(Hint: if you factor this, you'll find the minimum is exactly at $x = 4$, where $y = 0$. We will use your script to prove this!)*

**Requirements (Do not use any external math libraries other than standard Python):**
1.  Write a Python function `calculate_loss(x)` that returns the value of $x^2 - 8x + 16$.
2.  Write a loop that starts your "blindfolded hiker" at `current_x = 10`.
3.  Set a `step_size = 0.1`.
4.  Inside the loop, compare the loss at `current_x` against the loss at `current_x - step_size`.
5.  If stepping to the left (`current_x - step_size`) makes the loss *go down*, update `current_x` to take that step. 
6.  If the loss stops going down (i.e., stepping left makes the loss go up or stay flat), **break** the loop—you've hit the bottom of the valley!
7.  Print the final `current_x` and the final loss.


```python
def calculate_loss(x):
    return x**2 - 8*x + 16

current_x = 10
step_size = 0.1
current_loss = calculate_loss(current_x)
while True:
    new_x = current_x - step_size
    new_loss = calculate_loss(new_x)
    if current_loss > new_loss:
        current_x = new_x
        current_loss = new_loss
    else:
        break
print(current_x)
```
