## **Phase 1: Algebra & Geometry Bridge**
### **Module 2: Inequalities and the Concept of Bounds**

#### **1. The Supermarket Intuition**
In Module 1, we balanced the scale perfectly. But real life rarely requires perfect equality. Suppose you walk into the supermarket with exactly $\$20$ in your pocket. You don't *have* to spend exactly $\$20$. You can spend $\$5$, $\$15$, or $\$19.99$. 

Your total bill, let's call it $x$, simply has a ceiling. It is bounded. We write this mathematically as $x \le 20$. Inequalities are the mathematics of boundaries and constraints.

#### **2. Rigorous Theory: The Rules of Inequalities**
We use four main symbols:
* $<$ (strictly less than)
* $>$ (strictly greater than)
* $\le$ (less than or equal to)
* $\ge$ (greater than or equal to)

The rules for manipulating inequalities are almost identical to equations (you can add or subtract the same number from both sides safely). **However, there is one fatal trap.**

**The Multiplication/Division Trap:**
If you multiply or divide an inequality by a **negative number**, you MUST flip the inequality sign. 
* *Why? Let's prove it geometrically.* Take a simple truth: $3 < 5$. 
    On the number line, $3$ is to the left of $5$. 
    Now, multiply both sides by $-1$. We get $-3$ and $-5$. 
    Geometrically, multiplying by a negative number acts as a **mirror reflection** across the Origin ($0$). Because the entire number line is reflected, the relative positions are reversed. Now, $-3$ is to the *right* of $-5$. Therefore, $-3 > -5$.

#### **3. Machine Learning Tie-In: Constraints**
Why do we care about bounds? In Machine Learning, we constantly use inequalities to constrain our models. For example, a probability $P(x)$ must always satisfy the boundary $0 \le P(x) \le 1$. If your algorithm produces a probability of $1.2$, you have violated the bounds of reality!
