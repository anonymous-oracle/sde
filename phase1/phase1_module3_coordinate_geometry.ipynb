## **Phase 1: Algebra & Geometry Bridge**
### **Module 3: The Cartesian Plane and Pythagoras (Building the 2D World)**

#### **1. The Supermarket Intuition (Moving to 2D)**
So far, our entire mathematical universe has existed on a single, straight supermarket aisle (the 1D number line). You could only move forward (positive) or backward (negative). 

But what if you aren't just a shopper walking down an aisle? What if you are a bird flying looking down at the whole city? To describe a location in a city, you need *two* pieces of information: an Avenue (East/West) and a Street (North/South). 

In the 17th century, a mathematician named René Descartes was staring at his ceiling, watching a fly crawl around. He realized he could describe the fly's exact position by measuring its distance from two intersecting walls. This realization birthed the **Cartesian Coordinate System**.

#### **2. Rigorous Theory: Building the Plane**
Let us formally construct this 2D space.
1.  **The $x$-axis:** We take our standard horizontal number line from Phase 0. We call this the $x$-axis.
2.  **The $y$-axis:** We take a *second* number line, rotate it exactly $90^\circ$ (perpendicular), and lay it across the first one. We call this the $y$-axis.
3.  **The Origin $(0,0)$:** The exact point where these two lines cross is our new anchor for the universe. Both $x$ and $y$ are zero here.
4.  **Coordinates $(x, y)$:** Any point in this space is defined by an ordered pair. $(3, 4)$ means "start at the Origin, walk $3$ units Right (on the $x$-axis), and $4$ units Up (parallel to the $y$-axis)."

#### **3. The Pythagorean Theorem (The Geometry of Distance)**
Now, suppose you are at the Origin $(0,0)$ and you want to walk to the point $(3,4)$. 
You could walk 3 blocks East, turn 90 degrees, and walk 4 blocks North. Total distance = 7 blocks. (This is actually called "Manhattan Distance" in Machine Learning!).

But what if you can fly directly there like a bird? What is the *shortest* straight-line distance? 
To find this, we use the most famous theorem in geometry, courtesy of the ancient Greeks.

**The Theorem:** In any right-angled triangle, the square of the longest side (the hypotenuse, let's call it $c$) is strictly equal to the sum of the squares of the other two sides ($a$ and $b$).
$$a^2 + b^2 = c^2$$

If your horizontal movement is $a = 3$, and your vertical movement is $b = 4$, then your direct flight path $c$ is:
$$3^2 + 4^2 = c^2$$
$$9 + 16 = c^2$$
$$25 = c^2$$
To solve for $c$, we take the square root of both sides. Since a physical distance must be positive:
$$c = 5$$
