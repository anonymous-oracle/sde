## **Phase 1: Algebra & Geometry Bridge**
### **Module 3: The Cartesian Plane, Pythagoras, and Coordinate Geometry**

#### **1. The Supermarket Intuition**
Imagine the supermarket is laid out on a perfectly square grid of floor tiles. To tell someone exactly where the milk is, you need two pieces of information: how many tiles to walk East/West, and how many tiles to walk North/South. 

If you walk 3 tiles East and 4 tiles North, you are at a specific location. But what if you wanted to know the direct, straight-line distance from your starting point to the milk? You wouldn't count the tiles East and then North; you would draw a diagonal line cutting right across the grid. We need a mathematical way to calculate that diagonal.

#### **2. Rigorous Theory: Building the Grid**
We will now combine algebra with geometry by taking our 1-Dimensional Number Line from Phase 0 and crossing it with a second one.

* **The Cartesian Plane:** We place a horizontal number line (the **X-axis**) and a vertical number line (the **Y-axis**) so they intersect at exactly $0$ for both. This intersection is the **Origin** $(0, 0)$. Any point in this 2D space is an ordered pair $(x, y)$.
* **Pythagoras' Theorem:** In any triangle where one angle is a perfect $90^\circ$ (a right angle), the square of the longest side (the hypotenuse, $c$) is strictly equal to the sum of the squares of the two shorter sides ($a$ and $b$). 
    $$a^2 + b^2 = c^2$$
* **The Distance Formula:** Let’s find the straight-line distance $d$ between any two points $A(x_1, y_1)$ and $B(x_2, y_2)$. If we draw a horizontal line from $A$ and a vertical line from $B$, they form a right-angled triangle.
    * The horizontal base $a$ has a length of $(x_2 - x_1)$.
    * The vertical height $b$ has a length of $(y_2 - y_1)$.
    * Using Pythagoras: $d^2 = (x_2 - x_1)^2 + (y_2 - y_1)^2$.
    * Taking the square root gives us our universal distance formula:
        $$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$
* **The Equation of a Straight Line:** A straight line is simply a set of points that all share the exact same steepness. 
    * **Slope ($m$):** The steepness is the ratio of vertical change (rise) to horizontal change (run) between any two points on the line. 
        $$m = \frac{y_2 - y_1}{x_2 - x_1}$$
    * **Y-Intercept ($b$):** The exact point where the line crashes through the vertical Y-axis (where $x = 0$).
    * **Standard Form:** This gives us the foundational equation for any linear relationship: 
        $$y = mx + b$$
