## Big-O notation

```python
import time, random
my_list = [random.randint(1, 100000) for _ in range(5000)]
def check_target(arr, target):
    for i in arr:
        if target==i:
            return True
    return False

def check_duplicates(arr):
    for i in range(len(arr)):
        for j in range(len(arr)):
            if i!=j and arr[i]==arr[j]:
                return True
    return False

target_start = time.time()
check_target(my_list, 3876)
print(time.time() - target_start)

duplicate_start = time.time()
check_duplicates(my_list)
print(time.time() - duplicate_start)
```

```python
import time
vocab_dict = {"robot":439, "apple":46, "learn":894, "monarch":704, "ashborn":859}
vocab_list = list(vocab_dict.copy().keys())
iter_start = time.time()
word = "monarch"
for w in vocab_list:
    if w==word:
        print(w)
        break
print(time.time() - iter_start)

hash_start = time.time()
print(vocab_dict[word])
print(time.time() - hash_start)
```
