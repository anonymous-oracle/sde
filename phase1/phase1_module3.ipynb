## Big-O notation

### Phase 1 | Module 3: Algorithmic Complexity

Before we start building complex Deep Learning architectures, we need to understand how algorithms scale. Large Language Models have a famous, fatal flaw: the longer the text prompt you give them, the slower they get. But it's not a steady, linear slowdown—it's an exponential explosion. 

To understand why this happens, we need to learn **Big-O Notation**.

#### 1. The Supermarket Intuition
Imagine you are given a stack of $N$ unorganized test papers and asked to find a specific student's test. You have to look at the first one, then the second, then the third... up to $N$ times. 
* If there are 100 papers, it takes at most 100 checks. 
* If there are 1,000 papers, it takes 1,000 checks.
* The time it takes scales perfectly evenly with the amount of data. We call this **$O(N)$** or "Linear Time".

#### 2. The $O(N^2)$ Trap
Now imagine a much harder task: you are asked to check if *any two students* in that stack of $N$ papers have the exact same test score. 
* You take the first paper, and compare it to the remaining 99. 
* Then you take the second paper, and compare it to the remaining 98. 
* For $N$ papers, you are doing roughly $N \times N$ comparisons. 

This is **$O(N^2)$** or "Quadratic Time". 
If $N=100$, you do about 10,000 checks. But if $N=100,000$ (the size of a large document prompt in an LLM), an $O(N^2)$ algorithm requires **10,000,000,000** checks! 

This is the exact mathematical bottleneck of **Transformer Self-Attention**. In an LLM, every single word has to "look" at every other single word to understand the context. If you double the prompt length, you quadruple the compute time.

3. Write a function called `check_target(arr, target)` that loops through the array once to see if the `target` number exists. (This is $O(N)$).
4. Write a function called `check_duplicates(arr)` that uses a **nested for-loop** to check if any two items in the array are identical. (This is $O(N^2)$).
5. Use `time.time()` to record the start time, run your $O(N)$ function, record the end time, and print the difference.

Run the code, watch your computer sweat during the second function, and paste your code and the printed times here!

```python
import time, random
my_list = [random.randint(1, 100000) for _ in range(5000)]
def check_target(arr, target):
    for i in arr:
        if target==i:
            return True
    return False

def check_duplicates(arr):
    for i in range(len(arr)):
        for j in range(len(arr)):
            if i!=j and arr[i]==arr[j]:
                return True
    return False

target_start = time.time()
check_target(my_list, 3876)
print(time.time() - target_start)

duplicate_start = time.time()
check_duplicates(my_list)
print(time.time() - duplicate_start)
```
