### Phase 1 | Module 5: Graph Theory Fundamentals

We have reached a massive turning point in computer science. Arrays are flat. Trees branch out in one direction. But the real world is messy, interconnected, and circular. To model the real world, we need **Graphs**.

#### 1. The Connect-The-Dots Intuition
Imagine a map of a train system, or a social network.
* **Nodes (or Vertices):** The individual entities. On a map, these are the train stations. On Facebook, these are the users.
* **Edges (or Links):** The connections between them. The train tracks, or the "friend" relationship.

Graphs come in a few flavors:
* **Directed vs. Undirected:** On Instagram, I can follow you without you following me (Directed—a one-way street). On Facebook, if we are friends, the connection goes both ways (Undirected—a two-way street).
* **Weighted vs. Unweighted:** If the graph is a map, the edges might have "weights" representing the distance in miles between cities. 

#### 2. The ML Tie-in: GNNs and Knowledge Graphs
Why is a Staff Scientist learning connect-the-dots?
1. **Graph Neural Networks (GNNs):** Standard neural networks process grids of pixels (images) or sequences of words (text). But what if your data is a chemical molecule, or a massive social network? GNNs are state-of-the-art models designed specifically to learn from graph structures.
2. **Knowledge Graphs:** Modern LLMs hallucinate. To fix this, we plug them into Knowledge Graphs (like Wikipedia's Wikidata). The graph strictly defines relationships (e.g., `[Paris] --(is_capital_of)--> [France]`), giving the LLM rigid factual guardrails. 
3. **Classical ASR (WFSTs):** Later in Phase 4.5, we will build Weighted Finite State Transducers to decode speech. A WFST is literally just a massive, weighted, directed graph!

#### 3. How do computers see Graphs? (Adjacency Lists)
A computer can't "look" at a drawing of a graph. We have to translate it into data structures you already know: Hash Maps (Dictionaries) and Arrays (Lists).

The most common way to represent a graph is an **Adjacency List**.
We use a Python Dictionary where the `keys` are the Nodes, and the `values` are Lists of the nodes they connect to.
If Node A is connected to Node B, it looks like this:
`graph = {'A': ['B'], 'B': ['A']}`

```python
class SimpleGraph:
    def __init__(self):
        self.adjacency_list = {}
    def add_node(self, node):
        if node not in self.adjacency_list:
            self.adjacency_list[node] = []
    def add_edge(self, node1, node2):
        self.add_node(node1)
        self.add_node(node2)
        # adding undirected edge
        self.adjacency_list[node1].append(node2)
        self.adjacency_list[node2].append(node1)

    # "editor.tabCompletion": "on",
    # "editor.wordBasedSuggestions": "matchingDocuments",
    # "python.languageServer": "Pylance",
    # "jupyter.enableExtendedPythonKernelCompletions": true,
graph = SimpleGraph()
graph.add_node("Alice")
graph.add_node("Bob")
graph.add_node("Charlie")
graph.add_edge("Alice", "Bob")
graph.add_edge("Bob", "Charlie")
print(graph.adjacency_list)
```
