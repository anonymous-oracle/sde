### Phase 2 | Module 2: Matrices & Linear Transformations

If a Vector is a single 1D list of numbers, a **Matrix** is a 2D grid of numbers. 

#### 1. What is a Matrix geometrically?
We often think of matrices just as spreadsheets of data. But in Linear Algebra, a Matrix is an action. It is a **Linear Transformation**. 
Imagine you draw a vector on a piece of graph paper. If you multiply that vector by a Matrix, the Matrix literally "grabs" the graph paper and stretches, squishes, or rotates the entire space. 

#### 2. Matrix Multiplication (The Neural Network Engine)
How do we multiply a Matrix by a Vector (or another Matrix)? It is simply a series of **Dot Products**!

To multiply Matrix $W$ by Vector $x$, you take the dot product of the *first row* of $W$ with $x$, then the *second row* of $W$ with $x$, and so on.

**The Rule of Dimensions:**
You cannot multiply just any two matrices together. The inner dimensions must match.
If Matrix $A$ has a shape of $(Rows \times Columns)$, let's say $(3 \times 2)$, and Matrix $B$ has a shape of $(2 \times 4)$:
* The inner dimensions match ($2$ and $2$). The multiplication is allowed.
* The output Matrix will magically have the shape of the *outer* dimensions: $(3 \times 4)$.

#### 3. The ML Tie-in: Forward Pass
In a Neural Network, a "Dense Layer" is literally just a Matrix of weights (let's call it $W$). Your input data is a vector (let's call it $x$). 
When data passes through the network, the computer calculates $y = Wx$. The network is just mathematically stretching and rotating your data until it separates the "cats" from the "dogs"!
