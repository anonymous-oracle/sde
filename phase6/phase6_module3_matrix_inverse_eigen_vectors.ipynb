### Phase 2 | Module 3: Matrix Inverse, Eigenvectors & Eigenvalues

Now that we know matrices transform (stretch, rotate, squish) space, we need to ask two critical questions: Can we reverse the transformation? And what parts of the space didn't get knocked off their original path?

#### 1. The Matrix Inverse (The "Undo" Button)
If Matrix $A$ rotates your data 90 degrees to the left, is there a matrix that rotates it 90 degrees to the right, putting everything back where it started? 
Yes! It is called the **Inverse**, written as $A^{-1}$.
$$A \cdot A^{-1} = I$$
*(Note: $I$ is the Identity Matrix, which is the equivalent of the number 1 in matrix math—it does nothing to your data).*

**The Catch (Determinant):** You can't always hit undo. If a matrix perfectly flattens 2D graph paper into a 1D straight line (this means its **Determinant is 0**), you have lost information permanently. You cannot mathematically "un-squish" a line back into a 2D plane. We call this a *Singular* or *Non-invertible* matrix.

#### 2. Eigenvectors & Eigenvalues (The Skeleton of the Matrix)
When a matrix transforms space, almost every vector gets rotated and stretched. 
However, there are a few special, magical vectors that **do not rotate**. They stay pointing in their exact original direction; they only get stretched or squished. 

* The vector that refuses to rotate is called an **Eigenvector**.
* The factor by which it stretches or shrinks is called its **Eigenvalue** ($\lambda$).

$$A \cdot v = \lambda \cdot v$$
*(Multiplying the matrix $A$ by the eigenvector $v$ does the exact same thing as just scaling $v$ by a simple number $\lambda$)*.

**The ML Tie-in:** Why do we care? In Machine Learning, data is often massively high-dimensional (like 10,000-dimensional user profiles). If we find the Eigenvectors of our data's covariance matrix, we find the "Principal Components"—the most important structural pillars of our data. We can use this to compress 10,000 dimensions down to 50 dimensions without losing the core information (an algorithm called PCA)!

### Deep Dive: Unboxing the Matrix Algebra

#### 1. The Manual 2x2 Inverse
To reverse a 2x2 matrix, we need two things: the **Determinant** (to scale it) and the **Adjugate** (to rearrange it).

Let our Matrix be: $A = \begin{bmatrix} a & b \\ c & d \end{bmatrix}$

**Step 1: The Determinant ($|A|$ or $\det(A)$)**
Geometrically, the determinant is the factor by which the matrix scales the area of the space. 
The formula is simple cross-multiplication:
$$\det(A) = (a \cdot d) - (b \cdot c)$$
*(If this equals 0, the area was squished to a 1D line, and no inverse exists!)*


**Step 2: The Adjugate & Final Formula**
To build the inverse, we swap the main diagonal ($a$ and $d$), flip the signs of the off-diagonal ($b$ and $c$), and divide everything by the determinant.
$$A^{-1} = \frac{1}{\det(A)} \begin{bmatrix} d & -b \\ -c & a \end{bmatrix}$$

Let's test this on your matrix: $A = \begin{bmatrix} 4 & 1 \\ 2 & 3 \end{bmatrix}$
* $\det(A) = (4 \cdot 3) - (1 \cdot 2) = 12 - 2 = 10$.
* Swap and flip: $\begin{bmatrix} 3 & -1 \\ -2 & 4 \end{bmatrix}$.
* Divide by 10: $\begin{bmatrix} 0.3 & -0.1 \\ -0.2 & 0.4 \end{bmatrix}$. 
*(This perfectly matches the background output of your `np.linalg.inv` call!)*

#### 2. Manually finding Eigenvalues (The Characteristic Equation)
Remember the core rule: $A \cdot v = \lambda \cdot v$.
We want to find the magic $\lambda$ (Eigenvalue) and $v$ (Eigenvector). 

By using the Identity Matrix ($I = \begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix}$), we can rewrite the equation as:
$$(A - \lambda I)v = 0$$

For this to have a non-zero vector $v$, the matrix $(A - \lambda I)$ must squish space completely. This means its **Determinant must be exactly 0**.
$$\det(A - \lambda I) = 0$$
This is called the **Characteristic Equation**.

Let's solve it for your matrix $A = \begin{bmatrix} 4 & 1 \\ 2 & 3 \end{bmatrix}$:
1. Subtract $\lambda$ from the main diagonal: $\det\left(\begin{bmatrix} 4-\lambda & 1 \\ 2 & 3-\lambda \end{bmatrix}\right) = 0$
2. Cross-multiply (Determinant formula): $(4-\lambda)(3-\lambda) - (1 \cdot 2) = 0$
3. Expand: $12 - 4\lambda - 3\lambda + \lambda^2 - 2 = 0$
4. Simplify to a quadratic equation: $\lambda^2 - 7\lambda + 10 = 0$
5. Factor it: $(\lambda - 5)(\lambda - 2) = 0$

Therefore, the exact Eigenvalues are **$\lambda = 5$** and **$\lambda = 2$**. *(Exactly what NumPy printed!)*
