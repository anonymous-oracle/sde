### Combined Mastery Side Quest: Probability & Information Theory

#### Part 1: Basic Probability (The Coin & The Die)
Before we can measure "Information," we have to measure uncertainty. 
* **The Intuition:** If you flip a perfectly fair coin, there are 2 possible outcomes (Heads or Tails). You want to know the chance of getting Heads (1 target outcome).
* **The Math:** Probability ($P$) is simply the number of target outcomes divided by the total number of possible outcomes.
  $$P(\text{Heads}) = \frac{1}{2} = 0.5$$

In Machine Learning, a neural network's final layer (the Softmax layer we discussed earlier) outputs an array of these probabilities. E.g., $P(\text{Cat}) = 0.8$, $P(\text{Dog}) = 0.2$. The sum of all probabilities in a closed system must always equal exactly **$1.0$** (100%).

#### Part 2: What is "Information"? (Claude Shannon's Genius)
In 1948, a mathematician named Claude Shannon asked a bizarre question: *How do we mathematically measure a "surprise"?*
* **Low Surprise (Low Information):** If I tell you "The sun rose in the east today," you learn nothing new. You were 100% sure it was going to happen ($P=1.0$). Therefore, this message contains **0 bits of information**.
* **High Surprise (High Information):** If I tell you "A meteor just crashed through your roof," you are incredibly surprised. The probability of that happening was almost zero ($P=0.000001$). Therefore, this message contains a **massive amount of information**.

**The Mathematical Formula for "Surprise" (Self-Information):**
Remember our Logarithms side quest? We use the base-2 logarithm ($\log_2$) because computers process data in Binary (0s and 1s, called "bits").
To calculate the Information $I$ of an event $x$, we take the negative logarithm of its probability:
$$I(x) = -\log_2(P(x))$$
*(Note: We make it negative because probabilities are decimals between 0 and 1, and the log of a decimal is a negative number. The double negative makes our final "bits" positive!)*

#### Part 3: Entropy (The Average Surprise)
If Information is the surprise of *one* specific event, **Entropy** (denoted by $H$) is the *average surprise* of the entire system. 
Imagine a rigged coin that lands on Heads 99% of the time. You are never really surprised by the outcome. The system has very **Low Entropy**. 
But a perfectly fair coin (50/50) is completely unpredictable. It has **Maximum Entropy**.

**The ML Tie-in: Cross-Entropy Loss**
When a Large Language Model is trying to guess the next word, it outputs a probability distribution. If the correct next word is "apple", but the model was 99% sure the next word was "robot", the model receives a massive mathematical "surprise" penalty. This penalty is called **Cross-Entropy Loss**, and the model uses Calculus (the gradients you calculated in our first side quest!) to update its weights and minimize that surprise for next time.

```python
import math

def calculate_information(probability):
    # If the probability is 0, the surprise is technically infinite. 
    # We catch this to prevent a math error in Python.
    if probability <= 0:
        return float('inf')
    
    # Claude Shannon's formula: I(x) = -log2(P(x))
    # This tells us exactly how many "bits" of information the event contains.
    bits_of_info = -math.log2(probability)
    return bits_of_info

# Scenario 1: The Sun Rising (100% Probability)
p_sun = 1.0
print(f"Information in the sun rising: {calculate_information(p_sun)} bits")

# Scenario 2: Flipping a perfectly fair coin (50% Probability)
p_coin = 0.5
print(f"Information in a coin flip: {calculate_information(p_coin)} bits")

# Scenario 3: A highly unlikely event (1 in 1024 chance)
p_rare = 1 / 1024
print(f"Information in a rare event: {calculate_information(p_rare)} bits")
```
