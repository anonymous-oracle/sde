## **Phase 2: Precalculus Bridge**
### **Module 1: Functions (The Mathematical Machine)**

#### **1. The Factory Intuition**
Up until now, we have dealt with static equations ($y = 2x + 3$). Now, we need to think dynamically. 

Imagine a meat grinder in a factory. You put raw beef in the top, turn the handle, and ground beef comes out the bottom. If you put chicken in, ground chicken comes out. 

A **Function** is exactly this: a mathematical machine. You feed it a specific input, it performs a strict set of operations, and it spits out exactly one output. 
* **The Golden Rule of Functions:** For every input you put in, the machine can only give you **exactly one output**. If you put beef into the grinder and it spits out *both* ground beef and a live chicken, the machine is broken.

#### **2. Rigorous Theory: $f(x)$, Domain, and Range**
We must upgrade our notation. Instead of writing $y = 2x + 3$, we write:
$$f(x) = 2x + 3$$
This is read as *"f of x"*. 
* $f$ is the name of the machine. 
* $x$ is the placeholder for the raw material you are feeding it.
* The right side is the machinery itself. 
If we feed the machine a $5$, we write $f(5) = 2(5) + 3 = 13$.

**Domain (The Allowed Inputs):**
The Domain is the complete set of all numbers you are legally allowed to feed into the machine without breaking it. 
* Example: $f(x) = \frac{1}{x}$. You can put almost any number in here. But if you input $x = 0$, you get $\frac{1}{0}$, which we proved in Phase 0 is mathematically undefined. Therefore, the Domain is "All real numbers, where $x \neq 0$."

**Range (The Possible Outputs):**
The Range is the complete set of all numbers the machine is capable of spitting out.
* Example: $f(x) = x^2$. No matter what real number you put in (positive or negative), squaring it will always yield a positive number (or zero). Therefore, the Range is $y \ge 0$.

#### **3. Composite Functions (The Assembly Line)**
In Machine Learning (specifically Deep Learning), we don't just use one machine. We hook machines up in an assembly line. The output of the first machine becomes the raw material fed directly into the second machine.

If you have Machine $g(x)$ and Machine $f(x)$, putting them in an assembly line is written as **$f(g(x))$**. 
You evaluate the inside machine $g(x)$ first, and pass that result into the outside machine $f(x)$.
