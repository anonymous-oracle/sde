**1. Supermarket Intuition**
In Module 03, we learned that adding fractions with different denominators is tedious, requiring the LCM. In Module 03 Node 3, we solved this by forcing denominators into powers of 10 (Decimals). **Percentages** are simply another rigid normalization strategy: we force the denominator to be exactly $100$. 
The word literally translates from Latin *per centum*—"by the hundred." If a machine learning model has an accuracy of $0.94$, we say $94\%$. It is the universal language of probability and comparative growth.

**2. Bounded Rigorous Derivation**
Let $x$ be a rational number ($\mathbb{Q}$).

* **The Core Definition:** $$x\% = \frac{x}{100}$$

* **Conversion Algorithms:**
  1.  **Fraction to Percentage:** Multiply the fraction by $100$ and attach the $\%$ symbol. (Mathematically, you are multiplying by $\frac{100}{100}$, which is $1$, preserving the identity).
      $$\frac{a}{b} = \left(\frac{a}{b} \times 100\right)\%$$
  2.  **Decimal to Percentage:** Shift the decimal point two places to the right and attach the $\%$ symbol.
      $$0.xyz = \left(0.xyz \times 100\right)\% = xy.z\%$$
  3.  **Percentage to Fraction/Decimal:** Remove the $\%$ symbol and divide by $100$. 
      $$a\% = \frac{a}{100}$$

* **Percentage of a Quantity:**
  To find $x\%$ of a total quantity $Y$, apply the fractional multiplication unlocked in Module 03:
  $$\text{Value} = \frac{x}{100} \times Y$$
