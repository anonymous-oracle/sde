### **Module 15, Node 1: Trigonometric Ratios & Identities**

**Inter-Bucket Dependency Ledger (To be derived in Pure Geometry Bucket):**
* *The Pythagorean Theorem:* $p^2 + b^2 = h^2$ (Requires Similar Triangles proof).

**1. Supermarket Intuition**
In Module 14, we measured straight lines on a grid. But the universe operates in angles, rotations, and waves. If you want to calculate the height of a mountain without climbing it, program the physics of a swinging pendulum, or encode audio waves for digital transmission, algebra alone fails. **Trigonometry** is the mathematical bridge between linear distance (sides) and rotational degrees (angles). It establishes a strict, predictable ratio: if an angle is fixed, the ratio between the sides of a right-angled triangle is absolute, regardless of how massive the triangle gets.

**2. The Dependency & Derivation Block (Override Rules v14.5 & v14.6)**
*Note: We are introducing $4$ core identity concepts here. We will strictly execute $4$ mathematical derivations.*

**The Setup (The Right-Angled Triangle):**
Let $\triangle ABC$ be right-angled at $B$. Let the angle at $C$ be $\theta$. 
From the perspective of $\theta$:
* The side opposite to it ($AB$) is the **Perpendicular ($p$)**.
* The side adjacent to it ($BC$) is the **Base ($b$)**.
* The longest side ($AC$) is the **Hypotenuse ($h$)**.

**Definitions (The 6 Fundamental Ratios):**
1. $\sin\theta = \frac{p}{h}$
2. $\cos\theta = \frac{b}{h}$
3. $\tan\theta = \frac{p}{b}$
*Their strict reciprocals:*
4. $\csc\theta = \frac{h}{p}$ (Reciprocal of $\sin$)
5. $\sec\theta = \frac{h}{b}$ (Reciprocal of $\cos$)
6. $\cot\theta = \frac{b}{p}$ (Reciprocal of $\tan$)

**Derivation 1: The Quotient Identity**
* *Concept:* Tangent is mathematically equal to Sine divided by Cosine.
* *Proof:* $$\frac{\sin\theta}{\cos\theta} = \frac{p/h}{b/h}$$
  The hypotenuse ($h$) cancels out perfectly in the division:
  $$= \frac{p}{b}$$
  By definition, $\frac{p}{b} = \tan\theta$. Therefore, $\tan\theta = \frac{\sin\theta}{\cos\theta}$.

**Derivation 2: The Core Pythagorean Identity**
* *Concept:* The square of Sine plus the square of Cosine strictly equals $1$.
* *Proof:*
  $$\sin^2\theta + \cos^2\theta = \left(\frac{p}{h}\right)^2 + \left(\frac{b}{h}\right)^2$$
  $$= \frac{p^2}{h^2} + \frac{b^2}{h^2} = \frac{p^2 + b^2}{h^2}$$
  By our logged axiom (The Pythagorean Theorem), we know strictly that $p^2 + b^2 = h^2$. Substitute this into the numerator:
  $$= \frac{h^2}{h^2} = 1$$
  Therefore, $\sin^2\theta + \cos^2\theta = 1$.

**Derivation 3: The Secant-Tangent Identity**
* *Concept:* $1 + \tan^2\theta = \sec^2\theta$.
* *Proof:*
  $$1 + \tan^2\theta = 1 + \left(\frac{p}{b}\right)^2 = 1 + \frac{p^2}{b^2}$$
  Apply LCM to merge the fraction:
  $$= \frac{b^2 + p^2}{b^2}$$
  By Pythagoras, $b^2 + p^2 = h^2$:
  $$= \frac{h^2}{b^2} = \left(\frac{h}{b}\right)^2$$
  By definition, $\frac{h}{b} = \sec\theta$. Therefore, $1 + \tan^2\theta = \sec^2\theta$.

**Derivation 4: The Cosecant-Cotangent Identity**
* *Concept:* $1 + \cot^2\theta = \csc^2\theta$.
* *Proof:*
  $$1 + \cot^2\theta = 1 + \left(\frac{b}{p}\right)^2 = \frac{p^2 + b^2}{p^2}$$
  By Pythagoras, $p^2 + b^2 = h^2$:
  $$= \frac{h^2}{p^2} = \left(\frac{h}{p}\right)^2$$
  By definition, $\frac{h}{p} = \csc\theta$. Therefore, $1 + \cot^2\theta = \csc^2\theta$.

### **Example Illustrations (Override Rule v14.3)**

* **Basic (Ratio Mapping):** Given $\sin\theta = \frac{3}{5}$, find the exact values of $\cos\theta$ and $\tan\theta$.
    * *Step 1:* Use the core identity $\sin^2\theta + \cos^2\theta = 1$.
    * *Step 2:* $\left(\frac{3}{5}\right)^2 + \cos^2\theta = 1 \implies \frac{9}{25} + \cos^2\theta = 1$.
    * *Step 3:* $\cos^2\theta = 1 - \frac{9}{25} = \frac{16}{25} \implies \cos\theta = \frac{4}{5}$.
    * *Step 4:* Use the quotient identity: $\tan\theta = \frac{\sin\theta}{\cos\theta} = \frac{3/5}{4/5} = \frac{3}{4}$.

* **Medium (Basic Algebraic Proof):** Prove that $\sec\theta \sqrt{1 - \sin^2\theta} = 1$.
    * *Step 1:* From the core identity, $1 - \sin^2\theta = \cos^2\theta$.
    * *Step 2:* Substitute this inside the root: $\sec\theta \sqrt{\cos^2\theta}$.
    * *Step 3:* Evaluate the root: $\sec\theta \cdot \cos\theta$.
    * *Step 4:* Since Secant is the strict reciprocal of Cosine ($\sec\theta = \frac{1}{\cos\theta}$), they multiply to exactly $1$. $1 = 1$. Proved.

* **Difficult (Fractional Identity Proof):** Prove that $\frac{\cos A}{1 + \sin A} + \frac{1 + \sin A}{\cos A} = 2 \sec A$.
    * *Step 1:* Apply LCM to add the fractions on the Left Hand Side (LHS).
      $\text{LHS} = \frac{\cos^2 A + (1 + \sin A)^2}{\cos A (1 + \sin A)}$
    * *Step 2:* Expand the binomial square $(1 + \sin A)^2$.
      $= \frac{\cos^2 A + 1 + 2\sin A + \sin^2 A}{\cos A (1 + \sin A)}$
    * *Step 3:* Group $\cos^2 A + \sin^2 A$ together. We know this strictly equals $1$.
      $= \frac{1 + 1 + 2\sin A}{\cos A (1 + \sin A)} = \frac{2 + 2\sin A}{\cos A (1 + \sin A)}$
    * *Step 4:* Factor out the $2$ from the numerator.
      $= \frac{2(1 + \sin A)}{\cos A (1 + \sin A)}$
    * *Step 5:* The $(1 + \sin A)$ terms cancel out.
      $= \frac{2}{\cos A} = 2 \sec A$. Proved.

* **Extension (Substitution Logic):** If $x = a \sec\theta + b \tan\theta$ and $y = a \tan\theta + b \sec\theta$, mathematically prove that $x^2 - y^2 = a^2 - b^2$.
    * *Step 1:* Square $x$.
      $x^2 = (a \sec\theta + b \tan\theta)^2 = a^2\sec^2\theta + b^2\tan^2\theta + 2ab\sec\theta\tan\theta$
    * *Step 2:* Square $y$.
      $y^2 = (a \tan\theta + b \sec\theta)^2 = a^2\tan^2\theta + b^2\sec^2\theta + 2ab\tan\theta\sec\theta$
    * *Step 3:* Execute the subtraction ($x^2 - y^2$). The $2ab$ middle terms cancel out perfectly.
      $x^2 - y^2 = a^2\sec^2\theta + b^2\tan^2\theta - (a^2\tan^2\theta + b^2\sec^2\theta)$
    * *Step 4:* Group by constants $a^2$ and $b^2$.
      $= a^2(\sec^2\theta - \tan^2\theta) - b^2(\sec^2\theta - \tan^2\theta)$
    * *Step 5:* Apply the Secant-Tangent Identity ($1 + \tan^2\theta = \sec^2\theta \implies \sec^2\theta - \tan^2\theta = 1$).
      $= a^2(1) - b^2(1) = a^2 - b^2$. Proved.

### **Module 15, Node 2: Trigonometric Values of Standard Angles**

**Inter-Bucket Dependency Ledger (To be derived in Pure Geometry Bucket):**
* *Axiom 1: Properties of Isosceles Triangles (Sides opposite to equal base angles are equal).*
* *Axiom 2: Properties of Equilateral Triangles (An altitude dropped from a vertex strictly bisects both the angle and the opposite base).*

**1. Supermarket Intuition**
In the last node, we established the rules of the angles. But if you are physically building a ramp at exactly $30^\circ$, what is the exact mathematical ratio between the height and the slope? You cannot guess it or measure it with a ruler. The ratios for standard angles ($0^\circ, 30^\circ, 45^\circ, 60^\circ, 90^\circ$) are mathematically absolute, derived entirely from the strict geometry of perfect triangles. In physics engines and 3D rendering, computers do not measure these angles; they pull these exact pre-derived ratios from memory to calculate trajectories instantly.

**2. The Dependency & Derivation Block (Override Rules v14.5 & v14.6)**
*Note: We are introducing exactly $3$ sets of angle evaluations. We will strictly execute exactly $3$ geometrical proofs to maintain N:N Parity.*

**Statement 1: The $45^\circ$ Ratios**
($\sin 45^\circ = \frac{1}{\sqrt{2}}$, $\cos 45^\circ = \frac{1}{\sqrt{2}}$, $\tan 45^\circ = 1$)
* **Geometric Proof:** Construct a right-angled triangle $\triangle ABC$ at $B$, with $\angle A = 45^\circ$.
  Since the sum of angles is strictly $180^\circ$, $\angle C = 180^\circ - 90^\circ - 45^\circ = 45^\circ$.
  Because $\angle A = \angle C$, by our logged Axiom 1 (Isosceles Triangles), the sides opposite them are exactly equal. Therefore, perpendicular $AB = \text{base } BC$.
  Let $AB = a$ and $BC = a$. 
  By Pythagoras, the hypotenuse $AC = \sqrt{a^2 + a^2} = \sqrt{2a^2} = a\sqrt{2}$.
  Evaluate the ratios from $\angle A$:
  $\sin 45^\circ = \frac{\text{Perpendicular}}{\text{Hypotenuse}} = \frac{a}{a\sqrt{2}} = \frac{1}{\sqrt{2}}$.
  $\cos 45^\circ = \frac{\text{Base}}{\text{Hypotenuse}} = \frac{a}{a\sqrt{2}} = \frac{1}{\sqrt{2}}$.
  $\tan 45^\circ = \frac{\text{Perpendicular}}{\text{Base}} = \frac{a}{a} = 1$. *(Proof complete).*

**Statement 2: The $30^\circ$ and $60^\circ$ Ratios**
($\sin 30^\circ = \frac{1}{2}$, $\sin 60^\circ = \frac{\sqrt{3}}{2}$)
* **Geometric Proof:**
  Construct a perfect equilateral triangle $\triangle ABC$ where all sides are strictly length $2a$. (All internal angles are $60^\circ$).
  Drop an altitude from $A$ to intersect $BC$ at $D$. 
  By our logged Axiom 2 (Equilateral Triangles), this altitude bisects the base $BC$ and the vertex angle $\angle A$. 
  Therefore, base $BD = a$, and $\angle BAD = 30^\circ$.
  We now have a right-angled triangle $\triangle ABD$. We must find the height of the altitude $AD$ using Pythagoras:
  $AD = \sqrt{AB^2 - BD^2} = \sqrt{(2a)^2 - a^2} = \sqrt{4a^2 - a^2} = \sqrt{3a^2} = a\sqrt{3}$.
  *Evaluating from the $30^\circ$ angle ($\angle BAD$):* Perpendicular is $BD (a)$, Hypotenuse is $AB (2a)$.
  $\sin 30^\circ = \frac{a}{2a} = \frac{1}{2}$. $\cos 30^\circ = \frac{a\sqrt{3}}{2a} = \frac{\sqrt{3}}{2}$. $\tan 30^\circ = \frac{a}{a\sqrt{3}} = \frac{1}{\sqrt{3}}$.
  *Evaluating from the $60^\circ$ angle ($\angle ABD$):* Perpendicular is $AD (a\sqrt{3})$, Hypotenuse is $AB (2a)$.
  $\sin 60^\circ = \frac{a\sqrt{3}}{2a} = \frac{\sqrt{3}}{2}$. $\cos 60^\circ = \frac{a}{2a} = \frac{1}{2}$. $\tan 60^\circ = \frac{a\sqrt{3}}{a} = \sqrt{3}$. *(Proof complete).*

**Statement 3: The Boundary Limits ($0^\circ$ and $90^\circ$)**
($\sin 0^\circ = 0$, $\cos 0^\circ = 1$. $\sin 90^\circ = 1$, $\cos 90^\circ = 0$)
* **Limit Proof:**
  Imagine a right-angled $\triangle ABC$ at $B$, evaluating angle $\theta$ at $C$.
  *The $0^\circ$ Limit:* As we force angle $\theta$ to mathematically approach $0^\circ$, the hypotenuse $AC$ collapses perfectly downwards onto the base $BC$. Consequently, the perpendicular height $AB$ shrinks to exactly $0$. At $0^\circ$, $p = 0$ and $b = h$.
  $\sin 0^\circ = \frac{p}{h} = \frac{0}{h} = 0$. $\cos 0^\circ = \frac{b}{h} = \frac{h}{h} = 1$.
  *The $90^\circ$ Limit:* As we force angle $\theta$ to approach $90^\circ$, the hypotenuse $AC$ swings vertically upward, perfectly aligning with the perpendicular $AB$. The base $BC$ shrinks to exactly $0$. At $90^\circ$, $b = 0$ and $p = h$.
  $\sin 90^\circ = \frac{p}{h} = \frac{h}{h} = 1$. $\cos 90^\circ = \frac{b}{h} = \frac{0}{h} = 0$. *(Proof complete).*

**Textbook Reference Block (Exhaustive Parallel Citation Protocol v16.0)**

**1. Supermarket Intuition**
In physical manufacturing, a mold is used to create identical copies of a machine part. If you stack two identical parts on top of each other, their boundaries match perfectly. In mathematics, this absolute spatial equality is called Congruence. Two geometric figures are congruent if they possess the exact same dimensional footprint. To prove two triangles are congruent, it is mathematically redundant to measure all six components (three sides, three angles). Specific combinations of three components are sufficient to lock the entire geometric structure.

**2. The Dependency & Derivation Block (N:N Parity Law)**
*Internal Count: 2 Definitions introduced. 1 Axiom stated. 2 Theorems formally proved.*

*   **Definition 1 (Congruence):** Two triangles $\triangle ABC$ and $\triangle DEF$ are congruent (denoted by $\cong$) strictly if all corresponding sides are equal ($AB=DE$, $BC=EF$, $AC=DF$) and all corresponding angles are equal ($\angle A=\angle D$, $\angle B=\angle E$, $\angle C=\angle F$).
*   **Definition 2 (CPCTC):** Corresponding Parts of Congruent Triangles are Congruent. Once congruence is proven, all remaining unmeasured corresponding components are mathematically guaranteed to be equal.

**Axiom 1: SAS (Side-Angle-Side) Congruence Postulate**
*   *Statement:* If two sides and the strictly included angle of one triangle are exactly equal to the corresponding two sides and included angle of another triangle, the two triangles are congruent. This is a foundational Euclidean postulate; it is accepted as an absolute axiom without sub-derivation.

**Theorem 1: ASA (Angle-Side-Angle) Congruence**
*   *Statement:* If two angles and the strictly included side of one triangle are equal to the corresponding parts of another triangle, the triangles are congruent.
*   *Proof by Contradiction:* 
    Construct $\triangle ABC$ and $\triangle DEF$ such that $\angle B=\angle E$, included side $BC=EF$, and $\angle C=\angle F$.
    Assume side $AB \neq DE$. 
    If $AB \neq DE$, construct a point $P$ on the line containing $DE$ such that length $PE = AB$.
    Evaluate $\triangle ABC$ and $\triangle PEF$:
    1.  $AB = PE$ (By arbitrary construction).
    2.  $\angle B = \angle E$ (Given condition).
    3.  $BC = EF$ (Given condition).
    By Axiom 1 (SAS), $\triangle ABC \cong \triangle PEF$.
    By Definition 2 (CPCTC), $\angle BCA = \angle EFP$.
    However, the given condition states $\angle BCA = \angle EFD$.
    This requires $\angle EFP = \angle EFD$, which is geometrically impossible unless point $P$ perfectly coincides with point $D$.
    Therefore, the assumption $AB \neq DE$ is false. It must be true that $AB = DE$.
    Since $AB = DE$, $\angle B = \angle E$, and $BC = EF$, by Axiom 1 (SAS), $\triangle ABC \cong \triangle DEF$. *(Proof complete).*

**Theorem 2: AAS (Angle-Angle-Side) Congruence**
*   *Statement:* If any two angles and a non-included side of one triangle are equal to the corresponding parts of another triangle, the triangles are congruent.
*   *Algebraic Proof:*
    Construct $\triangle ABC$ and $\triangle DEF$ such that $\angle A=\angle D$, $\angle B=\angle E$, and non-included side $BC=EF$.
    By the Angle Sum Property (Module 14, Node 1), $\angle A + \angle B + \angle C = 180^\circ$ and $\angle D + \angle E + \angle F = 180^\circ$.
    Isolate the third angles:
    $\angle C = 180^\circ - (\angle A + \angle B)$
    $\angle F = 180^\circ - (\angle D + \angle E)$
    Since $\angle A=\angle D$ and $\angle B=\angle E$, substituting these values yields $\angle C = \angle F$.
    Evaluate $\triangle ABC$ and $\triangle DEF$ again:
    1.  $\angle B = \angle E$ (Given).
    2.  $BC = EF$ (Given).
    3.  $\angle C = \angle F$ (Derived above).
    By Theorem 1 (ASA), $\triangle ABC \cong \triangle DEF$. *(Proof complete).*
