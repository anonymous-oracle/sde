### **Module 05, Node 1: Set Theory Syntax & Typologies**

**Textbook Reference:** * Selina Concise Mathematics (Class 6, Chapter 10: Sets; Class 7, Set Concepts; Class 8, Sets).

**1. Supermarket Intuition**
In databases, a "table" is a collection of unique records. When you write a SQL query to find "all users over 18 who live in Bengaluru," you are executing a Set Theory operation. A Set is simply a well-defined collection of distinct objects. If the collection is not "well-defined" (e.g., "a collection of *good* students" is subjective), a computer cannot process it. We must establish the rigid syntax used to define, populate, and categorize these data structures.

**2. Bounded Rigorous Derivation**
Let $A$ be a set. 
* **Elements:** The objects contained within a set. If an object $x$ is inside set $A$, we write $x \in A$ ($x$ belongs to $A$). If not, $x \notin A$. Elements in a set must be distinct (no duplicates).

* **Syntax Notation:**
    1.  **Roster (Tabular) Form:** Explicitly listing all elements inside curly braces, separated by commas. 
        *Example: $V = \{a, e, i, o, u\}$*
    2.  **Set-Builder (Rule) Form:** Defining the set by the mathematical rule all its elements must follow. 
        *Example: $V = \{x \mid x \text{ is a vowel in the English alphabet}\}$*

* **Typologies and Cardinality:**
    * **Cardinal Number ($n(A)$):** The total count of distinct elements in a finite set. 
    * **Finite Set:** A set with a countable number of elements (e.g., factors of 12).
    * **Infinite Set:** A set with an uncountable number of elements (e.g., the set of all Integers, $\mathbb{Z}$).
    * **Empty (Null/Void) Set ($\emptyset$ or $\{\}$):** A set containing zero elements. Cardinality is $0$.
    * **Singleton Set:** A set containing exactly one element. Cardinality is $1$.
    * **Equivalent Sets:** Two sets $A$ and $B$ where $n(A) = n(B)$ (they have the same *number* of elements, but not necessarily the same elements).
    * **Equal Sets:** Two sets $A$ and $B$ that contain the exact same elements, regardless of order. $A = B$.

**1. Supermarket Intuition**
If you have a dataset of "Users who bought shoes" and another dataset of "Users who bought socks," defining the data is only step one. The actual business logic happens when you interact with them. You might want to run a marketing campaign targeting users who bought *both* (an Intersection), users who bought *either* (a Union), or users who bought shoes but strictly *not* socks (a Difference). In SQL, these are `INNER JOIN`, `FULL OUTER JOIN`, and `LEFT JOIN EXCLUDING INNER`. In mathematics, this is the foundational architecture of Set Operations.

**2. Bounded Rigorous Derivation**
Let $A$ and $B$ be sets.

* **The Universal Set ($U$ or $\xi$):** The overarching parent set containing all possible elements under consideration for a specific problem.
* **Subsets ($\subset$ and $\subseteq$):** * $A$ is a subset of $B$ ($A \subseteq B$) if every single element of $A$ is also an element of $B$.
    * $A$ is a *proper* subset of $B$ ($A \subset B$) if $A \subseteq B$ but $A \neq B$ (meaning $B$ has at least one element that $A$ does not).
* **Union ($A \cup B$):** The set of all elements that belong to $A$, or to $B$, or to both. 
    $$A \cup B = \{x \mid x \in A \text{ or } x \in B\}$$
* **Intersection ($A \cap B$):** The set of elements that belong to *both* $A$ and $B$ simultaneously. 
    $$A \cap B = \{x \mid x \in A \text{ and } x \in B\}$$
    *Note: If $A \cap B = \emptyset$, the sets are called **Disjoint Sets**.*
* **Difference ($A - B$):** The set of elements that belong strictly to $A$ and NOT to $B$.
    $$A - B = \{x \mid x \in A \text{ and } x \notin B\}$$
* **Complement ($A'$ or $A^c$):** The set of elements in the Universal set $U$ that are strictly not in $A$.
    $$A' = U - A = \{x \mid x \in U \text{ and } x \notin A\}$$
