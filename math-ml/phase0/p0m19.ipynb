### Lexical Geometry Postulates (D-Nodes)

#### D-Node 35: Locus & Equidistance

* **Locus (Plural: Loci):** The path traced by a point moving in a two-dimensional plane under specific, unyielding mathematical constraints or rules. It defines the exact set of all points—and only those points—that satisfy the specified conditions.
* **Equidistant:** Maintaining a completely identical, unchanging geometric distance value relative to two or more structural references (points, lines, or boundaries).

### Rigorous Derivations (Bottom-Up Traversal)

#### Theorem 46: The Perpendicular Bisector Locus

The locus of a point equidistant from two fixed distinct points is the perpendicular bisector of the straight line segment joining the two fixed points.

**Proof:**
Let $A$ and $B$ be two fixed reference points. Let a point $P$ move in the plane such that its distance to $A$ always matches its distance to $B$ ($PA = PB$). Draw a line connecting $A$ and $B$, and mark the exact midpoint $M$ of segment $AB$ ($AM = MB$). Connect $P$ to $M$.

1. Isolate and compare the triangles $\triangle PMA$ and $\triangle PMB$:
* $PA = PB$ (Given mathematical constraint).
* $AM = MB$ (By midpoint definition).
* $PM = PM$ (Shared common side).


2. By SSS Congruence (Theorem 8), the triangles are identical:

$$\triangle PMA \cong \triangle PMB$$


3. Evaluating matching components via CPCTC:

$$\angle PMA = \angle PMB$$


4. By the Linear Pair Postulate, these adjacent angles along the straight path $AB$ must sum to **180°**:

$$\angle PMA + \angle PMB = \mathbf{180^\circ} \implies 2\angle PMA = \mathbf{180^\circ} \implies \angle PMA = \mathbf{90^\circ}$$


5. Because $\angle PMA = \mathbf{90^\circ}$ and $M$ is the midpoint, the line path passing through $P$ and $M$ is strictly perpendicular to $AB$ and slices it in half. Thus, any point satisfying $PA = PB$ must lie on this perpendicular bisector.

*Status: [UNLOCKED]*

#### Theorem 47: The Angle Bisector Locus

The locus of a point equidistant from two intersecting straight lines is the pair of lines that bisect the angles formed between the two original lines.

**Proof:**
Let lines $AB$ and $CD$ intersect at an origin point $O$. Let a point $P$ move in the space between them such that its perpendicular distance to line $AB$ (segment $PM \perp AB$) identically matches its perpendicular distance to line $CD$ (segment $PN \perp CD$), meaning $PM = PN$. Connect $P$ to the intersection vertex $O$.

1. Isolate and evaluate the right-angled triangles $\triangle PMO$ and $\triangle PNO$:
* Hypotenuse $OP = \text{Hypotenuse } OP$ (Shared common side).
* Side $PM = \text{Side } PN$ (Given equidistance constraint).
* $\angle PMO = \angle PNO = \mathbf{90^\circ}$ (By perpendicular distance definition).


2. By RHS Congruence (Theorem 9), the triangles match perfectly:

$$\triangle PMO \cong \triangle PNO$$


3. Evaluating the tracking links via CPCTC locks the vertex angle:

$$\angle POM = \angle PON$$


4. Because $\angle POM = \angle PON$, the vector path $OP$ splits the internal angle evenly. Thus, any point maintaining equal distance to the lines must lie on this angle bisector.

*Status: [UNLOCKED]*

#### The Fundamental Construction Algorithms

By combining Theorem 46 and Theorem 47, we automate the construction of circle intersections for any target triangle $\triangle ABC$:

1. **Circumcircle Construction (The Out-Circle):** * Construct the perpendicular bisectors for side $AB$ and side $BC$ using a compass.
* By Theorem 46, their intersection point $O$ is simultaneously equidistant from $A, B,$ and $C$ ($OA = OB = OC$).
* Point $O$ acts as the *Circumcenter*. Placing the compass point at $O$ with a radius of $OA$ draws a circle that passes perfectly through all three vertices.


2. **Incircle Construction (The In-Circle):**
* Construct the angle bisectors for internal angles $\angle B$ and $\angle C$.
* By Theorem 47, their intersection point $I$ is simultaneously equidistant from all three boundary lines of the triangle.
* Point $I$ acts as the *Incenter*. Dropping a perpendicular from $I$ to base $BC$ establishes the radius length required to draw a circle that touches all three sides internally without crossing them.

### Intuition-Building Illustrative Examples (IBIE)

#### IBIE 1: The Multi-Constraint Spatial Intersection Trap

* **[The Problem]:** Two fixed coordinates $A$ and $B$ sit **6 cm** apart. Find the exact number of possible locations in the plane for a point $P$ that is simultaneously equidistant from $A$ and $B$, AND located exactly **5 cm** away from point $A$.
* **[The Heuristic Trap]:** Assuming that because two independent rules exist, they cannot cross, or trying to guess individual point locations without drawing the full path curves.
* **[The Failure State]:** Stating that only one point exists at the midpoint or that infinite locations exist along a single line.
* **[The Resolution Heuristic]:** Decompose the problem into separate paths. Rule 1 ($PA = PB$) defines a continuous line (the perpendicular bisector). Rule 2 ($PA = \mathbf{5\text{ cm}}$) defines a continuous circle centered at $A$ with a radius of **5 cm**. The true solutions are the intersection points of these two paths.
* **[Solution Pathway]:**
1. Draw the perpendicular bisector of $AB$. It passes through the midpoint $M$ (which is **3 cm** from $A$).
2. Draw a circle of radius **5 cm** centered at point $A$.
3. Because the circle's radius (**5 cm**) is strictly greater than the distance to the bisector line (**3 cm**), the curved circle boundary must cut across the straight bisector line at exactly two distinct points.
4. Therefore, exactly two valid locations exist for point $P$.



#### IBIE 2: The Angular Blindspot Vertex Trap

* **[The Problem]:** In $\triangle ABC$, a point $X$ must be located such that it is equidistant from side $AB$ and side $AC$, AND simultaneously equidistant from vertex $B$ and vertex $C$. Define the precise structural construction step to find $X$.
* **[The Heuristic Trap]:** Conflating the locus for equal distance to *sides* (lines) with the locus for equal distance to *vertices* (points).
* **[The Failure State]:** Constructing two separate angle bisectors, or intersecting two perpendicular bisectors.
* **[The Resolution Heuristic]:** Translate the vocabulary tokens to their matching path equations:
* Equidistant from lines $AB$ and $AC \rightarrow$ Requires the **Angle Bisector** of $\angle A$.
* Equidistant from points $B$ and $C \rightarrow$ Requires the **Perpendicular Bisector** of side $BC$.


* **[Solution Pathway]:** Construct the internal angle bisector of $\angle A$ and the perpendicular bisector of segment $BC$. The intersection point of these two distinct paths marks the exact location of point $X$.
