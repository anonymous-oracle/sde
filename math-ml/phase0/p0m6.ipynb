**1. Supermarket Intuition**
In Module 03, a fraction $\frac{3}{4}$ simply meant 3 parts of a whole. A **Ratio** is a fraction with *context*. It is the comparative relationship between two quantities of the *exact same kind*. If you have 3 apples and 4 oranges, the ratio of apples to oranges is $3:4$. 
A **Proportion** occurs when you equate two ratios. If you scale up to 6 apples and 8 oranges, the ratio $6:8$ is proportionally equal to $3:4$. In machine learning, scaling image pixel dimensions or normalizing loss gradients relies entirely on maintaining strict proportional relationships so the data doesn't distort.

**2. Bounded Rigorous Derivation**
* **The Ratio ($a:b$):** Mathematically identical to the fraction $\frac{a}{b}$, where $a$ is the *antecedent* and $b$ is the *consequent*. 
    * *The Unit Mandate:* You cannot form a ratio between 2 kilograms and 500 grams. Both must be converted to the same unit before the ratio is established. Once established, the ratio itself has no units.
* **Proportion ($a:b :: c:d$):** An equation stating that two ratios are equivalent: $\frac{a}{b} = \frac{c}{d}$.
    * Here, $a$ and $d$ are called the **Extremes** (the outer terms), and $b$ and $c$ are called the **Means** (the inner terms).
* **The Cross-Product Theorem:** For any proportion $\frac{a}{b} = \frac{c}{d}$ (where $b, d \neq 0$), the product of the extremes equals the product of the means.
    $$ad = bc$$
    *Proof:* Multiply both sides of the equation $\frac{a}{b} = \frac{c}{d}$ by the LCM of the denominators, which is $bd$. The $b$ cancels on the left, leaving $ad$. The $d$ cancels on the right, leaving $bc$. (This builds directly on our Module 03 LCM mechanics).
* **Continued Proportion:** Three quantities $a, b, c$ are in continued proportion if the ratio of the first to the second equals the ratio of the second to the third: $\frac{a}{b} = \frac{b}{c}$.
    * By the Cross-Product Theorem, this implies $b^2 = ac$. Here, $b$ is known as the **mean proportional**.

### **Module 06, Node 2: Direct and Inverse Variation**

**1. Supermarket Intuition**
Proportions do not exist statically; they describe how systems move. In data science, you are constantly looking for *correlations*. If I increase my server compute power, my processing time drops. If I increase my marketing budget, my user acquisition goes up. These are dynamic relationships.
* **Direct Variation:** Two variables move in the exact same direction. (More fuel = more distance).
* **Inverse Variation:** Two variables move in strictly opposite directions. (More speed = less time to destination).

**2. Bounded Rigorous Derivation**
Let $x$ and $y$ be two variable quantities. Let $k$ be a positive, non-zero constant ($k > 0$).

* **Direct Variation ($y \propto x$):**
  When $x$ increases, $y$ increases proportionally. Their *ratio* remains strictly constant.
  $$\frac{y}{x} = k \implies y = kx$$
  Therefore, for any two states of the system:
  $$\frac{y_1}{x_1} = \frac{y_2}{x_2}$$
  *(Notice this is exactly the Proportion equation we unlocked in Node 1).*

* **Inverse Variation ($y \propto \frac{1}{x}$):**
  When $x$ increases, $y$ decreases proportionally. Their *product* remains strictly constant.
  $$y = k \times \frac{1}{x} \implies xy = k$$
  Therefore, for any two states of the system:
  $$x_1 \times y_1 = x_2 \times y_2$$

* **Problem 2 (Basic):** It takes $8$ workers exactly $15 \text{ days}$ to build a wall. If all workers operate at identical efficiency, how many days will it take $12$ workers to build the exact same wall? (Identify the variation type first).
* **Problem 3 (Medium):** The time $t$ taken to fill a tank varies inversely as the number of identical pumps $p$ used. If $4$ pumps can fill the tank in $45 \text{ minutes}$, how many pumps are needed to fill it in $20 \text{ minutes}$?

### **Module 06, Node 3: Advanced Proportion Properties (Componendo & Dividendo)**

**Textbook Reference Block (Exhaustive Parallel Citation Protocol v14.2):**
* **Selina Concise Mathematics:** Class 10, Chapter 7: Ratio and Proportion -> Page Range: Standard Edition Span.
* **M.L. Aggarwal Understanding ICSE:** Class 10, Chapter 7: Ratio and Proportion -> Page Range: Standard Edition Span.
* *(Note: S. Chand, Frank, and others place this strictly in their Class 10 volumes under identical titles).*

**1. Supermarket Intuition**
In simple algebra, if you have $\frac{x+5}{x-5} = \frac{7}{2}$, you would cross-multiply, expand the terms, group the variables, and solve. That works, but it is computationally "expensive" in terms of human time. In advanced calculus or algorithmic design, you often encounter massive symmetric fractions. "Componendo and Dividendo" (C&D) is a structural shortcut. It allows you to rapidly strip away the addition and subtraction in a fraction by exploiting the inherent symmetries of proportions, turning complex polynomials back into simple ratios.

**2. Bounded Rigorous Derivation**
Let $\frac{a}{b} = \frac{c}{d}$ be our foundational proportion (where $b, d \neq 0$).

We derive five specific algebraic transformations. These are universally true for any valid proportion:
1.  **Invertendo:** Flip both sides.
    $$\frac{b}{a} = \frac{d}{c}$$
2.  **Alternando:** Swap the "means" (the cross-diagonals). 
    $$\frac{a}{c} = \frac{b}{d}$$
3.  **Componendo:** Add the denominator to the numerator. (We proved this in Node 1 by adding 1 to both sides).
    $$\frac{a+b}{b} = \frac{c+d}{d}$$
4.  **Dividendo:** Subtract the denominator from the numerator. (Proved by subtracting 1 from both sides).
    $$\frac{a-b}{b} = \frac{c-d}{d}$$
5.  **Componendo and Dividendo (C&D):** The ultimate tool. We divide the Componendo equation by the Dividendo equation. The denominators $b$ and $d$ cancel out.
    $$\frac{a+b}{a-b} = \frac{c+d}{c-d}$$
    *The Power of C&D:* If you apply C&D to an expression that already looks like $\frac{x+y}{x-y}$, it perfectly strips away the symmetry and reverts it back to $\frac{x}{y}$. It is its own inverse operation.
