### Rigorous Derivations (Bottom-Up Traversal)

#### Theorem 38: The Central Angle Theorem

The angle subtended by a circular arc at the center of a circle is exactly twice the angle subtended by the same arc at any point on the remaining part of the circumference.

**Proof:**
Let a circle have center $O$. Let arc $AB$ subtend angle $\angle AOB$ at the center, and angle $\angle ACB$ at a point $C$ on the remaining circumference. Draw a straight line from $C$ passing through the center $O$, extending to a point $D$ outside the triangle.
Isolate $\triangle AOC$:

* $OA = OC$ (Radii of the identical circle).
* By the Isosceles Triangle Theorem (Theorem 7), base angles match: $\angle OCA = \angle OAC$.

Examine the exterior line segment boundary at point $O$. By the Exterior Angle Theorem (Theorem 5), the exterior angle $\angle AOD$ equals the sum of the two opposite interior angles:


$$\angle AOD = \angle OCA + \angle OAC$$


Substituting the equal base angle values:


$$\angle AOD = 2\angle OCA \quad \text{— (Equation 1)}$$

Now isolate the adjacent triangle $\triangle BOC$:

* $OB = OC$ (Radii of the identical circle).
* By Theorem 7: $\angle OCB = \angle OBC$.

Applying Theorem 5 to the exterior configuration at point $O$:


$$\angle BOD = \angle OCB + \angle OBC \implies \angle BOD = 2\angle OCB \quad \text{— (Equation 2)}$$

Summing Equation 1 and Equation 2 completely:


$$\angle AOD + \angle BOD = 2\angle OCA + 2\angle OCB$$

$$\angle AOB = 2(\angle OCA + \angle OCB) \implies \angle AOB = 2\angle ACB$$

*(Note: If arc $AB$ is a semi-circle, the central angle $\angle AOB$ straightens to exactly $180^\circ$, forcing the circumferential angle to stay locked at half that value, which proves Theorem 40 automatically).*

*Status: [UNLOCKED]*

#### Theorem 39: Angles in the Same Segment are Equal

Connect the endpoints to the center $O$ to form the central reference angle $\angle AOB$. Applying Theorem 38 to both circumferential points independently:


$$\angle AOB = 2\angle ACB \implies \angle ACB = \frac{1}{2}\angle AOB$$

$$\angle AOB = 2\angle ADB \implies \angle ADB = \frac{1}{2}\angle AOB$$


Equating the two derived fractional scalar blocks:


$$\angle ACB = \angle ADB$$

#### Theorem 41: Opposite Angles of a Cyclic Quadrilateral are Supplementary

The sum of any pair of opposite interior angles of a cyclic quadrilateral is exactly equal to $180^\circ$.

**Proof:**
Let $ABCD$ be a cyclic quadrilateral inscribed inside a circumcircle with center $O$. Connect the opposite vertices $B$ and $D$ directly to the center $O$ using radii segments $OB$ and $OD$. Let the interior angle at vertex $A$ be denoted as $\angle BAD = x$, and the interior angle at vertex $C$ be denoted as $\angle BCD = y$.

1. Examine the central angle subtended by the minor arc $BCD$. This angle faces vertex $A$. By the Central Angle Theorem (Theorem 38), the interior central angle measures exactly twice the circumferential angle:



$$\angle BOD_{\text{interior}} = 2x$$


2. Examine the central angle subtended by the major arc $BAD$. This angle faces vertex $C$. By Theorem 38, the reflex central angle measures exactly twice its matching circumferential angle:



$$\angle BOD_{\text{reflex}} = 2y$$


3. The sum of the interior central angle and the reflex central angle completes one full point rotation around the center $O$, which is axiomatically locked at exactly $360^\circ$:

$$\angle BOD_{\text{interior}} + \angle BOD_{\text{reflex}} = 360^\circ$$



Substituting the derived angular components into the rotation equation:


$$2x + 2y = 360^\circ$$

Factoring out the scalar value of $2$:


$$2(x + y) = 360^\circ \implies x + y = 180^\circ$$

$$\therefore \angle BAD + \angle BCD = 180^\circ$$

By executing an identical derivation sequence after connecting vertices $A$ and $C$ to the center $O$, it is mathematically proven that $\angle ABC + \angle ADC = 180^\circ$.

*Status: [UNLOCKED]*

#### Theorem 42: Exterior Angle of a Cyclic Quadrilateral

If one side of a cyclic quadrilateral is produced, the resulting exterior angle is strictly equal to the interior opposite angle.

**Proof:**
Let side $AB$ of cyclic quadrilateral $ABCD$ be extended past vertex $B$ to an external point $E$, creating the exterior angle $\angle CBE$.

1. By the Linear Pair Postulate, the adjacent interior angle and the exterior angle along the straight line path $AE$ must sum to $180^\circ$:

$$\angle ABC + \angle CBE = 180^\circ \quad \text{— (Equation 1)}$$


2. By Theorem 41, the opposite interior angles of this inscribed cyclic polygon are supplementary:

$$\angle ABC + \angle ADC = 180^\circ \quad \text{— (Equation 2)}$$


3. Equate Equation 1 and Equation 2 since both share a common sum of $180^\circ$:

$$\angle ABC + \angle CBE = \angle ABC + \angle ADC$$



Subtracting the common structural component $\angle ABC$ from both sides of the equation locks the invariant:


$$\angle CBE = \angle ADC$$

#### Theorem 43: The Tangent-Radius Orthogonality Theorem

A tangent at any point on a circle is strictly perpendicular ($\perp$) to the radius drawn through the point of contact.

**Proof by Contradiction:**
Let a circle have center $O$, and let line $XY$ be a tangent touching the circle at the unique point of contact $P$. We must prove that $OP \perp XY$.

1. Assume that $OP$ is *not* perpendicular to $XY$.
2. If $OP$ is not perpendicular, construct an alternative line segment from the center $O$ that drops strictly perpendicular to the line $XY$. Let this perpendicular line meet $XY$ at a distinct point $Q$. Thus, $\angle OQP = 90^\circ$.
3. Look at the right-angled triangle $\triangle OQP$. Because $\angle OQP = 90^\circ$, by the Triangle Angle Sum property (Theorem 4), the remaining interior angles must be strictly acute ($<90^\circ$). Therefore, $\angle OQP > \angle OPQ$.
4. By the Angle-Side Inequality (Theorem 10), the side opposite the larger angle must be strictly longer than the side opposite the smaller angle. The side opposite $\angle OQP$ is $OP$, and the side opposite $\angle OPQ$ is $OQ$:

$$OP > OQ \implies OQ < OP$$


5. Since $P$ lies perfectly on the circle's circumference, $OP = r$ (the radius). Because $OQ < OP$, the distance $OQ$ is strictly less than the radius ($OQ < r$).
6. This length constraint mathematically forces the point $Q$ to lie *inside* the circle's internal surface space.
7. If point $Q$ lies inside the circle, the straight line path $XY$ passing through $Q$ must intersect the circle's boundary at two distinct points, converting line $XY$ into a *secant* (D-Node 34).
8. This directly contradicts the initial given structural state that line $XY$ is a *tangent*.
9. The alternative assumption is invalid. Therefore, no point other than $P$ can be closer to $O$, forcing $OP$ to be the absolute minimum distance path. Minimum distance maps to the perpendicular vector.

$$\therefore OP \perp XY$$

If two distinct tangent lines are drawn to a circle from a singular external point, the lengths of the tangent segments from the external point to the points of contact are strictly equal.

$$\angle OAP = \angle OBP = 90^\circ$$


2. Isolate and evaluate the right-angled triangles $\triangle OAP$ and $\triangle OBP$:
* Hypotenuse $OP = \text{Hypotenuse } OP$ (Shared common side).
* Side $OA = \text{Side } OB$ (Radii of the identical circle, D-Node 19).
* $\angle OAP = \angle OBP = 90^\circ$ (Perpendicularity invariant).


3. By RHS Congruence (Theorem 9), the triangles are structurally identical:

$$\triangle OAP \cong \triangle OBP$$


4. Evaluating matching outer links via CPCTC (D-Node 7) locks the segment metrics:

$$PA = PB$$



$$\therefore \text{The tangent segments are equal.}$$

#### Theorem 45: The Alternate Segment Theorem

The angle formed between a tangent and a chord through the point of contact is strictly equal to the angle subtended by the chord in the alternate segment.

**Proof:**
Let line $XY$ be a tangent to a circle at the point of contact $P$. Let $PA$ be a chord drawn through $P$, forming the angle $\angle APX$ with the tangent. Let $C$ be any point in the alternate major segment, forming the circumferential angle $\angle PCA$. We must prove that $\angle APX = \angle PCA$.

1. **Construction:** Draw the diameter line from $P$ passing through the center $O$, meeting the circumference at point $B$. Connect point $B$ to point $A$ with a chord segment.
2. By Theorem 43 (Tangent-Radius Orthogonality), the diameter meeting the tangent line forces a complete right-angle lock:

$$\angle BPX = 90^\circ$$


3. The full angle $\angle BPX$ is composed of two adjacent sub-angles: $\angle BPA + \angle APX$. Therefore:

$$\angle BPA + \angle APX = 90^\circ \implies \angle APX = 90^\circ - \angle BPA \quad \text{— (Equation 1)}$$


4. Now analyze the internal right triangle $\triangle BPA$. By Thales' Semi-Circle Theorem (Theorem 40), the angle subtended by the diameter path $BP$ at vertex $A$ is exactly:

$$\angle BAP = 90^\circ$$


5. Applying the Triangle Angle Sum property (Theorem 4) to $\triangle BPA$:

$$\angle ABP + \angle BPA + 90^\circ = 180^\circ \implies \angle ABP + \angle BPA = 90^\circ \implies \angle ABP = 90^\circ - \angle BPA \quad \text{— (Equation 2)}$$


6. Equating Equation 1 and Equation 2 yields:

$$\angle APX = \angle ABP$$


7. Observe vertices $B$ and $C$. Both angles $\angle ABP$ and $\angle PCA$ are subtended by the identical base chord arc $PA$. By Theorem 39 (Angles in the Same Segment are Equal):

$$\angle ABP = \angle PCA$$


8. Substituting this structural equivalence identity into our equation locks the proof:

$$\angle APX = \angle PCA$$

#### IBIE 1: The Adjacent Angle Misalignment Trap

* **[The Problem]:** A tangent line $XY$ touches a circle at $P$. A chord $PA$ forms an angle $\angle APX = 55^\circ$. A point $C$ lies on the minor arc side of the chord. Is $\angle PCA = 55^\circ$?
* **[The Heuristic Trap]:** Grabbing the nearest available circumferential vertex without verifying if it resides within the true *alternate* structural segment.
* **[The Failure State]:** Stating $\angle PCA = 55^\circ$, forgetting that the minor segment angle must face the obtuse reflection boundary.
* **[The Resolution Heuristic]:** The Alternate Segment Theorem maps an acute angle to the *opposite major segment* vertex. If a vertex lies on the same side as the minor segment, it forms a cyclic quadrilateral with the alternate vertex, making it supplementary.
* **[Solution Pathway]:**
1. Find the alternate major segment angle via Theorem 45: $\angle P_{\text{major}}A = 55^\circ$.
2. Because the major vertex and minor vertex $C$ form a cyclic quadrilateral inscribed in the circle, apply Theorem 41:

$$\angle PCA + 55^\circ = 180^\circ \implies \angle PCA = 125^\circ$$
