#### **1. Proof by Contradiction: $\sqrt{2}$ is Irrational (Module 02)**
*The fundamental proof that numbers exist beyond fractions.*
* **Assume the opposite:** Assume $\sqrt{2}$ is rational. Therefore, it can be written as an irreducible fraction $\frac{p}{q}$, where $p$ and $q$ are integers with *no common factors* other than 1, and $q \neq 0$.
* $$\sqrt{2} = \frac{p}{q}$$
* Square both sides: $$2 = \frac{p^2}{q^2} \implies 2q^2 = p^2$$
* **Logic step:** Since $p^2$ is exactly 2 times an integer ($q^2$), $p^2$ must be an even number. If a square is even, its square root must also be even. Therefore, $p$ is even.
* Let $p = 2k$ (where $k$ is an integer). Substitute this back into the equation:
* $$2q^2 = (2k)^2 \implies 2q^2 = 4k^2 \implies q^2 = 2k^2$$
* **Logic step:** By the exact same logic, $q^2$ is even, which means $q$ is even.
* **The Contradiction:** If both $p$ and $q$ are even, they share a common factor of 2. This directly contradicts our foundational assumption that $\frac{p}{q}$ was irreducible. The original assumption must be false. $\sqrt{2}$ is irrational.

#### **3. Completing the Square: Deriving the Quadratic Formula (Module 12)**
*The algorithmic engine of non-linear mathematics.*
* Start with the standard form: $$ax^2 + bx + c = 0$$
* Divide the entire equation by $a$ (to make the coefficient of $x^2$ strictly 1):
    $$x^2 + \frac{b}{a}x + \frac{c}{a} = 0$$
* Transpose the constant term:
    $$x^2 + \frac{b}{a}x = -\frac{c}{a}$$
* **The Core Trick (Completing the Square):** We want the LHS to look like $(x + k)^2 = x^2 + 2kx + k^2$. To achieve this, we take half of the $x$ coefficient ($\frac{b}{2a}$), square it, and add it strictly to *both* sides of the equation.
    $$x^2 + \frac{b}{a}x + \left(\frac{b}{2a}\right)^2 = -\frac{c}{a} + \left(\frac{b}{2a}\right)^2$$
* The LHS is now a perfect square. Evaluate the RHS:
    $$\left(x + \frac{b}{2a}\right)^2 = -\frac{c}{a} + \frac{b^2}{4a^2}$$
    $$\left(x + \frac{b}{2a}\right)^2 = \frac{-4ac + b^2}{4a^2}$$
* Take the square root of both sides (introducing the $\pm$ bound):
    $$x + \frac{b}{2a} = \pm\frac{\sqrt{b^2 - 4ac}}{2a}$$
* Isolate $x$:
    $$x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$$

### **CORRECTIVE TRACK: FORMAL DERIVATIONS AND PROOFS**

**Derivation 1: The Distributive Law (Module 01)**
*   *Statement:* $a(b + c) = ab + ac$
*   *Geometric Proof:* Construct a rectangle $R$ with length $a$ and width $(b + c)$. By definition, the total area of $R$ is strictly $a \times (b + c)$. 
    Draw a line perpendicular to the width, splitting the width into two separate segments of length $b$ and $c$. This creates two smaller adjacent rectangles. 
    The area of the first rectangle is $a \times b$. The area of the second is $a \times c$. 
    Because the total area must equal the sum of its parts, $a(b + c) = ab + ac$.

**Derivation 2: Density of Rational Numbers (Module 02)**
*   *Statement:* Between any two rational numbers, there exist infinite rational numbers.
*   *Algebraic Proof:* Let $a$ and $b$ be rational numbers ($\mathbb{Q}$) such that $a < b$. 
    Calculate the arithmetic mean: $c = \frac{a+b}{2}$. Because $\mathbb{Q}$ is strictly closed under addition and division by non-zero integers, $c \in \mathbb{Q}$.
    Rewrite the inequality: $\frac{2a}{2} < \frac{a+b}{2} < \frac{2b}{2}$.
    This simplifies to $a < c < b$. A new rational number $c$ strictly exists between $a$ and $b$.
    This operation $\frac{x+y}{2}$ can be applied recursively to $a$ and $c$, and $c$ and $b$, ad infinitum. Therefore, infinite rationals exist between $a$ and $b$.

**Derivation 3: Cross-Multiplication Equivalence (Module 03)**
*   *Statement:* If $\frac{a}{b} = \frac{c}{d}$, then $ad = bc$.
*   *Algebraic Proof:* Given the equation $\frac{a}{b} = \frac{c}{d}$, where $b \neq 0$ and $d \neq 0$.
    Multiply both sides of the equation by the lowest common multiple of the denominators, which is $(b \cdot d)$.
    $\frac{a}{b} (bd) = \frac{c}{d} (bd)$
    The $b$ terms cancel on the Left Hand Side (LHS), and the $d$ terms cancel on the Right Hand Side (RHS).
    $a(d) = c(b) \implies ad = bc$.

**Derivation 4: Divisibility by 9 Algorithm (Module 04)**
*   *Statement:* A number is divisible by 9 strictly if the sum of its digits is divisible by 9.
*   *Algebraic Proof:* Let $N$ be a four-digit base-10 integer. By place-value expansion, $N = 1000a + 100b + 10c + d$.
    Deconstruct the base-10 multipliers into multiples of 9 plus a remainder of 1:
    $N = (999 + 1)a + (99 + 1)b + (9 + 1)c + d$
    Distribute the variables:
    $N = 999a + a + 99b + b + 9c + c + d$
    Group all multiples of 9 together:
    $N = (999a + 99b + 9c) + (a + b + c + d)$
    Factor out 9 from the first group:
    $N = 9(111a + 11b + c) + (a + b + c + d)$
    The first term $9(111a + 11b + c)$ is strictly divisible by 9. Therefore, for the entire integer $N$ to be divisible by 9, the remaining term $(a + b + c + d)$—which is the sum of the digits—must also be divisible by 9.

**Derivation 5: HCF and LCM Fundamental Theorem (Module 04)**
*   *Statement:* $\text{HCF}(A, B) \times \text{LCM}(A, B) = A \times B$
*   *Algebraic Proof:* Let $A$ and $B$ be integers completely decomposed into their prime factors. For simplicity, assume they share prime factors $p_1$ and $p_2$, such that $A = p_1^{x_1} p_2^{x_2}$ and $B = p_1^{y_1} p_2^{y_2}$.
    By definition of Highest Common Factor, we take the minimum exponents: 
    $\text{HCF} = p_1^{\min(x_1, y_1)} \cdot p_2^{\min(x_2, y_2)}$
    By definition of Lowest Common Multiple, we take the maximum exponents: 
    $\text{LCM} = p_1^{\max(x_1, y_1)} \cdot p_2^{\max(x_2, y_2)}$
    Multiply HCF and LCM:
    $\text{HCF} \times \text{LCM} = \left( p_1^{\min(x_1, y_1)} \cdot p_2^{\min(x_2, y_2)} \right) \cdot \left( p_1^{\max(x_1, y_1)} \cdot p_2^{\max(x_2, y_2)} \right)$
    By exponent addition rules:
    $= p_1^{\min(x_1, y_1) + \max(x_1, y_1)} \cdot p_2^{\min(x_2, y_2) + \max(x_2, y_2)}$
    For any two real numbers $x$ and $y$, the sum of their minimum and maximum is strictly equal to their total sum ($\min(x,y) + \max(x,y) = x + y$).
    $= p_1^{x_1 + y_1} \cdot p_2^{x_2 + y_2}$
    This is mathematically identical to multiplying $A \times B$.

**Derivation 6: De Morgan’s First Law (Module 05)**
*   *Statement:* $(A \cup B)' = A' \cap B'$
*   *Set-Theoretic Proof:* We must prove mutual inclusion (LHS $\subseteq$ RHS and RHS $\subseteq$ LHS).
    *Part 1:* Let generic element $x \in (A \cup B)'$.
    By definition of a complement, $x \notin (A \cup B)$.
    If an element is strictly not in the union of two sets, it means it exists in neither set.
    Therefore, $x \notin A$ AND $x \notin B$.
    By definition of a complement, $x \in A'$ AND $x \in B'$.
    By definition of intersection (AND), $x \in (A' \cap B')$. Thus, $(A \cup B)' \subseteq A' \cap B'$.
    *Part 2:* Let generic element $y \in A' \cap B'$.
    By definition of intersection, $y \in A'$ AND $y \in B'$.
    By definition of complement, $y \notin A$ AND $y \notin B$.
    Since it is in neither set, it cannot be in their union: $y \notin (A \cup B)$.
    Therefore, $y \in (A \cup B)'$. Thus, $A' \cap B' \subseteq (A \cup B)'$.
    Since both subsets contain each other, they are strictly equal: $(A \cup B)' = A' \cap B'$.

**Derivation 7: Componendo and Dividendo (Module 06)**
*   *Statement:* If $\frac{a}{b} = \frac{c}{d}$, then $\frac{a+b}{a-b} = \frac{c+d}{c-d}$.
*   *Algebraic Proof:* 
    Start with $\frac{a}{b} = \frac{c}{d}$.
    *Operation 1 (Componendo):* Add $1$ to both sides.
    $\frac{a}{b} + 1 = \frac{c}{d} + 1 \implies \frac{a+b}{b} = \frac{c+d}{d} \quad \text{--- (Equation 1)}$
    *Operation 2 (Dividendo):* Subtract $1$ from both sides of the original equation.
    $\frac{a}{b} - 1 = \frac{c}{d} - 1 \implies \frac{a-b}{b} = \frac{c-d}{d} \quad \text{--- (Equation 2)}$
    Divide Equation 1 by Equation 2:
    $$\frac{\frac{a+b}{b}}{\frac{a-b}{b}} = \frac{\frac{c+d}{d}}{\frac{c-d}{d}}$$
    The $b$ denominators cancel on the LHS, and the $d$ denominators cancel on the RHS.
    $\frac{a+b}{a-b} = \frac{c+d}{c-d}$.

**Derivation 8: Successive Discount Formula (Module 07)**

**Derivation 10: Fundamental Law of Exponents (Module 09)**

    $a^m \cdot a^n = (a \cdot a \dots \text{to } m \text{ terms}) \times (a \cdot a \dots \text{to } n \text{ terms})$
    The total continuous string of $a$ multiplications is strictly $m + n$ terms long.
    Therefore, $a^m \cdot a^n = a^{m+n}$.
