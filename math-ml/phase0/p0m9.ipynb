**1. Supermarket Intuition**
Until now, we have dealt with arithmetic: exact, static quantities ($5, -12, \frac{3}{4}$). But in machine learning, you do not write algorithms for a single specific number. You write algorithms for *any* data point that might be fed into the system. You need placeholders. Algebra introduces **variables** (letters) to represent unknown or changing values, and **constants** to represent fixed values. An algebraic expression is simply a blueprint for a calculation waiting for data to be injected into it. 

**2. Bounded Rigorous Derivation**
Let us establish the syntax of this new language.

* **Terms and Coefficients:**
    * An expression is made up of **terms** separated strictly by $+$ or $-$ signs. (e.g., $3x^2 - 5xy + 7$ has three terms).
    * Within a term, the numerical multiplier is the **coefficient**. (In $-5xy$, the coefficient of $xy$ is $-5$).
* **Like and Unlike Terms:**
    * **Like Terms:** Terms that have the exact same literal (variable) factors, including identical exponents. (e.g., $4x^2y$ and $-9x^2y$).
    * **Unlike Terms:** Terms with different variable combinations. (e.g., $3x$ and $3x^2$).
* **Fundamental Operations:**
    * **Addition/Subtraction:** You can *only* add or subtract Like Terms. You do this by operating strictly on their numerical coefficients. $5x + 3x = (5+3)x = 8x$.
    * **Multiplication/Division:** You can multiply/divide any terms. You multiply the coefficients together, and you apply the **Laws of Exponents** (Module 01) to the variables. $(3x^2) \times (4x^3) = (3 \times 4)(x^{2+3}) = 12x^5$.
* **Substitution:**
    Replacing the variables with specific numerical values (Integers, Rationals) and evaluating the resulting arithmetic expression strictly using the **BODMAS hierarchy** (Module 04).
